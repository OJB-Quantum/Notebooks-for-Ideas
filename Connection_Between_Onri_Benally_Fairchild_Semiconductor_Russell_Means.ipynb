```python
"""This script constructs a directed Graphviz diagram mapping historical connections.
"""

import graphviz
import matplotlib.pyplot as plt
from IPython.display import display, Image

# Establish global notebook environment resolution and font defaults
plt.rcParams['figure.dpi'] = 250
plt.rcParams['font.sans-serif'] = ['DejaVu Sans']
plt.rcParams['font.weight'] = 'normal'

# Control Knobs
RANK_SEPARATION = "0.10"
NODE_SEPARATION = "0.09"
STAGGER_STEP = 2

if __name__ == '__main__':
    dag = graphviz.Digraph(name='Historical_Connections', format='png')

    dag.attr(
        dpi=str(plt.rcParams['figure.dpi']),
        splines='spline',
        bgcolor='white',
        rankdir='TB',
        ranksep=RANK_SEPARATION,
        nodesep=NODE_SEPARATION
    )
    dag.attr(
        'node',
        shape='box',
        style='solid',
        color='black',
        fontcolor='black',
        fontname='DejaVu Sans',
        fontsize='10'
    )
    dag.attr(
        'edge',
        color='black',
        fontcolor='black',
        arrowhead='normal',
        fontname='DejaVu Sans',
        fontsize='9'
    )

    # Define nodes with manual line breaks to eliminate redundant text-wrapping functions
    dag.node('Onri', 'Onri Jay Benally')
    dag.node('Relatives', 'Relatives of Onri\n(Fairchild Employees)')
    dag.node('Fairchild', 'Fairchild Semiconductor\n(Shiprock, New Mexico Plant)')
    dag.node('AIM', 'American Indian Movement\n(AIM)')
    dag.node('Russell', 'Russell Means\n(Activist & Leader)')
    dag.node('Tatanka', 'Tatanka Means\n(Actor & Comedian)')
    dag.node('Tatanka_Mom', "Gloria Means")
    dag.node('Chinle', 'Chinle\n(Navajo Nation)')

    # Define labeled causal and relational edges with staggering applied
    dag.edge('Onri', 'Relatives', label=' Descendant of', minlen=str(1 * STAGGER_STEP))
    dag.edge('Relatives', 'Fairchild', label=' Employed at (1965-1975)', minlen=str(2 * STAGGER_STEP))
    dag.edge('Russell', 'AIM', label=' Led', minlen=str(2 * STAGGER_STEP))
    dag.edge('AIM', 'Fairchild', label=' Armed takeover (1975)', minlen=str(1 * STAGGER_STEP))
    dag.edge('Russell', 'Tatanka', label=' Father of', minlen=str(3 * STAGGER_STEP))

    # Define biographical and geographical edges within the Navajo Nation
    dag.edge('Tatanka_Mom', 'Tatanka', label=' Mother of', minlen=str(1 * STAGGER_STEP))
    dag.edge('Tatanka_Mom', 'Chinle', label=' From', minlen=str(3 * STAGGER_STEP))
    dag.edge('Tatanka', 'Chinle', label=' Grew up in', minlen=str(1 * STAGGER_STEP))
    dag.edge(
        'Onri',
        'Tatanka',
        label=' Met at Red Valley/Cove\n High School (Navajo Nation)',
        dir='none',
        minlen=str(2 * STAGGER_STEP)
    )

    # Render and display in-memory
    binary_png = dag.pipe(format='png')
    display(Image(data=binary_png))
```
