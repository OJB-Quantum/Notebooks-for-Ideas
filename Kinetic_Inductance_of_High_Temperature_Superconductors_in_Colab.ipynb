# Kinetic Inductance in Superconducting Thin Films: High-Temperature and Disordered Materials

High-impedance quantum circuits, such as fluxonium-type qubits and photon detectors, require inductors that provide large values without occupying large physical areas. Obtaining large geometric inductance typically requires increased conductor length or an expanded magnetic field volume, both of which increase the device footprint and introduce unwanted parasitic capacitance. Superconductors provide an alternative via kinetic inductance, an effect rooted in the inertial mass of charge carriers.

### The Physics of Sheet Inductance

In a superconductor, current consists of the collective motion of Cooper pairs. These pairs possess inertia; accelerating them requires energy. This inertial energy manifests as inductance. For thin films where the thickness is much smaller than the penetration depth ($d \ll \lambda$), the property is described as sheet kinetic inductance $L_{k,\square}$, measured in henries per square ($\text{H}/\square$):

$$
L_{k,\square} \approx \frac{\mu_0 \lambda^2}{d}
$$

This microscopic property translates to circuit-level inductance $L_k$ based on the geometry of the conductor. For a strip with length $\ell$ and width $w$, the relation is:

$$
L_k \approx L_{k,\square} \frac{\ell}{w}
$$

Reducing the film thickness or increasing the penetration depth allows designers to achieve high inductance within a compact footprint.

### Microscopic Origins and Carrier Density

The penetration depth $\lambda$ determines the distance a magnetic field penetrates the superconducting bulk. This length scale depends on the effective mass and charge of the carriers, as well as the superfluid density $n_s$:

$$
\lambda^2 = \frac{m^\ast}{\mu_0 n_s (q^\ast)^2}
$$

Lowering the superfluid density increases the penetration depth. Approaching the Superconductor-Insulator Transition (SIT) leads to a drop in $n_s$. The superfluid density of Amorphous Indium Oxide ($\text{InO}_x$) can be strongly suppressed as the SIT is approached [5]. As a disordered, low-$T_c$ material, $\text{InO}_x$ generates a large $\lambda$, which drives $L_{k,\square}$ to values orders of magnitude higher than those found in crystalline superconductors.

### Anisotropy in High-Temperature Superconductors

High-Temperature Superconductors (HTS), specifically cuprates like Yttrium Barium Copper Oxide (YBCO), possess a layered crystal structure. This electronic architecture produces direction-dependent superfluid stiffness and penetration depth. Current flowing within the copper-oxide planes ($\lambda_{ab}$) exhibits a different superfluid response than current flowing perpendicular to those planes ($\lambda_c$).

The sheet inductance for current transport in the $ab$-plane is:

$$
L_{k,\square}^{(ab)} \approx \frac{\mu_0 \lambda_{ab}^2}{d}
$$

Orienting the film or the circuit traces relative to the crystallographic axes allows for the tuning of the inductive response. For example, (103)-oriented YBCO devices can exhibit inductances differing by roughly a factor of 20 between predominantly $ab$ plane and $c$ axis transport [4].

### The Conventional Dirty Limit

In many superconductors, the electronic mean free path $l$ is short compared to the Pippard coherence length $\xi_0$. This condition defines the dirty limit. In this regime, the effective penetration depth $\lambda_{\mathrm{eff}}$ follows an asymptotic scaling relative to the clean London penetration depth $\lambda_L$:

$$
\lambda_{\mathrm{eff}} \propto \lambda_L \sqrt{\frac{\xi_0}{l}}
$$

Within the conventional dirty-limit scaling, decreasing the mean free path increases $\lambda_{\mathrm{eff}}$ and consequently raises the kinetic inductance. Cuprate superconductors require additional care due to their unconventional pairing and anisotropic electronic structure. Measurements of $\text{Bi}_2\text{Sr}_2\text{CaCu}_2\text{O}_{8+\delta}$ indicate considerably shorter quasiparticle mean free paths than in comparatively clean YBCO, illustrating the broad disorder landscape across cuprate superconductors [1]. Excessive disorder may simultaneously suppress the critical temperature $T_c$ or increase microwave loss.

### Connection to Circuit Impedance

The impact of kinetic inductance on a quantum device is quantified by the kinetic inductance fraction $\alpha$. This ratio compares the kinetic component to the total inductance, including the geometric component $L_g$:

$$
\alpha = \frac{L_k}{L_g + L_k}
$$

As $\alpha$ approaches unity, the device behavior is dominated by the material's superfluid inertia. In microwave resonators, this inductance links directly to the imaginary component of the complex conductivity $\sigma_2$ for a thin film in the predominantly inductive, low-loss regime [12]:

$$
L_{k,\square} \approx \frac{1}{\omega d \sigma_2}
$$

### Material Performance Landscape

Transition temperature and inductive response represent separate axes of material performance. A high $T_c$ ensures superconductivity survives at elevated temperatures, but it does not guarantee a high $L_k$. Conversely, materials like $\text{InO}_x$ or granular aluminium provide enormous inductance despite a low $T_c$ [5, 2]. Mapping these two properties reveals a diverse landscape where material selection depends on whether the priority is thermal stability or electromagnetic impedance.

```python
"""Visualization of Kinetic Inductance and Critical Temperature in Superconductors.

This script analyzes the relationship between Critical Temperature (Tc) and
Kinetic Inductance (Lk) using a dataset of ambient, high pressure, and
provisional superconducting records.
"""

from typing import Any
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from adjustText import adjust_text

# ---------------------------------------------------
# Control knobs
# ---------------------------------------------------
DATA_CUTOFF_DATE = "2026-08-26"
FIGURE_SIZE = (14.0, 8.2)
plt.rcParams["figure.dpi"] = 250
plt.rcParams["font.sans-serif"] = ["Tahoma", "DejaVu Sans"]
plt.rcParams["font.weight"] = "normal"
plt.rcParams["axes.titlepad"] = 12.0

# ---------------------------------------------------
# Material data definitions
# ---------------------------------------------------
ambient_records = pd.DataFrame(
    [
        (1911, 4.20, "Hg", "Mercury", "Ambient", 10.0),
        (1954, 18.05, "Nb3Sn", "Niobium tin", "Ambient", 20.0),
        (1974, 23.20, "Nb3Ge", "Niobium germanium", "Ambient", 25.0),
        (1986, 35.00, "Ba La Cu O", "Barium lanthanum copper oxide", "Ambient", 80.0),
        (1987, 93.00, "YBCO", "Yttrium barium copper oxide", "Ambient", 150.0),
        (1988, 108.00, "Bi-2223", "Bismuth strontium calcium copper oxide", "Ambient", 120.0),
        (1988, 125.00, "Tl 2223", "Tl2Ba2Ca2Cu3O10", "Ambient", 110.0),
        (1993, 133.00, "Hg 1223", "HgBa2Ca2Cu3O8+d", "Ambient", 100.0),
        (2001, 39.00, "MgB2", "Magnesium diboride", "Ambient", 40.0),
        (2008, 55.00, "SmFeAs(O,F)", "Samarium iron arsenide oxide fluoride", "Ambient", 65.0),
        (2026, 151.00, "Hg 1223 PQP", "Pressure quenched HgBa2Ca2Cu3O8+d", "Ambient", 90.0),
    ],
    columns=["year", "tc_k", "plot_label", "material", "pressure", "lk_ph"],
)

high_pressure_records = pd.DataFrame(
    [
        (1994, 164.0, "Hg 1223", "HgBa2Ca2Cu3O8+d", "High Pressure", 95.0),
        (2015, 203.0, "H3S", "Sulfur hydride system", "High Pressure", 60.0),
        (2019, 260.0, "LaH10", "Lanthanum superhydride", "High Pressure", 55.0),
        (2021, 243.0, "YH9", "Yttrium nonahydride", "High Pressure", 50.0),
    ],
    columns=["year", "tc_k", "plot_label", "material", "pressure", "lk_ph"],
)

retracted_reports = pd.DataFrame(
    [
        (2020, 288.0, "C S H", "Carbonaceous sulfur hydride", "High Pressure", 40.0),
        (2023, 294.0, "Lu N H", "Nitrogen doped lutetium hydride", "High Pressure", 40.0),
    ],
    columns=["year", "tc_k", "plot_label", "material", "pressure", "lk_ph"],
)

provisional_report = pd.DataFrame(
    [
        (2025, 298.0, "LaSc2H24", "Lanthanum scandium hydride", "High Pressure", 50.0),
    ],
    columns=["year", "tc_k", "plot_label", "material", "pressure", "lk_ph"],
)

indium_oxide = pd.DataFrame(
    [(2010, 3.4, "InOx", "Amorphous Indium Oxide", "Ambient", 5000.0)],
    columns=["year", "tc_k", "plot_label", "material", "pressure", "lk_ph"],
)

# ---------------------------------------------------
# Core functions
# ---------------------------------------------------
def get_combined_dataset() -> pd.DataFrame:
    """Combines all superconducting records into a single DataFrame.

    Returns:
        pd.DataFrame: Merged dataset containing all studied materials.
    """
    return pd.concat(
        [
            ambient_records,
            high_pressure_records,
            retracted_reports,
            provisional_report,
            indium_oxide,
        ],
        ignore_index=True,
    )


def create_point_label(axis: Any, row: Any) -> Any:
    """Generates a text label for a specific data point.

    Args:
        axis: The matplotlib axis to place the text on.
        row: The data row containing plot_label and values.

    Returns:
        Any: A matplotlib Text object.
    """
    label_text = f"{row.plot_label}\n$L_k$: {row.lk_ph} pH"
    return axis.text(
        row.tc_k,
        row.lk_ph,
        label_text,
        ha="center",
        va="center",
        fontsize=8,
        bbox=dict(boxstyle="round,pad=0.3", facecolor="white", alpha=0.7),
    )


def plot_kinetic_inductance_scatter(df: pd.DataFrame) -> None:
    """Creates a scatterplot of Critical Temperature versus Kinetic Inductance.

    Args:
        df (pd.DataFrame): The combined superconductor dataset.
    """
    fig, ax = plt.subplots(figsize=FIGURE_SIZE)

    """
    The scatterplot utilizes a log scale for the Y axis because the kinetic
    inductance of amorphous materials like InOx is orders of magnitude
    higher than crystalline cuprates or hydrides.
    """
    scatter = ax.scatter(
        df["tc_k"],
        df["lk_ph"],
        c=df["lk_ph"],
        cmap="viridis",
        s=100,
        edgecolors="black"
    )

    texts = [create_point_label(ax, row) for row in df.itertuples(index=False)]

    ax.set_yscale("log")
    ax.set_title(
        f"Superconductor Critical Temperature vs. Kinetic Inductance\n"
        f"Data status checked {DATA_CUTOFF_DATE}",
        fontsize=16
    )
    ax.set_xlabel("Critical Temperature $T_c$ (K)", fontsize=12)
    ax.set_ylabel("Kinetic Inductance $L_k$ (pH/sq)", fontsize=12)
    ax.grid(True, which="both", ls="solid", alpha=0.2)

    fig.colorbar(scatter, ax=ax, label="Kinetic Inductance (pH/sq)")

    """
    fig.canvas.draw() is required in notebook environments to ensure the
    renderer computes actual pixel positions of text before adjust_text
    attempts to move them to avoid overlaps.
    """
    fig.canvas.draw()

    adjust_text(
        texts,
        ax=ax,
        arrowprops=dict(
            arrowstyle="->",
            color="gray",
            lw=0.5,
            shrinkA=5,
            shrinkB=5
        )
    )

    fig.subplots_adjust(left=0.1, right=0.9, top=0.9, bottom=0.1)
    plt.show()


def plot_property_heatmap(df: pd.DataFrame) -> None:
    """Creates a correlation heatmap of numerical superconducting properties.

    Args:
        df (pd.DataFrame): The combined superconductor dataset.
    """
    numeric_df = df[["tc_k", "lk_ph", "year"]]
    corr_matrix = numeric_df.corr()

    fig, ax = plt.subplots(figsize=(8, 6))
    sns.heatmap(
        corr_matrix,
        annot=True,
        cmap="RdBu_r",
        center=0,
        square=True,
        linewidths=1,
        ax=ax
    )
    ax.set_title("Correlation Heatmap $T_c$ versus $L_k$", fontsize=15)
    fig.subplots_adjust(left=0.15, right=0.95, top=0.9, bottom=0.15)
    plt.show()


def main() -> None:
    """Main execution function to generate the graphical visualizations."""
    dataset = get_combined_dataset()

    """
    The processing pipeline first renders a high-resolution scatterplot
    to visualize the massive kinetic inductance of amorphous materials,
    followed by a correlation heatmap to quantify property relationships.
    """
    plot_kinetic_inductance_scatter(dataset)
    plot_property_heatmap(dataset)

if __name__ == "__main__":
    main()
```
