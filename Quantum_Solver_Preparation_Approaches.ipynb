```python
!uv pip install graphviz ipython

"""This script constructs a compact, first-layer Graphviz diagram for ansatz-free quantum simulation in-memory.

The diagram restricts the hierarchy to the root node and the two primary paradigm branches:
Schrodingerisation and Wave Matrix Lindbladization. Control parameters at the top level configure
a 250 DPI resolution, DejaVu Sans typography, spline edge routing, and vertical branch staggering.
"""

import graphviz
from IPython.display import display, Image

# Control Knobs
GRAPH_DPI = "250"
NODE_FONT = "DejaVu Sans"
NODE_FONT_WEIGHT = "normal"
EDGE_ROUTING = "spline"
NODE_SHAPE = "box"
BACKGROUND_COLOR = "white"
FOREGROUND_COLOR = "black"
RANK_DIRECTION = "TB"
MAX_CHARACTERS_PER_LINE = 28
RANK_SEPARATION = "0.50"
NODE_SEPARATION = "0.30"
BRANCH_STAGGER_STEP = 1  # Controls the vertical rank offset between the two first-layer branches


def wrap_text(text: str, max_width: int = MAX_CHARACTERS_PER_LINE) -> str:
    """Wraps text to a specified maximum width, hyphenating oversized strings.

    Args:
        text: The original string input.
        max_width: The maximum allowable characters per line.

    Returns:
        A formatted string featuring newline characters at proper intervals.
    """
    lines = []
    chunk_step = max_width - 1
    for paragraph in text.split('\n'):
        current_line = ""
        words = paragraph.split()
        for word in words:
            if len(word) > max_width:
                if current_line:
                    lines.append(current_line)
                    current_line = ""
                num_chunks = (len(word) - 1) // chunk_step
                for idx in range(num_chunks):
                    start = idx * chunk_step
                    lines.append(word[start:start + chunk_step] + "-")
                word = word[num_chunks * chunk_step:]

            if current_line:
                if len(current_line) + 1 + len(word) <= max_width:
                    current_line += " " + word
                else:
                    lines.append(current_line)
                    current_line = word
            else:
                current_line = word
        if current_line:
            lines.append(current_line)
    return '\n'.join(lines)


def build_first_layer_simulation_tree() -> graphviz.Digraph:
    """Constructs the directed Graphviz tree containing only the root and first-layer branches."""
    tree = graphviz.Digraph(name='First_Layer_Quantum_Simulation', format='png')

    tree.attr(
        dpi=GRAPH_DPI,
        splines=EDGE_ROUTING,
        bgcolor=BACKGROUND_COLOR,
        rankdir=RANK_DIRECTION,
        ranksep=RANK_SEPARATION,
        nodesep=NODE_SEPARATION
    )
    tree.attr(
        'node',
        shape=NODE_SHAPE,
        style='solid',
        color=FOREGROUND_COLOR,
        fontcolor=FOREGROUND_COLOR,
        fontname=NODE_FONT,
        fontsize='10'
    )
    tree.attr(
        'edge',
        color=FOREGROUND_COLOR,
        arrowhead='normal',
        fontname=NODE_FONT,
        fontsize='9'
    )

    tree.node(
        'Root',
        wrap_text('General Quantum Simulation Approaches for Preparation Beyond an Ansatz')
    )

    first_layer_branches = [
        (
            'Branch_Schrod',
            '1. Schrodingerisation\n(Continuous-Variable Hamiltonian Embedding)'
        ),
        (
            'Branch_WML',
            '2. Wave Matrix Lindbladization\n(Sample-Based Open-System Simulation)'
        ),
    ]

    for idx, (branch_id, branch_label) in enumerate(first_layer_branches):
        tree.node(branch_id, wrap_text(branch_label))
        tree.edge(
            'Root',
            branch_id,
            minlen=str((idx + 1) * BRANCH_STAGGER_STEP)
        )

    return tree


def display_graph(tree: graphviz.Digraph) -> None:
    """Renders the Graphviz Digraph in memory and displays the PNG without HTML wrappers."""
    binary_png = tree.pipe(format='png')
    display(Image(data=binary_png))


if __name__ == '__main__':
    first_layer_tree = build_first_layer_simulation_tree()
    display_graph(first_layer_tree)
```

```python
!uv pip install graphviz ipython

"""This script constructs an in-memory Graphviz diagram comparing Schrodingerisation, Wave Matrix Lindbladization, and Ehrenfestization.

The diagram maps three ansatz-free quantum simulation and state preparation paradigms across three
vertically staggered, parallel workflow columns that converge on target state extraction. Control
parameters at the top level configure a 250 DPI resolution, DejaVu Sans typography, spline edge
routing, and a compact portrait aspect ratio.
"""

import graphviz
from IPython.display import display, Image

# Control Knobs
GRAPH_DPI = "250"
NODE_FONT = "DejaVu Sans"
NODE_FONT_WEIGHT = "normal"
EDGE_ROUTING = "spline"
NODE_SHAPE = "box"
BACKGROUND_COLOR = "white"
FOREGROUND_COLOR = "black"
RANK_DIRECTION = "TB"
MAX_CHARACTERS_PER_LINE = 28
TARGET_ASPECT_RATIO = "1.45"        # Balances height-to-width ratio near 3:2 portrait proportions
RANK_SEPARATION = "0.42"            # Vertical distance (inches) between consecutive ranks
NODE_SEPARATION = "0.28"            # Horizontal distance (inches) between parallel columns
BRANCH_STAGGER_STEP = 1             # Vertical rank stagger step across primary paradigm headers
ALIGNMENT_EDGE_WEIGHT = "10"        # High weight keeps vertical sequential chains straight
CROSS_COLUMN_EDGE_WEIGHT = "1"      # Lower weight permits smooth branching and convergence splines


def wrap_text(text: str, max_width: int = MAX_CHARACTERS_PER_LINE) -> str:
    """Wraps text to a specified maximum width, hyphenating oversized strings.

    Args:
        text: The original string input.
        max_width: The maximum allowable characters per line.

    Returns:
        A formatted string featuring newline characters at proper intervals.
    """
    lines = []
    chunk_step = max_width - 1
    for paragraph in text.split('\n'):
        current_line = ""
        words = paragraph.split()
        for word in words:
            if len(word) > max_width:
                if current_line:
                    lines.append(current_line)
                    current_line = ""
                num_chunks = (len(word) - 1) // chunk_step
                for idx in range(num_chunks):
                    start = idx * chunk_step
                    lines.append(word[start:start + chunk_step] + "-")
                word = word[num_chunks * chunk_step:]

            if current_line:
                if len(current_line) + 1 + len(word) <= max_width:
                    current_line += " " + word
                else:
                    lines.append(current_line)
                    current_line = word
            else:
                current_line = word
        if current_line:
            lines.append(current_line)
    return '\n'.join(lines)


def build_ansatz_free_simulation_dag() -> graphviz.Digraph:
    """Constructs the directed Graphviz DAG for Schrodingerisation, Wave Matrix Lindbladization, and Ehrenfestization."""
    dag = graphviz.Digraph(name='Ansatz_Free_Quantum_Simulation', format='png')

    dag.attr(
        dpi=GRAPH_DPI,
        splines=EDGE_ROUTING,
        bgcolor=BACKGROUND_COLOR,
        rankdir=RANK_DIRECTION,
        ranksep=RANK_SEPARATION,
        nodesep=NODE_SEPARATION,
        ratio=TARGET_ASPECT_RATIO
    )
    dag.attr(
        'node',
        shape=NODE_SHAPE,
        style='solid',
        color=FOREGROUND_COLOR,
        fontcolor=FOREGROUND_COLOR,
        fontname=NODE_FONT,
        fontsize='10'
    )
    dag.attr(
        'edge',
        color=FOREGROUND_COLOR,
        arrowhead='normal',
        fontname=NODE_FONT,
        fontsize='9'
    )

    dag.node(
        'Root',
        wrap_text('General Quantum Simulation & State Preparation Beyond an Ansatz\n(Deterministic Dilation of Non-Unitary Dynamics)')
    )

    paradigms = [
        (
            'Cat_Schrod',
            '1. Schrodingerisation\n(Continuous-Variable Hamiltonian Embedding)',
            [
                (
                    'S1',
                    'Target Linear ODE / PDE:\ndu(t)/dt = -A u(t), u(0) = u0\nGeneral non-unitary evolution.'
                ),
                (
                    'S2',
                    'Hermitian-Antihermitian Split:\nA = A1 + i A2\nA1 = (A + A^+)/2 >= 0 (Dissipative)\nA2 = (A - A^+)/(2i) (Dispersive).'
                ),
                (
                    'S3',
                    'Warped Phase Transformation:\nIntroduce auxiliary mode p > 0;\nw(t, p) = exp(-p) u(t) converts\ndecay A1 into spatial drift d_p.'
                ),
                (
                    'S4',
                    'Fourier Transform (p -> η):\nMaps -i d_p to wavenumber η;\nyields extended Hamiltonian\nH_ext = η (x) A1 - I_η (x) A2.'
                ),
                (
                    'S5',
                    'Dilated Unitary Evolution:\nSynthesize exp(-i H_ext t) on\nH_p (x) H_sys via block encoding\nor Trotter product formulas.'
                ),
                (
                    'S6',
                    'Inverse Transform & Projection:\nApply QFT^+ (η -> p) and project\nonto p > 0 sector to extract\nnormalized state u(t)/||u(t)||.'
                ),
            ]
        ),
        (
            'Cat_WML',
            '2. Wave Matrix Lindbladization\n(Sample-Based Open-System Simulation)',
            [
                (
                    'W1',
                    'Target GKSL Master Equation:\ndρ/dt = L(ρ) = -i[H, ρ] +\nSum_k (Lk ρ Lk^+ - 1/2 {Lk^+ Lk, ρ})\nMarkovian open-system channel.'
                ),
                (
                    'W2',
                    'Operator Normalization:\nNormalize jump operators via\nHilbert-Schmidt Schatten-2 norm\n||Lk||_2 = sqrt(Tr[Lk^+ Lk]) = 1.'
                ),
                (
                    'W3',
                    'Quantum Program State Encoding:\nEncode jump operators into Choi\nstates |ψk> = sqrt(d)(I (x) Lk)|Φ>\n& H = Sum_j cj σj density states.'
                ),
                (
                    'W4',
                    'Multiplexed Program Copies:\nPrepare n ~ t^2/ε independent\ncopies of program state ω\n(sampling or Trotter schedule).'
                ),
                (
                    'W5',
                    'Repeated Bipartite Interactions:\nApply controlled partial-SWAP\nexp(-i SWAP Δt) & dissipative\nancilla channels over Δt = t/n.'
                ),
                (
                    'W6',
                    'Partial Trace & Channel Output:\nTrace out program registers Tr_env\nto prepare evolved state exp(Lt)(ρ)\nwithin diamond-norm error ε.'
                ),
            ]
        ),
        (
            'Cat_Ehrenfest',
            '3. Ehrenfestization\n(Expectation-Value Commutator Closure)',
            [
                (
                    'E1',
                    'Target Statistical Classical System:\nd<A>/dt = -i<[A, H]> + <dA/dt>\nMap distributions to quantum expectation values.'
                ),
                (
                    'E2',
                    'Semiclassical State Representation:\nEncode statistical initial conditions\ninto compatible quantum matrix transformations.'
                ),
                (
                    'E3',
                    'Commutator Closure:\nEnforce expectation-value closure to yield\na finite, deterministically solvable operator hierarchy.'
                ),
                (
                    'E4',
                    'QPU Hardware Compilation:\nCompile the mapped matrix transformations\ninto discrete unitary sequences tailored for target QPU topology.'
                ),
                (
                    'E5',
                    'Unitary Hardware Execution:\nEvolve the semiclassical state natively on\nthe QPU via the synthesized quantum transformation steps.'
                ),
                (
                    'E6',
                    'Observable Measurement:\nExtract classical statistical properties directly\nby evaluating target operator expectation values.'
                ),
            ]
        ),
    ]

    dag.node(
        'Target_Output',
        wrap_text('Ansatz-Free Target Quantum State Prepared:\nBypasses Barren Plateaus & Non-Convex VQA Optimization')
    )

    for p_idx, (cat_id, cat_title, steps) in enumerate(paradigms):
        dag.node(cat_id, wrap_text(cat_title))
        dag.edge(
            'Root',
            cat_id,
            minlen=str(1 + p_idx * BRANCH_STAGGER_STEP),
            weight=CROSS_COLUMN_EDGE_WEIGHT
        )

        for s_idx, (step_id, step_text) in enumerate(steps):
            dag.node(step_id, wrap_text(step_text))
            if s_idx == 0:
                dag.edge(cat_id, step_id, minlen='1', weight=ALIGNMENT_EDGE_WEIGHT)
            else:
                prev_step_id = steps[s_idx - 1][0]
                dag.edge(prev_step_id, step_id, minlen='1', weight=ALIGNMENT_EDGE_WEIGHT)

        # Connect final step of each column to the shared synthesis node
        dag.edge(
            steps[-1][0],
            'Target_Output',
            minlen='1',
            weight=CROSS_COLUMN_EDGE_WEIGHT
        )

    # Conceptual Cross-Links Highlighting Shared Structural Duality
    dag.edge(
        'S3',
        'W3',
        label=wrap_text('Auxiliary register dilation', 16),
        style='dashed',
        dir='both',
        constraint='false',
        weight=CROSS_COLUMN_EDGE_WEIGHT
    )
    dag.edge(
        'S5',
        'W5',
        label=wrap_text('Unitary interaction primitive', 16),
        style='dashed',
        dir='both',
        constraint='false',
        weight=CROSS_COLUMN_EDGE_WEIGHT
    )
    dag.edge(
        'W5',
        'E5',
        label=wrap_text('QPU hardware mapping', 16),
        style='dashed',
        dir='both',
        constraint='false',
        weight=CROSS_COLUMN_EDGE_WEIGHT
    )

    return dag


def display_graph(tree: graphviz.Digraph) -> None:
    """Renders the Graphviz Digraph in memory and displays the PNG without HTML wrappers."""
    binary_png = tree.pipe(format='png')
    display(Image(data=binary_png))


if __name__ == '__main__':
    simulation_dag = build_ansatz_free_simulation_dag()
    display_graph(simulation_dag)
```
