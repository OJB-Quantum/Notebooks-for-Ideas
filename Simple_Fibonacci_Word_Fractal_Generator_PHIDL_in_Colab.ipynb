```python
"""This script generates and renders a Fibonacci word fractal using PHIDL."""

import gdspy
import matplotlib as mpl
import matplotlib.pyplot as plt
import numpy as np
import phidl
from phidl import quickplot as qp

# Suppress the specific matplotlib logging warning to maintain a clean output
logging.getLogger('matplotlib.axes._base').setLevel(logging.ERROR)

# --- Control Knobs ---
ITERATION_NUMBER: int = 9
SEGMENT_LENGTH: float = 1.0
PATH_WIDTH: float = 0.15
# ---------------------

# Establish the explicit visual styling configurations
mpl.rcParams['figure.dpi'] = 300
mpl.rcParams['font.sans-serif'] = ['Tahoma', 'DejaVu Sans']
mpl.rcParams['font.weight'] = 'normal'

def generate_fibonacci_word(iteration: int) -> str:
    """Generates the Fibonacci word sequence up to a defined iteration boundary."""
    if iteration == 0:
        return "0"
    if iteration == 1:
        return "01"

    word_n_minus_2 = "0"
    word_n_minus_1 = "01"
    word_current = ""

    for _ in range(2, iteration + 1):
        word_current = word_n_minus_1 + word_n_minus_2
        word_n_minus_2 = word_n_minus_1
        word_n_minus_1 = word_current

    return word_n_minus_1

def create_fractal_points(word: str, length: float) -> list[tuple[float, float]]:
    """Calculates the Cartesian coordinates for the corresponding fractal vertices."""
    x, y = 0.0, 0.0
    angle = 90.0
    points = [(x, y)]

    for k, char in enumerate(word, start=1):
        x += length * np.cos(np.radians(angle))
        y += length * np.sin(np.radians(angle))
        points.append((x, y))

        if char == '0':
            if k % 2 == 0:
                angle += 90.0
            else:
                angle -= 90.0

    return points

def main() -> None:
    """Executes the spatial fractal generation and visualizes the PHIDL device."""
    word = generate_fibonacci_word(ITERATION_NUMBER)
    points = create_fractal_points(word, SEGMENT_LENGTH)

    device = phidl.Device("Fibonacci_Fractal")
    fractal_path = gdspy.FlexPath(points, PATH_WIDTH)
    device.add(fractal_path)

    qp(device)
    plt.show()

if __name__ == "__main__":
    main()
```
