# Standing Waves, Electrical Length, and Impedance Matching in Transmission Lines

A transmission line supports forward and reflected electromagnetic waves if its termination produces a finite reflection coefficient. Shortening the physical line reduces the fraction of a standing-wave pattern that can occupy the structure, yet reflection strength arises primarily from impedance mismatch. This separation between **standing-wave amplitude** and **standing-wave spatial occupancy** provides a more precise design rule than requiring the physical line length to stay below one wavelength. A matched transmission line can extend across many guided wavelengths with negligible standing-wave modulation, whereas an electrically short open or short circuit can retain a reflection coefficient magnitude of unity [1,2].

For microwave and superconducting circuits, the relevant wavelength is the **guided wavelength** $\lambda_g$, whose value can differ considerably from the free-space wavelength. Dielectric loading lowers phase velocity, and superconducting kinetic inductance can lower it further [3,4]. Transmission-line length should generally be evaluated using the dimensionless electrical length $L/\lambda_g$ together with the impedance mismatch represented by $\Gamma_L$.

## 2. First Principles and Fundamental Theory

Distributed transmission-line behavior follows from the telegrapher equations, which relate voltage and current to the per-unit-length electromagnetic parameters [1,2]:

$$
\frac{dV}{dx}=-(R'+j\omega L')I
$$

and

$$
\frac{dI}{dx}=-(G'+j\omega C')V.
$$

Combining these equations gives the complex propagation constant

$$
\gamma=\alpha+j\beta=\sqrt{(R'+j\omega L')(G'+j\omega C')}.
$$

For an ideal lossless line with $R'=0$ and $G'=0$, the propagation constant reduces to

$$
\gamma=j\beta,
$$

with

$$
\beta=\omega\sqrt{L'C'}.
$$

The corresponding phase velocity is

$$
v_p=\frac{\omega}{\beta}=\frac{1}{\sqrt{L'C'}}.
$$

Guided wavelength follows from

$$
\lambda_g=\frac{2\pi}{\beta}=\frac{v_p}{f}.
$$

For a conventional quasi-TEM transmission line whose conductor inductance can be treated geometrically,

$$
v_p\approx\frac{c}{\sqrt{\epsilon_{\mathrm{eff}}}}.
$$

Thus,

$$
\lambda_g\approx\frac{c}{f\sqrt{\epsilon_{\mathrm{eff}}}}.
$$

A termination with impedance $Z_L$ produces the load reflection coefficient [1]

$$
\Gamma_L=\frac{Z_L-Z_0}{Z_L+Z_0}.
$$

Perfect impedance matching occurs at

$$
Z_L=Z_0,
$$

which gives

$$
\Gamma_L=0.
$$

Taking $x=0$ at the load, the total voltage phasor on a lossless line can be written as

$$
V(x)=V^+\left(e^{-j\beta x}+\Gamma_L e^{j\beta x}\right).
$$

Writing the reflection coefficient in polar form as

$$
\Gamma_L=|\Gamma_L|e^{j\phi_\Gamma},
$$

gives the normalized voltage magnitude

$$
\frac{|V(x)|}{|V^+|}=\sqrt{1+|\Gamma_L|^2+2|\Gamma_L|\cos(2\beta x+\phi_\Gamma)}.
$$

This equation separates two physical controls. The magnitude $|\Gamma_L|$ governs the modulation depth, whereas $\beta x$ governs the spatial phase of that modulation.

For a passive lossless line with $|\Gamma_L|<1$, the voltage standing-wave ratio is

$$
\mathrm{VSWR}=\frac{1+|\Gamma_L|}{1-|\Gamma_L|}.
$$

A matched termination gives $\mathrm{VSWR}=1$. Open and short terminations have $|\Gamma_L|=1$, producing complete voltage-wave reflection in the ideal lossless model.

The spatial modulation contains the factor $2\beta x$. Adjacent voltage maxima occur after the phase advances by $2\pi$:

$$
2\beta\Delta x=2\pi.
$$

Using $\beta=2\pi/\lambda_g$ gives

$$
\Delta x_{\mathrm{max}\rightarrow\mathrm{max}}=\frac{\lambda_g}{2}.
$$

A maximum and its neighboring minimum are separated by

$$
\Delta x_{\mathrm{max}\rightarrow\mathrm{min}}=\frac{\lambda_g}{4}.
$$

Hence the standing-wave **voltage envelope repeats every $\lambda_g/2$**, though the forward and reflected traveling waves individually possess wavelength $\lambda_g$ [1].

## 3. Operational Scaling and Implementation

Electrical length provides a compact measure of how much phase accumulates across a physical line:

$$
\theta=\beta L=\frac{2\pi L}{\lambda_g}.
$$

A line with $L=\lambda_g$ possesses

$$
\theta=2\pi.
$$

The standing-wave envelope repeats every $\lambda_g/2$, meaning the number of envelope periods represented by a line of length $L$ is

$$
N_{\mathrm{env}}=\frac{L}{\lambda_g/2}=\frac{2L}{\lambda_g}.
$$

Equivalently,

$$
N_{\mathrm{env}}=\frac{\theta}{\pi}.
$$

This relation provides the precise version of the original intuition. Shortening a line reduces the number of standing-wave oscillations that can spatially develop across its physical extent, yet it leaves the load reflection coefficient unchanged.

| Condition               | Electrical interpretation                  | Standing-wave implication                                                 |
| :---------------------- | :----------------------------------------- | :------------------------------------------------------------------------ |
| $\Gamma_L\approx0$      | Load approximately matched to $Z_0$        | Very weak standing-wave modulation regardless of $L$                      |
| $L<\lambda_g/2$         | Less than $180^\circ$ of electrical length | Less than one complete envelope period fits on the line                   |
| $L=\lambda_g/2$         | $180^\circ$ electrical length              | One complete envelope period fits                                         |
| $L=\lambda_g$           | $360^\circ$ electrical length              | Two complete envelope periods fit                                         |
| $L\lesssim\lambda_g/10$ | Approximately $36^\circ$ or less           | Common first screening criterion for electrically short interconnects [2] |
| $L\gg\lambda_g$         | Distributed-wave regime                    | Many maxima and minima can appear if a finite reflection exists           |

The $\lambda_g/10$ criterion concerns the applicability of a lumped approximation. It does not suppress the underlying reflection coefficient. For example, an open circuit with $L=0.05\lambda_g$ still has $|\Gamma_L|=1$ at the termination, although the physical line samples only a short section of the resulting spatial pattern.

Impedance matching directly addresses amplitude to minimize standing waves:

$$
Z_L\rightarrow Z_0
\quad\Longrightarrow\quad
|\Gamma_L|\rightarrow0
\quad\Longrightarrow\quad
\mathrm{VSWR}\rightarrow1.
$$

Reducing the number of spatial oscillations on a fixed physical line requires the following scaling:

$$
\frac{L}{\lambda_g}=\frac{Lf}{v_p}.
$$

Lower frequency, shorter physical length, or greater phase velocity decreases electrical length.

## 4. Advanced Phenomena and Nuances

Real transmission lines possess finite attenuation. A reflected wave traveling from the load back toward the source experiences additional attenuation, so the local reflection-coefficient magnitude for a uniform lossy line scales as

$$
|\Gamma(x)|=|\Gamma_L|e^{-2\alpha x}.
$$

The factor of two arises from comparing waves whose propagation histories differ by a round-trip distance. Standing-wave modulation may diminish with increasing distance from the load. Lossy transmission lines require additional care in interpreting reflection coefficients and reflected power [5].

Dispersion introduces another refinement. If $v_p$ varies with frequency, the guided wavelength follows

$$
\lambda_g(f)=\frac{v_p(f)}{f}.
$$

Electrical length then loses exact proportionality to frequency. Broadband circuits should evaluate $L/\lambda_g(f)$ across the relevant spectrum, particularly toward frequencies where $\lambda_g$ reaches its smallest values.

Superconducting coplanar waveguides introduce an additional inductive contribution associated with the inertia of Cooper pairs. Their distributed inductance can be written as

$$
L'=L_g'+L_k'.
$$

Experimental and analytical studies of superconducting CPWs show that $L_k'$ can contribute measurably to resonant frequency and propagation characteristics [3,4]. The kinetic-inductance fraction is defined as

$$
\alpha_k=\frac{L_k'}{L'},
$$

which implies

$$
L_g'=(1-\alpha_k)L'.
$$

If the geometric quasi-TEM velocity is approximated by $c/\sqrt{\epsilon_{\mathrm{eff}}}$, the kinetic-inductance correction gives approximately

$$
v_p\approx\frac{c}{\sqrt{\epsilon_{\mathrm{eff}}}}\sqrt{1-\alpha_k}.
$$

Accordingly,

$$
\lambda_g\approx
\frac{c}{f\sqrt{\epsilon_{\mathrm{eff}}}}
\sqrt{1-\alpha_k}.
$$

Greater kinetic-inductance fraction shortens the guided wavelength and increases electrical length for a fixed physical geometry. This effect gains importance in thin superconducting films and high-kinetic-inductance microwave structures [3,4].

## 5. Practical Application and Observation

Consider an illustrative quasi-TEM transmission line operating at

$$
f=6\,\mathrm{GHz},
$$

with

$$
\epsilon_{\mathrm{eff}}=6
$$

and negligible kinetic-inductance correction. Its phase velocity is approximately

$$
v_p\approx\frac{c}{\sqrt{6}}\approx1.224\times10^8\,\mathrm{m\,s^{-1}}.
$$

The corresponding guided wavelength is

$$
\lambda_g\approx20.4\,\mathrm{mm}.
$$

Different physical lengths then produce the following electrical scales.

|              $L$ | $L/\lambda_g$ | Electrical length $\theta$ | $N_{\mathrm{env}}=2L/\lambda_g$ |
| ---------------: | ------------: | -------------------------: | ------------------------------: |
|  $2,\mathrm{mm}$ |       $0.098$ |               $35.3^\circ$ |                         $0.196$ |
|  $5,\mathrm{mm}$ |       $0.245$ |               $88.2^\circ$ |                         $0.490$ |
| $10,\mathrm{mm}$ |       $0.490$ |              $176.5^\circ$ |                         $0.980$ |
| $20,\mathrm{mm}$ |       $0.980$ |             $352.97^\circ$ |                         $1.961$ |

The $2\,\mathrm{mm}$ line samples only a small part of the standing-wave envelope at $6\,\mathrm{GHz}$. A $20\,\mathrm{mm}$ line spans almost one guided wavelength and can display approximately two complete envelope oscillations if its termination produces appreciable reflection.

Measurement with a vector network analyzer connects directly to these quantities. The measured $S_{11}$ referenced to the line impedance represents the complex input reflection coefficient, whose magnitude can be converted to return loss or VSWR. Spatial field simulation provides a complementary view by revealing the corresponding maxima and minima along the physical structure.

## 6. Conceptual Synthesis

```text
Standing-wave behavior
│
├── Reflection strength
│   └── |ΓL|
│       ├── ZL = Z0  →  |ΓL| = 0  →  essentially no standing-wave ripple
│       └── ZL ≠ Z0  →  |ΓL| > 0  →  finite spatial modulation
│
├── Spatial repetition
│   └── β = 2π / λg
│       └── envelope period = λg / 2
│
├── Physical line occupancy
│   └── L / λg
│       └── Nenv = 2L / λg
│
├── Frequency
│   └── f ↑  →  λg ↓  →  L / λg ↑
│
├── Dielectric loading
│   └── εeff ↑  →  vp ↓  →  λg ↓
│
└── Superconducting kinetic inductance
    └── Lk′ ↑  →  vp ↓  →  λg ↓  →  electrical length ↑
```

Impedance matching offers the direct control mechanism to suppress standing-wave amplitude. Shortening the line toward $L\lesssim\lambda_g/10$ provides a common engineering starting point to reduce spatial phase variation and recover approximately lumped behavior. A condition such as $L\leq\lambda_g$ still permits as many as two complete standing-wave envelope periods and represents a distributed transmission line in many microwave applications.

````python
"""Visualize standing waves, mismatch, and electrical length.

The model treats a uniform lossless quasi-TEM transmission line. It separates
reflection strength, which is governed by the load reflection coefficient,
from spatial standing-wave density, which is governed by electrical length.
"""

import math
import matplotlib as mpl
import matplotlib.pyplot as plt
import numpy as np


# -----------------------------------------------------------------------------
# Control knobs
# -----------------------------------------------------------------------------

FREQUENCY_GHZ = 6.0
LINE_LENGTH_MM = 20.0

CHARACTERISTIC_IMPEDANCE_OHM = 50.0
LOAD_IMPEDANCE_OHM = 100.0 + 0.0j

EFFECTIVE_RELATIVE_PERMITTIVITY = 6.0

# alpha_k = L_k' / (L_g' + L_k').
# Set to 0.0 for an ordinary geometric-inductance approximation.
KINETIC_INDUCTANCE_FRACTION = 0.0

LINE_LENGTHS_IN_WAVELENGTHS = (
    0.10,
    0.25,
    0.50,
    1.00,
    2.00,
)

REFLECTION_MAGNITUDES = (
    0.00,
    0.20,
    0.50,
    0.90,
)

FREQUENCY_SWEEP_MIN_GHZ = 1.0
FREQUENCY_SWEEP_MAX_GHZ = 12.0
FREQUENCY_SWEEP_POINTS = 600

SPATIAL_POINTS = 2500
HEATMAP_SPATIAL_POINTS = 1400

FIGURE_DPI = 250
FIGURE_SIZE = (9.5, 5.7)

SHOW_GRID = True


# -----------------------------------------------------------------------------
# Physical constants
# -----------------------------------------------------------------------------

C_0_M_S = 299_792_458.0


# -----------------------------------------------------------------------------
# Plot configuration
# -----------------------------------------------------------------------------

mpl.rcParams.update(
    {
        "figure.dpi": FIGURE_DPI,
        "savefig.dpi": FIGURE_DPI,
        "font.size": 11,
        "axes.titlesize": 13,
        "axes.labelsize": 11,
        "legend.fontsize": 9,
    }
)


# -----------------------------------------------------------------------------
# Transmission-line model
# -----------------------------------------------------------------------------

def validate_controls() -> None:
    """Check the user-adjustable physical parameters."""
    if FREQUENCY_GHZ <= 0.0:
        raise ValueError("FREQUENCY_GHZ must exceed zero.")

    if LINE_LENGTH_MM <= 0.0:
        raise ValueError("LINE_LENGTH_MM must exceed zero.")

    if CHARACTERISTIC_IMPEDANCE_OHM <= 0.0:
        raise ValueError(
            "CHARACTERISTIC_IMPEDANCE_OHM must exceed zero."
        )

    if EFFECTIVE_RELATIVE_PERMITTIVITY <= 0.0:
        raise ValueError(
            "EFFECTIVE_RELATIVE_PERMITTIVITY must exceed zero."
        )

    if not 0.0 <= KINETIC_INDUCTANCE_FRACTION < 1.0:
        raise ValueError(
            "KINETIC_INDUCTANCE_FRACTION must lie in [0, 1)."
        )

    if FREQUENCY_SWEEP_MIN_GHZ <= 0.0:
        raise ValueError(
            "FREQUENCY_SWEEP_MIN_GHZ must exceed zero."
        )

    if FREQUENCY_SWEEP_MAX_GHZ <= FREQUENCY_SWEEP_MIN_GHZ:
        raise ValueError(
            "FREQUENCY_SWEEP_MAX_GHZ must exceed the minimum."
        )


def phase_velocity_m_s(
    epsilon_eff: float,
    kinetic_fraction: float,
) -> float:
    """Return an approximate quasi-TEM phase velocity.

    The geometric phase velocity is c / sqrt(epsilon_eff). A kinetic
    inductance fraction alpha_k modifies it by sqrt(1 - alpha_k).
    """
    geometric_velocity = C_0_M_S / math.sqrt(epsilon_eff)

    return geometric_velocity * math.sqrt(1.0 - kinetic_fraction)


def reflection_coefficient(
    load_impedance_ohm: complex,
    characteristic_impedance_ohm: float,
) -> complex:
    """Return the voltage reflection coefficient at the load."""
    if np.isinf(abs(load_impedance_ohm)):
        return 1.0 + 0.0j

    denominator = (
        load_impedance_ohm + characteristic_impedance_ohm
    )

    if abs(denominator) < np.finfo(float).eps:
        raise ValueError(
            "The selected load makes the reflection coefficient singular."
        )

    return (
        load_impedance_ohm - characteristic_impedance_ohm
    ) / denominator


def standing_wave_magnitude(
    position_m: np.ndarray,
    beta_rad_m: float | np.ndarray,
    gamma_load: complex,
) -> np.ndarray:
    """Return voltage magnitude normalized to the incident amplitude.

    Distance is measured from the load toward the source. The function
    evaluates |exp(-j beta x) + Gamma exp(+j beta x)|.
    """
    phase = np.asarray(beta_rad_m) * np.asarray(position_m)

    forward = np.exp(-1j * phase)
    reflected = gamma_load * np.exp(1j * phase)

    return np.abs(forward + reflected)


def compute_vswr(gamma_magnitude: float) -> float:
    """Return VSWR for a passive lossless-line reflection magnitude."""
    if np.isclose(gamma_magnitude, 1.0):
        return math.inf

    if gamma_magnitude < 1.0:
        return (
            1.0 + gamma_magnitude
        ) / (
            1.0 - gamma_magnitude
        )

    return math.nan


# -----------------------------------------------------------------------------
# Derived quantities
# -----------------------------------------------------------------------------

validate_controls()

frequency_hz = FREQUENCY_GHZ * 1.0e9
line_length_m = LINE_LENGTH_MM * 1.0e-3

phase_velocity = phase_velocity_m_s(
    EFFECTIVE_RELATIVE_PERMITTIVITY,
    KINETIC_INDUCTANCE_FRACTION,
)

guided_wavelength_m = phase_velocity / frequency_hz
beta = 2.0 * np.pi / guided_wavelength_m

gamma_load = reflection_coefficient(
    LOAD_IMPEDANCE_OHM,
    CHARACTERISTIC_IMPEDANCE_OHM,
)

gamma_magnitude = abs(gamma_load)
gamma_phase_deg = np.rad2deg(np.angle(gamma_load))

electrical_length_ratio = (
    line_length_m / guided_wavelength_m
)

electrical_length_deg = (
    360.0 * electrical_length_ratio
)

envelope_periods = 2.0 * electrical_length_ratio
vswr = compute_vswr(gamma_magnitude)

reflected_power_fraction = gamma_magnitude**2


# -----------------------------------------------------------------------------
# Numerical summary
# -----------------------------------------------------------------------------

print("Transmission-line standing-wave summary")
print("=" * 46)
print(f"Frequency:                {FREQUENCY_GHZ:.4f} GHz")
print(f"Physical line length:     {LINE_LENGTH_MM:.4f} mm")
print(
    "Phase velocity:           "
    f"{phase_velocity:,.3f} m/s "
    f"({phase_velocity / C_0_M_S:.4f} c)"
)
print(
    "Guided wavelength:        "
    f"{guided_wavelength_m * 1.0e3:.4f} mm"
)
print(
    "Electrical length L/lambda_g: "
    f"{electrical_length_ratio:.6f}"
)
print(
    "Electrical phase length:  "
    f"{electrical_length_deg:.3f} deg"
)
print(
    "Envelope periods in L:    "
    f"{envelope_periods:.6f}"
)
print(
    "Load reflection Gamma_L:  "
    f"{gamma_load.real:+.6f}"
    f"{gamma_load.imag:+.6f}j"
)
print(
    "|Gamma_L|:                "
    f"{gamma_magnitude:.6f}"
)
print(
    "Reflection phase:         "
    f"{gamma_phase_deg:.3f} deg"
)

if math.isinf(vswr):
    print("VSWR:                     infinity")
elif math.isnan(vswr):
    print("VSWR:                     outside passive-line formula")
else:
    print(f"VSWR:                     {vswr:.6f}")

print(
    "Reflected power fraction: "
    f"{100.0 * reflected_power_fraction:.3f}% "
    "(ideal real-Z0 line)"
)


# -----------------------------------------------------------------------------
# Figure 1 - Reflection strength at fixed electrical length
# -----------------------------------------------------------------------------

x_m = np.linspace(
    0.0,
    line_length_m,
    SPATIAL_POINTS,
)

reference_phase = np.angle(gamma_load)

plt.figure(figsize=FIGURE_SIZE)

for magnitude in REFLECTION_MAGNITUDES:
    gamma_test = magnitude * np.exp(1j * reference_phase)

    voltage_magnitude = standing_wave_magnitude(
        x_m,
        beta,
        gamma_test,
    )

    plt.plot(
        x_m * 1.0e3,
        voltage_magnitude,
        linewidth=1.8,
        label=rf"$|\Gamma_L|={magnitude:.2f}$",
    )

plt.xlabel("Distance from load (mm)")
plt.ylabel(r"$|V(x)|/|V^+|$")
plt.title(
    "Standing-Wave Strength Is Controlled by Reflection"
)

if SHOW_GRID:
    plt.grid(alpha=0.25)

plt.legend()
plt.tight_layout()
plt.show()


# -----------------------------------------------------------------------------
# Figure 2 - Physical occupancy versus L / lambda_g
# -----------------------------------------------------------------------------

normalized_position = np.linspace(
    0.0,
    1.0,
    SPATIAL_POINTS,
)

plt.figure(figsize=FIGURE_SIZE)

for length_in_lambda in LINE_LENGTHS_IN_WAVELENGTHS:
    comparison_length_m = (
        length_in_lambda * guided_wavelength_m
    )

    comparison_x_m = (
        normalized_position * comparison_length_m
    )

    voltage_magnitude = standing_wave_magnitude(
        comparison_x_m,
        beta,
        gamma_load,
    )

    number_of_envelope_periods = 2.0 * length_in_lambda

    label = (
        rf"$L={length_in_lambda:.2f}\lambda_g$"
        rf", $N_{{env}}={number_of_envelope_periods:.2f}$"
    )

    plt.plot(
        normalized_position,
        voltage_magnitude,
        linewidth=1.7,
        label=label,
    )

plt.xlabel(r"Normalized position $x/L$")
plt.ylabel(r"$|V(x)|/|V^+|$")
plt.title(
    "Shorter Electrical Length Contains Fewer "
    "Standing-Wave Periods"
)

if SHOW_GRID:
    plt.grid(alpha=0.25)

plt.legend()
plt.tight_layout()
plt.show()


# -----------------------------------------------------------------------------
# Figure 3 - Frequency-position heatmap for a fixed physical line
# -----------------------------------------------------------------------------

frequency_sweep_ghz = np.linspace(
    FREQUENCY_SWEEP_MIN_GHZ,
    FREQUENCY_SWEEP_MAX_GHZ,
    FREQUENCY_SWEEP_POINTS,
)

frequency_sweep_hz = frequency_sweep_ghz * 1.0e9

heatmap_x_m = np.linspace(
    0.0,
    line_length_m,
    HEATMAP_SPATIAL_POINTS,
)

beta_sweep = (
    2.0
    * np.pi
    * frequency_sweep_hz
    / phase_velocity
)

phase_matrix = (
    beta_sweep[:, np.newaxis]
    * heatmap_x_m[np.newaxis, :]
)

forward_matrix = np.exp(-1j * phase_matrix)
reflected_matrix = gamma_load * np.exp(1j * phase_matrix)

voltage_heatmap = np.abs(
    forward_matrix + reflected_matrix
)

plt.figure(figsize=FIGURE_SIZE)

image = plt.imshow(
    voltage_heatmap,
    origin="lower",
    aspect="auto",
    extent=(
        0.0,
        LINE_LENGTH_MM,
        FREQUENCY_SWEEP_MIN_GHZ,
        FREQUENCY_SWEEP_MAX_GHZ,
    ),
)

plt.xlabel("Distance from load (mm)")
plt.ylabel("Frequency (GHz)")
plt.title(
    "Standing-Wave Spatial Density Versus Frequency"
)

colorbar = plt.colorbar(image)
colorbar.set_label(r"$|V(x)|/|V^+|$")

plt.tight_layout()
plt.show()


# -----------------------------------------------------------------------------
# Figure 4 - Envelope-period count versus electrical length
# -----------------------------------------------------------------------------

length_ratio_axis = np.linspace(
    0.0,
    3.0,
    1200,
)

envelope_count_axis = 2.0 * length_ratio_axis

plt.figure(figsize=FIGURE_SIZE)

plt.plot(
    length_ratio_axis,
    envelope_count_axis,
    linewidth=2.0,
)

plt.scatter(
    [electrical_length_ratio],
    [envelope_periods],
    s=55,
    label=(
        rf"Selected line: "
        rf"$L/\lambda_g={electrical_length_ratio:.3f}$"
    ),
)

plt.xlabel(r"Electrical length $L/\lambda_g$")
plt.ylabel(r"Envelope periods $N_{\mathrm{env}}$")
plt.title(
    r"$N_{\mathrm{env}}=2L/\lambda_g$"
)

if SHOW_GRID:
    plt.grid(alpha=0.25)

plt.legend()
plt.tight_layout()
plt.show()
````
