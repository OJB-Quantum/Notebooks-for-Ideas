```python
"""Build and render a hierarchical 128 KiB ASAP7 SRAM array.

The zoom selector rejects tap cells, dummy cells, fillers, decaps, and other helper structures before identifying a genuine SRAM bitcell or an explicitly labeled supporting-cell fallback.
"""

# -----------------------------------------------------------------------------
# Notebook code cell 1
# -----------------------------------------------------------------------------
# Install only the geometry packages.
#
# NumPy and Matplotlib are intentionally left at the versions supplied by
# the active Colab runtime. PHIDL 1.7.2 supports NumPy 2.x.

import subprocess
import sys

GEOMETRY_REQUIREMENTS = (
    "gdstk==1.0.0",
    "gdspy==1.6.13",
    "phidl==1.7.2",
)

subprocess.run(
    [sys.executable, "-m", "pip", "install", "--quiet", "uv"],
    check=True,
)
subprocess.run(
    [
        "uv",
        "pip",
        "install",
        "--python",
        sys.executable,
        "--upgrade",
        "--no-deps",
        *GEOMETRY_REQUIREMENTS,
    ],
    check=True,
)

probe_code = '''
import gdspy
import gdstk
import matplotlib
import numpy
import phidl
from phidl import Device

print("numpy", numpy.__version__)
print("matplotlib", matplotlib.__version__)
print("gdstk", gdstk.__version__)
print("gdspy", gdspy.__version__)
print("phidl", getattr(phidl, "__version__", "unknown"))
print("PHIDL Device", Device.__name__)
'''

probe = subprocess.run(
    [sys.executable, "-c", probe_code],
    check=True,
    capture_output=True,
    text=True,
)
print(probe.stdout)

# -----------------------------------------------------------------------------
# Notebook code cell 2
# -----------------------------------------------------------------------------
# -------------------------------------------------------------------------
# CAPACITY AND FLOORPLAN
# -------------------------------------------------------------------------
TARGET_BYTES = 128 * 1024
BANK_MACRO_NAME = "srambank_256x4x64_6t122"
BANK_WORD_ROWS = 256
BANK_COLUMN_MUX = 4
BANK_WORD_BITS = 64

BANK_ARRAY_ROWS = 4
BANK_ARRAY_COLUMNS = 4
INTERBANK_GAP_UM = 8.0
BANK_MARGIN_UM = 2.0
PERIPHERY_ROWS = 8

# -------------------------------------------------------------------------
# SOURCE ASSETS
# -------------------------------------------------------------------------
DOWNLOAD_DIRECTORY = "/content/asap7_sources"
OUTPUT_DIRECTORY = "/content/asap7_sram_128kb_output"

SRAM_GDS_URL = (
    "https://raw.githubusercontent.com/The-OpenROAD-Project/"
    "asap7_sram_0p0/main/gds/srambank_64b.gds"
)
SRAM_MACRO_LEF_URL = (
    "https://raw.githubusercontent.com/The-OpenROAD-Project/"
    "asap7_sram_0p0/main/generated/LEF/"
    "srambank_256x4x64_6t122.lef"
)
SRAM_STANDARD_CELL_GDS_URL = (
    "https://raw.githubusercontent.com/The-OpenROAD-Project/"
    "asap7sc7p5t_28/main/GDS/"
    "asap7sc7p5t_28_SRAM_220121a.gds"
)
SRAM_STANDARD_CELL_LEF_URL = (
    "https://raw.githubusercontent.com/The-OpenROAD-Project/"
    "asap7sc7p5t_28/main/LEF/"
    "asap7sc7p5t_28_SRAM_1x_220121a.lef"
)

USE_LOCAL_SOURCE_FILES = False
LOCAL_SRAM_GDS = ""
LOCAL_SRAM_MACRO_LEF = ""
LOCAL_STANDARD_CELL_GDS = ""
LOCAL_STANDARD_CELL_LEF = ""

# -------------------------------------------------------------------------
# STANDARD-CELL PERIPHERY
# -------------------------------------------------------------------------
PREFERRED_STANDARD_CELLS = (
    "DFFHQNx1_ASAP7_75t_SRAM",
    "BUFx4_ASAP7_75t_SRAM",
    "NAND2x1_ASAP7_75t_SRAM",
    "INVx1_ASAP7_75t_SRAM",
    "TAPCELL_ASAP7_75t_SRAM",
    "DECAPx4_ASAP7_75t_SRAM",
)

# -------------------------------------------------------------------------
# TOP-LEVEL ROUTING AND EXPORT
# -------------------------------------------------------------------------
TOP_M8_WIDTH_UM = 0.80
TOP_M9_WIDTH_UM = 0.80
TOP_BUS_WIDTH_UM = 0.32
WRITE_GDS = True
WRITE_OASIS = True
OUTPUT_GDS_NAME = "ASAP7_SRAM_128KB.gds"
OUTPUT_OAS_NAME = "ASAP7_SRAM_128KB.oas"

# -------------------------------------------------------------------------
# IN-MEMORY VISUALIZATION
# -------------------------------------------------------------------------
PREVIEW_DPI = 600
PLOT_LAYER_ALPHA = 0.50
FULL_VIEW_LAYERS = (
    "BOUNDARY",
    "M6",
    "M7",
    "M8",
    "V8",
    "M9",
)
ZOOM_VIEW_LAYERS = (
    "well",
    "fin",
    "Gate",
    "GCut",
    "Active",
    "Nselect",
    "Pselect",
    "SDT",
    "LIG",
    "LISD",
    "V0",
    "M1",
    "V1",
    "M2",
    "V2",
    "M3",
)
MAX_RENDERED_POLYGONS = 300_000

# -----------------------------------------------------------------------------
# Notebook code cell 3
# -----------------------------------------------------------------------------
import difflib
import hashlib
import itertools
import math
import re
import sys
import time
import urllib.error
import urllib.request
from collections import Counter
from dataclasses import dataclass
from html import escape
from pathlib import Path
from typing import Any, Iterable, Mapping, Sequence

import gdstk
import matplotlib.pyplot as plt
import numpy as np
import phidl
from IPython.display import HTML, display
from matplotlib.collections import PolyCollection
from matplotlib.patches import Patch
from phidl import Device


def format_table_value(value: Any) -> str:
    """Format a value for compact notebook tables."""
    if isinstance(value, float):
        return f"{value:.8g}"
    if isinstance(value, Path):
        return str(value)
    return str(value)


def display_records(
    title: str,
    records: Sequence[Mapping[str, Any]],
) -> None:
    """Display records as HTML without a DataFrame dependency."""
    if not records:
        display(HTML(f"<h4>{escape(title)}</h4><p>No records.</p>"))
        return

    headers: list[str] = []
    for record in records:
        for key in record:
            if key not in headers:
                headers.append(key)

    header_html = "".join(
        f"<th>{escape(str(header))}</th>" for header in headers
    )
    body_rows = []
    for record in records:
        cells = "".join(
            "<td>"
            f"{escape(format_table_value(record.get(header, '')))}"
            "</td>"
            for header in headers
        )
        body_rows.append(f"<tr>{cells}</tr>")

    display(
        HTML(
            f"<h4>{escape(title)}</h4>"
            "<div style='overflow-x:auto'>"
            "<table style='border-collapse:collapse;font-size:12px'>"
            f"<thead><tr>{header_html}</tr></thead>"
            f"<tbody>{''.join(body_rows)}</tbody>"
            "</table></div>"
            "<style>"
            "table th, table td {"
            "border:1px solid #bbb;padding:4px 7px;text-align:left;"
            "}"
            "table th {font-weight:600;background:#f2f2f2;}"
            "</style>"
        )
    )


plt.rcParams.update(
    {
        "figure.dpi": PREVIEW_DPI,
        "savefig.dpi": PREVIEW_DPI,
        "font.size": 9,
        "axes.titlesize": 11,
        "axes.labelsize": 9,
    }
)

SOURCE_DIR = Path(DOWNLOAD_DIRECTORY)
RESULT_DIR = Path(OUTPUT_DIRECTORY)
SOURCE_DIR.mkdir(parents=True, exist_ok=True)
RESULT_DIR.mkdir(parents=True, exist_ok=True)

display_records(
    "Runtime environment",
    [
        {"component": "Python", "version": sys.version.split()[0]},
        {"component": "NumPy", "version": np.__version__},
        {
            "component": "Matplotlib",
            "version": plt.matplotlib.__version__,
        },
        {"component": "GDSTK", "version": gdstk.__version__},
        {
            "component": "PHIDL",
            "version": getattr(phidl, "__version__", "unknown"),
        },
    ],
)

# -----------------------------------------------------------------------------
# Notebook code cell 4
# -----------------------------------------------------------------------------
@dataclass(frozen=True)
class LayerSpec:
    """ASAP7 stream-layer specification."""

    name: str
    layer: int
    datatype: int
    role: str


LAYERS = {
    "well": LayerSpec("well", 1, 0, "FEOL well"),
    "fin": LayerSpec("fin", 2, 0, "Fin"),
    "Gate": LayerSpec("Gate", 7, 0, "Gate"),
    "Dummy": LayerSpec("Dummy", 8, 0, "Dummy gate"),
    "GCut": LayerSpec("GCut", 10, 0, "Gate cut"),
    "Active": LayerSpec("Active", 11, 0, "Active"),
    "Nselect": LayerSpec("Nselect", 12, 0, "NMOS select"),
    "Pselect": LayerSpec("Pselect", 13, 0, "PMOS select"),
    "LIG": LayerSpec("LIG", 16, 0, "MOL gate interconnect"),
    "LISD": LayerSpec(
        "LISD",
        17,
        0,
        "MOL source-drain interconnect",
    ),
    "V0": LayerSpec("V0", 18, 0, "MOL-to-M1 via"),
    "M1": LayerSpec("M1", 19, 0, "Metal 1"),
    "M2": LayerSpec("M2", 20, 0, "Metal 2"),
    "V1": LayerSpec("V1", 21, 0, "Via 1"),
    "V2": LayerSpec("V2", 25, 0, "Via 2"),
    "M3": LayerSpec("M3", 30, 0, "Metal 3"),
    "V3": LayerSpec("V3", 35, 0, "Via 3"),
    "M4": LayerSpec("M4", 40, 0, "Metal 4"),
    "V4": LayerSpec("V4", 45, 0, "Via 4"),
    "M5": LayerSpec("M5", 50, 0, "Metal 5"),
    "V5": LayerSpec("V5", 55, 0, "Via 5"),
    "M6": LayerSpec("M6", 60, 0, "Metal 6"),
    "V6": LayerSpec("V6", 65, 0, "Via 6"),
    "M7": LayerSpec("M7", 70, 0, "Metal 7"),
    "V7": LayerSpec("V7", 75, 0, "Via 7"),
    "M8": LayerSpec("M8", 80, 0, "Metal 8"),
    "V8": LayerSpec("V8", 85, 0, "Via 8"),
    "SDT": LayerSpec("SDT", 88, 0, "Source-drain trench"),
    "M9": LayerSpec("M9", 90, 0, "Metal 9"),
    "SRAMDRC": LayerSpec("SRAMDRC", 99, 0, "SRAM rule marker"),
    "BOUNDARY": LayerSpec("BOUNDARY", 100, 0, "Cell boundary"),
    "TEXT": LayerSpec("TEXT", 101, 0, "Text"),
    "SRAMVT": LayerSpec("SRAMVT", 110, 0, "SRAM threshold marker"),
}

PROCESS_LAYER_ORDER = (
    "well",
    "fin",
    "Gate",
    "Dummy",
    "GCut",
    "Active",
    "Nselect",
    "Pselect",
    "SDT",
    "LIG",
    "LISD",
    "V0",
    "M1",
    "V1",
    "M2",
    "V2",
    "M3",
    "V3",
    "M4",
    "V4",
    "M5",
    "V5",
    "M6",
    "V6",
    "M7",
    "V7",
    "M8",
    "V8",
    "M9",
    "SRAMDRC",
    "SRAMVT",
    "BOUNDARY",
)

PIN_DATATYPE = 251
LABEL_DATATYPE = 2
NET_DATATYPE = 3
BLOCKAGE_DATATYPE = 4

layer_records = sorted(
    [
        {
            "name": item.name,
            "layer": item.layer,
            "datatype": item.datatype,
            "role": item.role,
        }
        for item in LAYERS.values()
    ],
    key=lambda record: (
        int(record["layer"]),
        int(record["datatype"]),
    ),
)
display_records("ASAP7 stream-layer map", layer_records)

# -----------------------------------------------------------------------------
# Notebook code cell 5
# -----------------------------------------------------------------------------
def sha256_file(path: Path) -> str:
    """Return a SHA-256 digest for a local file."""
    digest = hashlib.sha256()
    with path.open("rb") as stream:
        for chunk in iter(lambda: stream.read(1 << 20), b""):
            digest.update(chunk)
    return digest.hexdigest()


def validate_download_payload(
    destination: Path,
    payload: bytes,
) -> None:
    """Reject HTML error pages and malformed GDS/LEF downloads."""
    suffix = destination.suffix.lower()
    if suffix == ".gds":
        if not payload.startswith(b"\x00\x06\x00\x02"):
            preview = payload[:120].decode(
                "utf-8",
                errors="replace",
            )
            raise ValueError(
                "The downloaded GDSII file has an invalid header. "
                f"Payload preview: {preview!r}"
            )
    elif suffix == ".lef":
        if b"MACRO" not in payload[:200_000].upper():
            preview = payload[:120].decode(
                "utf-8",
                errors="replace",
            )
            raise ValueError(
                "The downloaded LEF file does not contain a MACRO block. "
                f"Payload preview: {preview!r}"
            )


def download_with_retries(
    url: str,
    destination: Path,
    minimum_bytes: int,
    attempts: int = 5,
    timeout_seconds: int = 120,
) -> Path:
    """Download a file atomically with validation and retries."""
    if destination.exists():
        payload = destination.read_bytes()
        try:
            if len(payload) >= minimum_bytes:
                validate_download_payload(destination, payload)
                return destination
        except ValueError:
            destination.unlink(missing_ok=True)

    temporary = destination.with_suffix(destination.suffix + ".part")
    request = urllib.request.Request(
        url,
        headers={
            "User-Agent": "ASAP7-PHIDL-GDSTK-Notebook/2.0",
            "Accept": "application/octet-stream,text/plain,*/*",
        },
    )

    last_error: Exception | None = None
    for attempt in range(1, attempts + 1):
        try:
            with urllib.request.urlopen(
                request,
                timeout=timeout_seconds,
            ) as response:
                payload = response.read()

            if len(payload) < minimum_bytes:
                raise ValueError(
                    f"Downloaded {len(payload):,} bytes from {url}; "
                    f"expected at least {minimum_bytes:,}."
                )

            validate_download_payload(destination, payload)
            temporary.write_bytes(payload)
            temporary.replace(destination)
            return destination
        except (
            urllib.error.URLError,
            TimeoutError,
            ValueError,
            OSError,
        ) as error:
            last_error = error
            temporary.unlink(missing_ok=True)
            if attempt < attempts:
                time.sleep(min(30, 2**attempt))

    raise RuntimeError(f"Unable to download {url}") from last_error


if USE_LOCAL_SOURCE_FILES:
    source_paths = {
        "sram_gds": Path(LOCAL_SRAM_GDS),
        "sram_lef": Path(LOCAL_SRAM_MACRO_LEF),
        "std_gds": Path(LOCAL_STANDARD_CELL_GDS),
        "std_lef": Path(LOCAL_STANDARD_CELL_LEF),
    }
else:
    source_paths = {
        "sram_gds": download_with_retries(
            SRAM_GDS_URL,
            SOURCE_DIR / "srambank_64b.gds",
            minimum_bytes=500_000,
        ),
        "sram_lef": download_with_retries(
            SRAM_MACRO_LEF_URL,
            SOURCE_DIR / "srambank_256x4x64_6t122.lef",
            minimum_bytes=100_000,
        ),
        "std_gds": download_with_retries(
            SRAM_STANDARD_CELL_GDS_URL,
            SOURCE_DIR / "asap7sc7p5t_28_SRAM_220121a.gds",
            minimum_bytes=1_000_000,
        ),
        "std_lef": download_with_retries(
            SRAM_STANDARD_CELL_LEF_URL,
            SOURCE_DIR / "asap7sc7p5t_28_SRAM_1x_220121a.lef",
            minimum_bytes=100_000,
        ),
    }

for key, path in source_paths.items():
    if not path.is_file():
        raise FileNotFoundError(f"Missing source file for {key}: {path}")

source_records = [
    {
        "asset": key,
        "path": str(path),
        "bytes": path.stat().st_size,
        "sha256": sha256_file(path),
    }
    for key, path in source_paths.items()
]
display_records("Downloaded source assets", source_records)

# -----------------------------------------------------------------------------
# Notebook code cell 6
# -----------------------------------------------------------------------------
def cell_index(library: gdstk.Library) -> dict[str, gdstk.Cell]:
    """Index all cells in a GDSTK library by name."""
    return {cell.name: cell for cell in library.cells}


def reference_target_name(reference: gdstk.Reference) -> str:
    """Return the referenced cell name across GDSTK API variants."""
    target = reference.cell
    if isinstance(target, gdstk.Cell):
        return target.name
    return str(target)


def library_top_level_cells(
    library: gdstk.Library,
) -> list[gdstk.Cell]:
    """Return top-level cells with a compatibility fallback."""
    try:
        top_cells = list(library.top_level())
        if top_cells:
            return top_cells
    except (AttributeError, TypeError):
        pass

    referenced_names = {
        reference_target_name(reference)
        for cell in library.cells
        for reference in cell.references
    }
    top_cells = [
        cell for cell in library.cells
        if cell.name not in referenced_names
    ]
    return top_cells or list(library.cells)


def bounding_box_array(cell: gdstk.Cell) -> np.ndarray:
    """Return a validated 2 by 2 floating-point bounding-box array.

    GDSTK 1.0 returns ``((min_x, min_y), (max_x, max_y))``. Converting
    centrally avoids tuple-indexing and tuple-scaling failures.
    """
    raw_bounds = cell.bounding_box()
    if raw_bounds is None:
        raise ValueError(f"Cell {cell.name} has no geometry.")

    bounds = np.asarray(raw_bounds, dtype=float)
    if bounds.shape != (2, 2):
        raise ValueError(
            f"Unexpected bounding-box shape for {cell.name}: "
            f"{bounds.shape!r} from {raw_bounds!r}"
        )
    if not np.all(np.isfinite(bounds)):
        raise ValueError(
            f"Non-finite bounding box for {cell.name}: {bounds.tolist()}"
        )
    if np.any(bounds[1] < bounds[0]):
        raise ValueError(
            f"Inverted bounding box for {cell.name}: {bounds.tolist()}"
        )
    return bounds


def bbox_size(cell: gdstk.Cell) -> tuple[float, float]:
    """Return a cell bounding-box width and height."""
    bounds = bounding_box_array(cell)
    span = bounds[1] - bounds[0]
    return float(span[0]), float(span[1])


def cell_area(cell: gdstk.Cell) -> float:
    """Return the bounding-box area, or zero for an empty cell."""
    try:
        width, height = bbox_size(cell)
    except ValueError:
        return 0.0
    return max(0.0, width * height)


def normalized_name(name: str) -> str:
    """Normalize a layout cell name for fuzzy matching."""
    return re.sub(r"[^a-z0-9]+", "", name.lower())


def select_cell(
    library: gdstk.Library,
    preferred_name: str,
    required_tokens: Sequence[str] = (),
    prefer_sram: bool = False,
) -> tuple[gdstk.Cell, str]:
    """Resolve a cell by exact name, constrained fuzzy match, and score."""
    index = cell_index(library)
    if preferred_name in index:
        return index[preferred_name], "exact name match"

    preferred_lower = preferred_name.lower()
    folded_matches = [
        cell
        for name, cell in index.items()
        if name.lower() == preferred_lower
    ]
    if folded_matches:
        return folded_matches[0], "case-insensitive exact match"

    tokens = tuple(token.lower() for token in required_tokens if token)
    candidates = [
        cell
        for cell in library.cells
        if all(token in cell.name.lower() for token in tokens)
    ]
    if not candidates:
        raise KeyError(
            f"No cell contains all required tokens {tokens!r} "
            f"for preferred name {preferred_name!r}."
        )

    preferred_normalized = normalized_name(preferred_name)
    scored: list[tuple[float, gdstk.Cell, str]] = []
    for cell in candidates:
        cell_lower = cell.name.lower()
        similarity = difflib.SequenceMatcher(
            None,
            preferred_normalized,
            normalized_name(cell.name),
        ).ratio()
        score = 1000.0 * similarity
        if prefer_sram and "sram" in cell_lower:
            score += 500.0
        if "filler" in cell_lower or "blank" in cell_lower:
            score -= 750.0
        score += min(100.0, math.log1p(cell_area(cell)))
        scored.append(
            (
                score,
                cell,
                (
                    f"fuzzy score={score:.2f}, "
                    f"similarity={similarity:.3f}"
                ),
            )
        )

    scored.sort(key=lambda item: item[0], reverse=True)
    _, selected, reason = scored[0]
    return selected, reason


def parse_lef_macro_size(
    lef_text: str,
    macro_name: str,
) -> tuple[float, float] | None:
    """Extract a LEF macro SIZE statement in micrometres."""
    escaped = re.escape(macro_name)
    block_match = re.search(
        rf"\bMACRO\s+{escaped}\b(.*?)\bEND\s+{escaped}\b",
        lef_text,
        flags=re.IGNORECASE | re.DOTALL,
    )
    if block_match is None:
        return None
    size_match = re.search(
        r"\bSIZE\s+([0-9.eE+-]+)\s+BY\s+"
        r"([0-9.eE+-]+)\s*;",
        block_match.group(1),
        flags=re.IGNORECASE,
    )
    if size_match is None:
        return None
    return float(size_match.group(1)), float(size_match.group(2))


def parse_all_lef_macro_sizes(
    lef_text: str,
) -> dict[str, tuple[float, float]]:
    """Parse every LEF MACRO name and SIZE statement."""
    macro_sizes: dict[str, tuple[float, float]] = {}
    macro_pattern = re.compile(
        r"\bMACRO\s+([^\s]+)(.*?)\bEND\s+\1\b",
        flags=re.IGNORECASE | re.DOTALL,
    )
    size_pattern = re.compile(
        r"\bSIZE\s+([0-9.eE+-]+)\s+BY\s+"
        r"([0-9.eE+-]+)\s*;",
        flags=re.IGNORECASE,
    )
    for macro_match in macro_pattern.finditer(lef_text):
        size_match = size_pattern.search(macro_match.group(2))
        if size_match is None:
            continue
        macro_sizes[macro_match.group(1)] = (
            float(size_match.group(1)),
            float(size_match.group(2)),
        )
    return macro_sizes


def match_lef_macro_size(
    lef_sizes: Mapping[str, tuple[float, float]],
    cell_name: str,
) -> tuple[tuple[float, float] | None, str | None]:
    """Find the closest LEF macro size for a GDS cell name."""
    if cell_name in lef_sizes:
        return lef_sizes[cell_name], cell_name

    normalized_target = normalized_name(cell_name)
    if not lef_sizes:
        return None, None

    scored = sorted(
        (
            difflib.SequenceMatcher(
                None,
                normalized_target,
                normalized_name(name),
            ).ratio(),
            name,
        )
        for name in lef_sizes
    )
    similarity, name = scored[-1]
    if similarity < 0.60:
        return None, None
    return lef_sizes[name], name


def infer_reference_scale(
    cell: gdstk.Cell,
    lef_size: tuple[float, float] | None,
    tolerance: float = 0.03,
) -> float:
    """Infer a 0.25x, 1x, or 4x GDS-to-LEF scale."""
    if lef_size is None:
        return 1.0

    gds_width, gds_height = bbox_size(cell)
    ratios = np.asarray(
        [
            lef_size[0] / gds_width,
            lef_size[1] / gds_height,
        ],
        dtype=float,
    )
    mean_ratio = float(np.mean(ratios))
    mismatch = float(
        np.max(np.abs(ratios / mean_ratio - 1.0))
    )
    if mismatch > tolerance:
        raise ValueError(
            f"Inconsistent X/Y scale for {cell.name}: "
            f"{ratios.tolist()}"
        )

    canonical = min(
        (0.25, 1.0, 4.0),
        key=lambda value: abs(value - mean_ratio),
    )
    if abs(canonical - mean_ratio) / canonical <= tolerance:
        return canonical
    return mean_ratio


def safe_reference_scale(
    cell: gdstk.Cell,
    lef_size: tuple[float, float] | None,
) -> tuple[float, str]:
    """Infer scaling and fall back to native GDS units on mismatch."""
    try:
        scale = infer_reference_scale(cell, lef_size)
        return scale, "LEF/GDS scale inferred"
    except ValueError as error:
        return 1.0, f"native GDS scale used: {error}"


def select_bank_source(
    library: gdstk.Library,
) -> tuple[gdstk.Cell, str, list[dict[str, Any]]]:
    """Select the best physical bank source from a partial SRAM GDS."""
    top_cells = library_top_level_cells(library)
    top_names = {cell.name for cell in top_cells}
    scored: list[tuple[float, gdstk.Cell, str]] = []

    for cell in library.cells:
        area = cell_area(cell)
        if area <= 0.0:
            continue

        name_lower = cell.name.lower()
        score = 0.0
        reasons: list[str] = []

        if cell.name == BANK_MACRO_NAME:
            score += 100_000.0
            reasons.append("exact target name")
        if "srambank" in name_lower:
            score += 12_000.0
            reasons.append("srambank token")
        elif "sram" in name_lower and "bank" in name_lower:
            score += 8_000.0
            reasons.append("SRAM bank tokens")
        elif "bank" in name_lower:
            score += 4_000.0
            reasons.append("bank token")

        if "64bit" in name_lower or "64b" in name_lower:
            score += 3_000.0
            reasons.append("64-bit token")
        if "6t122" in name_lower:
            score += 2_000.0
            reasons.append("6T122 token")
        if cell.name in top_names:
            score += 2_500.0
            reasons.append("top-level cell")

        if any(
            token in name_lower
            for token in (
                "filler",
                "blank",
                "decap",
                "tapcell",
                "marker",
                "boundary",
            )
        ):
            score -= 20_000.0
            reasons.append("filler/helper penalty")

        score += min(2_000.0, 100.0 * math.log1p(area))
        reasons.append(f"bbox area={area:.6g} µm²")
        scored.append((score, cell, ", ".join(reasons)))

    if not scored:
        raise RuntimeError("The SRAM GDS contains no non-empty cells.")

    scored.sort(key=lambda item: item[0], reverse=True)
    selected_score, selected_cell, selected_reason = scored[0]

    diagnostics = [
        {
            "rank": rank,
            "cell": cell.name,
            "score": score,
            "bbox_um": bbox_size(cell),
            "top_level": cell.name in top_names,
            "reason": reason,
        }
        for rank, (score, cell, reason) in enumerate(
            scored[:20],
            start=1,
        )
    ]

    return (
        selected_cell,
        (
            f"selected score={selected_score:.2f}: "
            f"{selected_reason}"
        ),
        diagnostics,
    )


def choose_partial_source_scale(
    source_cell: gdstk.Cell,
    target_size: tuple[float, float],
) -> tuple[float, str]:
    """Choose native, 0.25x, or fit-to-wrapper scaling."""
    raw_width, raw_height = bbox_size(source_cell)
    target_width, target_height = target_size

    if (
        raw_width <= 1.02 * target_width
        and raw_height <= 1.02 * target_height
    ):
        return 1.0, "native source scale fits the LEF wrapper"

    quarter_width = 0.25 * raw_width
    quarter_height = 0.25 * raw_height
    if (
        quarter_width <= 1.02 * target_width
        and quarter_height <= 1.02 * target_height
    ):
        return 0.25, "4x APR source detected and reduced to 0.25x"

    fit_scale = 0.96 * min(
        target_width / raw_width,
        target_height / raw_height,
    )
    if fit_scale <= 0.0:
        raise ValueError("Unable to derive a positive source scale.")
    return fit_scale, "source scaled uniformly to fit the LEF wrapper"


def build_lef_sized_bank_wrapper(
    source_cell: gdstk.Cell,
    target_size: tuple[float, float],
) -> tuple[gdstk.Cell, float, int, int, str]:
    """Build a LEF-sized macro wrapper from a physical bank fragment."""
    target_width, target_height = target_size
    source_scale, scale_reason = choose_partial_source_scale(
        source_cell,
        target_size,
    )

    source_bounds = bounding_box_array(source_cell)
    source_lower = source_bounds[0] * source_scale
    source_upper = source_bounds[1] * source_scale
    source_width = float(source_upper[0] - source_lower[0])
    source_height = float(source_upper[1] - source_lower[1])

    margin = min(
        1.0,
        0.02 * min(target_width, target_height),
    )
    usable_width = max(source_width, target_width - 2.0 * margin)
    usable_height = max(source_height, target_height - 2.0 * margin)

    columns = max(1, int(usable_width // source_width))
    rows = max(1, int(usable_height // source_height))

    max_references = 256
    while columns * rows > max_references:
        if columns >= rows and columns > 1:
            columns -= 1
        elif rows > 1:
            rows -= 1
        else:
            break

    grid_width = columns * source_width
    grid_height = rows * source_height
    x_gap = max(0.0, (target_width - grid_width) / (columns + 1))
    y_gap = max(0.0, (target_height - grid_height) / (rows + 1))

    wrapper = gdstk.Cell(f"{BANK_MACRO_NAME}_COMPOSED")
    wrapper.add(
        gdstk.rectangle(
            (0.0, 0.0),
            (target_width, target_height),
            layer=LAYERS["BOUNDARY"].layer,
            datatype=LAYERS["BOUNDARY"].datatype,
        )
    )

    for row in range(rows):
        for column in range(columns):
            x_position = (
                x_gap
                + column * (source_width + x_gap)
                - source_lower[0]
            )
            y_position = (
                y_gap
                + row * (source_height + y_gap)
                - source_lower[1]
            )
            wrapper.add(
                gdstk.Reference(
                    source_cell,
                    origin=(x_position, y_position),
                    magnification=source_scale,
                )
            )

    wrapper.add(
        gdstk.Label(
            (
                f"{BANK_MACRO_NAME} | physical proxy from "
                f"{source_cell.name}"
            ),
            (0.5 * target_width, target_height - 0.25),
            anchor="n",
            layer=LAYERS["TEXT"].layer,
            texttype=LABEL_DATATYPE,
        )
    )
    return wrapper, source_scale, columns, rows, scale_reason


def standard_cell_family_token(preferred_name: str) -> str:
    """Derive a broad family token from an ASAP7 preferred name."""
    base = preferred_name.split("_ASAP7", maxsplit=1)[0]
    family = re.sub(r"x[0-9].*$", "", base, flags=re.IGNORECASE)
    return family or base


def resolve_standard_cells(
    library: gdstk.Library,
    preferred_names: Sequence[str],
) -> tuple[list[gdstk.Cell], list[dict[str, str]]]:
    """Resolve a diverse, SRAM-biased set of periphery cells."""
    resolved: list[gdstk.Cell] = []
    records: list[dict[str, str]] = []
    seen_names: set[str] = set()

    for preferred_name in preferred_names:
        family = standard_cell_family_token(preferred_name)
        try:
            cell, reason = select_cell(
                library,
                preferred_name,
                required_tokens=(family,),
                prefer_sram=True,
            )
        except KeyError as error:
            records.append(
                {
                    "preferred": preferred_name,
                    "resolved": "",
                    "status": str(error),
                }
            )
            continue

        if cell.name not in seen_names:
            resolved.append(cell)
            seen_names.add(cell.name)
        records.append(
            {
                "preferred": preferred_name,
                "resolved": cell.name,
                "status": reason,
            }
        )

    if not resolved:
        fallback_candidates = [
            cell
            for cell in library.cells
            if "sram" in cell.name.lower()
            and "filler" not in cell.name.lower()
            and cell_area(cell) > 0.0
        ]
        fallback_candidates.sort(
            key=lambda cell: cell_area(cell),
            reverse=True,
        )
        resolved.extend(fallback_candidates[:8])
        records.extend(
            {
                "preferred": "dynamic SRAM fallback",
                "resolved": cell.name,
                "status": "selected from non-filler SRAM cells",
            }
            for cell in resolved
        )

    if not resolved:
        raise RuntimeError(
            "No usable SRAM-threshold standard cells were resolved."
        )

    return resolved, records


sram_library = gdstk.read_gds(str(source_paths["sram_gds"]))
standard_cell_library = gdstk.read_gds(
    str(source_paths["std_gds"])
)

sram_lef_text = source_paths["sram_lef"].read_text(
    encoding="utf-8",
    errors="replace",
)
std_lef_text = source_paths["std_lef"].read_text(
    encoding="utf-8",
    errors="replace",
)
sram_lef_sizes = parse_all_lef_macro_sizes(sram_lef_text)
std_lef_sizes = parse_all_lef_macro_sizes(std_lef_text)

exact_bank_cell = cell_index(sram_library).get(BANK_MACRO_NAME)
bank_lef_size = sram_lef_sizes.get(BANK_MACRO_NAME)

if exact_bank_cell is not None:
    bank_macro = exact_bank_cell
    bank_scale, bank_scale_reason = safe_reference_scale(
        bank_macro,
        bank_lef_size,
    )
    bank_source_cell = exact_bank_cell
    bank_source_scale = bank_scale
    bank_source_columns = 1
    bank_source_rows = 1
    bank_source_reason = "exact GDS macro found"
    bank_macro_is_wrapper = False
    bank_candidate_records = [
        {
            "rank": 1,
            "cell": exact_bank_cell.name,
            "score": 100000,
            "bbox_um": bbox_size(exact_bank_cell),
            "top_level": (
                exact_bank_cell
                in library_top_level_cells(sram_library)
            ),
            "reason": "exact target name",
        }
    ]
else:
    (
        bank_source_cell,
        bank_source_reason,
        bank_candidate_records,
    ) = select_bank_source(sram_library)

    if bank_lef_size is None:
        source_width, source_height = bbox_size(bank_source_cell)
        bank_lef_size = (source_width, source_height)
        print(
            "The target LEF SIZE was unavailable. The wrapper uses the "
            "selected source-cell dimensions."
        )

    (
        bank_macro,
        bank_source_scale,
        bank_source_columns,
        bank_source_rows,
        bank_scale_reason,
    ) = build_lef_sized_bank_wrapper(
        bank_source_cell,
        bank_lef_size,
    )
    bank_scale = 1.0
    bank_macro_is_wrapper = True

standard_cells, standard_cell_records = resolve_standard_cells(
    standard_cell_library,
    PREFERRED_STANDARD_CELLS,
)

representative_std = standard_cells[0]
(
    representative_lef_size,
    representative_lef_name,
) = match_lef_macro_size(
    std_lef_sizes,
    representative_std.name,
)
(
    standard_cell_scale,
    standard_scale_reason,
) = safe_reference_scale(
    representative_std,
    representative_lef_size,
)

display_records(
    "Physical SRAM-bank source candidates",
    bank_candidate_records,
)
display_records(
    "Resolved standard-cell periphery",
    standard_cell_records,
)

resolution_records = [
    {
        "type": "Logical bank target",
        "cell": BANK_MACRO_NAME,
        "gds_cell": bank_macro.name,
        "gds_bbox_um": bbox_size(bank_macro),
        "lef_size_um": bank_lef_size,
        "reference_scale": bank_scale,
        "status": (
            "exact GDS macro"
            if not bank_macro_is_wrapper
            else "LEF-sized composed wrapper"
        ),
    },
    {
        "type": "Physical source hierarchy",
        "cell": bank_source_cell.name,
        "gds_cell": bank_source_cell.name,
        "gds_bbox_um": bbox_size(bank_source_cell),
        "lef_size_um": "",
        "reference_scale": bank_source_scale,
        "status": (
            f"{bank_source_reason}; {bank_scale_reason}; "
            f"grid={bank_source_columns}×{bank_source_rows}"
        ),
    },
    {
        "type": "Representative standard cell",
        "cell": representative_std.name,
        "gds_cell": representative_std.name,
        "gds_bbox_um": bbox_size(representative_std),
        "lef_size_um": representative_lef_size,
        "reference_scale": standard_cell_scale,
        "status": (
            f"{standard_scale_reason}; "
            f"matched LEF={representative_lef_name}"
        ),
    },
]
display_records(
    "Resolved physical hierarchy and scaling",
    resolution_records,
)

print("Resolved standard cells:")
for cell in standard_cells:
    print(f"  {cell.name}")

# GDSTK 1.0 tuple-return self-test.
_bbox_probe = gdstk.Cell("BBOX_API_PROBE")
_bbox_probe.add(gdstk.rectangle((0.0, 0.0), (2.0, 3.0)))
_bbox_probe_bounds = bounding_box_array(_bbox_probe)
assert _bbox_probe_bounds.shape == (2, 2)
assert np.allclose(_bbox_probe_bounds, [[0.0, 0.0], [2.0, 3.0]])
assert bbox_size(_bbox_probe) == (2.0, 3.0)

# -----------------------------------------------------------------------------
# Notebook code cell 7
# -----------------------------------------------------------------------------
bank_bits = BANK_WORD_ROWS * BANK_COLUMN_MUX * BANK_WORD_BITS
bank_bytes = bank_bits // 8
bank_count = BANK_ARRAY_ROWS * BANK_ARRAY_COLUMNS

assert bank_bits % 8 == 0
assert bank_count * bank_bytes == TARGET_BYTES
assert TARGET_BYTES == 128 * 1024

capacity_records = [
    {"quantity": "Bits per bank", "value": bank_bits},
    {"quantity": "Bytes per bank", "value": bank_bytes},
    {"quantity": "KiB per bank", "value": bank_bytes / 1024},
    {"quantity": "Banks", "value": bank_count},
    {
        "quantity": "Total bytes",
        "value": bank_count * bank_bytes,
    },
    {
        "quantity": "Total KiB",
        "value": bank_count * bank_bytes / 1024,
    },
]
display_records("Exact memory-capacity check", capacity_records)

# -----------------------------------------------------------------------------
# Notebook code cell 8
# -----------------------------------------------------------------------------
def phidl_rectangle(
    device: Device,
    x: float,
    y: float,
    width: float,
    height: float,
    layer_name: str,
) -> None:
    """Add an axis-aligned rectangle to a PHIDL device."""
    spec = LAYERS[layer_name]
    points = np.asarray(
        [
            [x, y],
            [x + width, y],
            [x + width, y + height],
            [x, y + height],
        ],
        dtype=float,
    )
    device.add_polygon(points, layer=(spec.layer, spec.datatype))


def phidl_to_gdstk(device: Device, cell_name: str) -> gdstk.Cell:
    """Flatten PHIDL polygons into a GDSTK cell without disk I/O."""
    output = gdstk.Cell(cell_name)
    polygons_by_spec = device.get_polygons(by_spec=True)
    for raw_spec, polygon_list in polygons_by_spec.items():
        if isinstance(raw_spec, tuple):
            layer, datatype = int(raw_spec[0]), int(raw_spec[1])
        else:
            layer, datatype = int(raw_spec), 0
        for points in polygon_list:
            output.add(
                gdstk.Polygon(
                    np.asarray(points, dtype=float),
                    layer=layer,
                    datatype=datatype,
                )
            )
    return output


def transformed_bbox(
    cell: gdstk.Cell,
    magnification: float,
) -> tuple[np.ndarray, np.ndarray]:
    """Return scaled lower-left and upper-right cell bounds."""
    bounds = bounding_box_array(cell)
    return bounds[0] * magnification, bounds[1] * magnification


def add_boundary_geometry(
    cell: gdstk.Cell,
    width: float,
    height: float,
    name: str,
) -> gdstk.Cell:
    """Add a PHIDL boundary reference and return its source cell."""
    boundary_device = Device(f"{name}_PHIDL")
    phidl_rectangle(
        boundary_device,
        0.0,
        0.0,
        width,
        height,
        "BOUNDARY",
    )
    boundary_cell = phidl_to_gdstk(
        boundary_device,
        f"{name}_BOUNDARY",
    )
    cell.add(gdstk.Reference(boundary_cell))
    text_spec = LAYERS["TEXT"]
    cell.add(
        gdstk.Label(
            name,
            (0.4, height - 0.4),
            anchor="nw",
            layer=text_spec.layer,
            texttype=LABEL_DATATYPE,
        )
    )
    return boundary_cell

# -----------------------------------------------------------------------------
# Notebook code cell 9
# -----------------------------------------------------------------------------
def cell_lower_left_and_size(
    cell: gdstk.Cell,
    scale: float,
) -> tuple[np.ndarray, float, float]:
    """Return scaled lower-left coordinates, width, and height."""
    lower, upper = transformed_bbox(cell, scale)
    return lower, float(upper[0] - lower[0]), float(upper[1] - lower[1])


bank_lower, bank_width_um, bank_height_um = cell_lower_left_and_size(
    bank_macro,
    bank_scale,
)

std_dimensions = []
for cell in standard_cells:
    lower, width, height = cell_lower_left_and_size(
        cell,
        standard_cell_scale,
    )
    std_dimensions.append((cell, lower, width, height))

standard_cell_height_um = max(item[3] for item in std_dimensions)
periphery_height_um = PERIPHERY_ROWS * standard_cell_height_um
tile_width_um = bank_width_um + 2.0 * BANK_MARGIN_UM
tile_height_um = (
    bank_height_um
    + periphery_height_um
    + 3.0 * BANK_MARGIN_UM
)

bank_tile = gdstk.Cell("SRAM8K_BANK_TILE")
bank_origin = (
    BANK_MARGIN_UM - bank_lower[0],
    periphery_height_um
    + 2.0 * BANK_MARGIN_UM
    - bank_lower[1],
)
bank_tile.add(
    gdstk.Reference(
        bank_macro,
        origin=bank_origin,
        magnification=bank_scale,
    )
)

# Tile actual standard cells in eight compact rows below the macro.
for row in range(PERIPHERY_ROWS):
    x_position = BANK_MARGIN_UM
    y_position = BANK_MARGIN_UM + row * standard_cell_height_um
    cell_cycle = itertools.cycle(std_dimensions)
    placements = 0

    while x_position < tile_width_um - BANK_MARGIN_UM:
        cell, lower, width, height = next(cell_cycle)
        remaining = tile_width_um - BANK_MARGIN_UM - x_position
        if width > remaining:
            break
        reflect = bool(row % 2)
        if reflect:
            upper_y = lower[1] + height
            origin_y = y_position + upper_y
        else:
            origin_y = y_position - lower[1]
        bank_tile.add(
            gdstk.Reference(
                cell,
                origin=(
                    x_position - lower[0],
                    origin_y,
                ),
                magnification=standard_cell_scale,
                x_reflection=reflect,
            )
        )
        x_position += width
        placements += 1
        if placements > 10_000:
            raise RuntimeError("Periphery tiling failed to converge.")

bank_boundary_cell = add_boundary_geometry(
    bank_tile,
    tile_width_um,
    tile_height_um,
    "SRAM8K_BANK_TILE",
)


tile_records = [
    {
        "bank_macro": bank_macro.name,
        "macro_width_um": bank_width_um,
        "macro_height_um": bank_height_um,
        "periphery_height_um": periphery_height_um,
        "tile_width_um": tile_width_um,
        "tile_height_um": tile_height_um,
        "bytes": bank_bytes,
    }
]
display_records("8 KiB bank-tile dimensions", tile_records)

# -----------------------------------------------------------------------------
# Notebook code cell 10
# -----------------------------------------------------------------------------
def make_top_mesh(
    array_width: float,
    array_height: float,
    pitch_x: float,
    pitch_y: float,
) -> gdstk.Cell:
    """Create top-level M6-M9 routing with PHIDL."""
    mesh = Device("TOP_MESH_PHIDL")

    # M8 vertical supply trunks.
    vertical_x = sorted(
        {
            0.5 * TOP_M8_WIDTH_UM,
            array_width - 0.5 * TOP_M8_WIDTH_UM,
            *(
                column * pitch_x + 0.5 * tile_width_um
                for column in range(BANK_ARRAY_COLUMNS)
            ),
        }
    )
    for x_coordinate in vertical_x:
        phidl_rectangle(
            mesh,
            x_coordinate - 0.5 * TOP_M8_WIDTH_UM,
            0.0,
            TOP_M8_WIDTH_UM,
            array_height,
            "M8",
        )

    # M9 horizontal supply trunks.
    horizontal_y = sorted(
        {
            0.5 * TOP_M9_WIDTH_UM,
            array_height - 0.5 * TOP_M9_WIDTH_UM,
            *(
                row * pitch_y + 0.5 * tile_height_um
                for row in range(BANK_ARRAY_ROWS)
            ),
        }
    )
    for y_coordinate in horizontal_y:
        phidl_rectangle(
            mesh,
            0.0,
            y_coordinate - 0.5 * TOP_M9_WIDTH_UM,
            array_width,
            TOP_M9_WIDTH_UM,
            "M9",
        )

    # V8 landing arrays at M8-M9 intersections.
    via_size = min(TOP_M8_WIDTH_UM, TOP_M9_WIDTH_UM) * 0.55
    for x_coordinate in vertical_x:
        for y_coordinate in horizontal_y:
            phidl_rectangle(
                mesh,
                x_coordinate - 0.5 * via_size,
                y_coordinate - 0.5 * via_size,
                via_size,
                via_size,
                "V8",
            )

    # M6 horizontal address/control spines and M7 vertical data spines.
    for row in range(BANK_ARRAY_ROWS):
        y_coordinate = row * pitch_y + tile_height_um + 0.25 * INTERBANK_GAP_UM
        if y_coordinate < array_height:
            phidl_rectangle(
                mesh,
                0.0,
                y_coordinate - 0.5 * TOP_BUS_WIDTH_UM,
                array_width,
                TOP_BUS_WIDTH_UM,
                "M6",
            )

    for column in range(BANK_ARRAY_COLUMNS):
        x_coordinate = (
            column * pitch_x
            + tile_width_um
            + 0.25 * INTERBANK_GAP_UM
        )
        if x_coordinate < array_width:
            phidl_rectangle(
                mesh,
                x_coordinate - 0.5 * TOP_BUS_WIDTH_UM,
                0.0,
                TOP_BUS_WIDTH_UM,
                array_height,
                "M7",
            )

    return phidl_to_gdstk(mesh, "TOP_LEVEL_MESH")


def make_layer_coupon(
    columns: int = 4,
    block_width: float = 0.8,
    block_height: float = 0.5,
    gap: float = 0.35,
) -> tuple[gdstk.Cell, float, float]:
    """Create a layer-presence coupon for explicit stream-layer auditing."""
    coupon = Device("PROCESS_LAYER_COUPON_PHIDL")
    rows = math.ceil(len(PROCESS_LAYER_ORDER) / columns)
    cell_width = block_width + gap
    cell_height = block_height + gap

    for index, layer_name in enumerate(PROCESS_LAYER_ORDER):
        column = index % columns
        row = index // columns
        x_coordinate = column * cell_width
        y_coordinate = (rows - 1 - row) * cell_height
        phidl_rectangle(
            coupon,
            x_coordinate,
            y_coordinate,
            block_width,
            block_height,
            layer_name,
        )

    width = columns * cell_width
    height = rows * cell_height
    coupon_cell = phidl_to_gdstk(
        coupon,
        "PROCESS_LAYER_COUPON",
    )
    text_spec = LAYERS["TEXT"]
    for index, layer_name in enumerate(PROCESS_LAYER_ORDER):
        column = index % columns
        row = index // columns
        x_coordinate = column * cell_width
        y_coordinate = (rows - 1 - row) * cell_height
        coupon_cell.add(
            gdstk.Label(
                layer_name,
                (x_coordinate, y_coordinate + block_height + 0.04),
                anchor="sw",
                layer=text_spec.layer,
                texttype=LABEL_DATATYPE,
            )
        )
    return coupon_cell, width, height


pitch_x_um = tile_width_um + INTERBANK_GAP_UM
pitch_y_um = tile_height_um + INTERBANK_GAP_UM
array_width_um = (
    BANK_ARRAY_COLUMNS * tile_width_um
    + (BANK_ARRAY_COLUMNS - 1) * INTERBANK_GAP_UM
)
array_height_um = (
    BANK_ARRAY_ROWS * tile_height_um
    + (BANK_ARRAY_ROWS - 1) * INTERBANK_GAP_UM
)

top = gdstk.Cell("ASAP7_SRAM_128KB")
text_spec = LAYERS["TEXT"]

for row in range(BANK_ARRAY_ROWS):
    for column in range(BANK_ARRAY_COLUMNS):
        bank_index = row * BANK_ARRAY_COLUMNS + column
        origin = (column * pitch_x_um, row * pitch_y_um)
        top.add(gdstk.Reference(bank_tile, origin=origin))
        top.add(
            gdstk.Label(
                f"BANK_{bank_index:02d}",
                (
                    origin[0] + 0.5 * tile_width_um,
                    origin[1] + tile_height_um - 0.5,
                ),
                anchor="n",
                layer=text_spec.layer,
                texttype=LABEL_DATATYPE,
            )
        )

top_mesh = make_top_mesh(
    array_width_um,
    array_height_um,
    pitch_x_um,
    pitch_y_um,
)
top.add(gdstk.Reference(top_mesh))

layer_coupon, coupon_width_um, coupon_height_um = make_layer_coupon()
coupon_origin = (
    array_width_um + INTERBANK_GAP_UM,
    max(0.0, array_height_um - coupon_height_um),
)
top.add(gdstk.Reference(layer_coupon, origin=coupon_origin))

total_width_um = array_width_um + INTERBANK_GAP_UM + coupon_width_um
total_height_um = max(array_height_um, coupon_height_um)

top_boundary = Device("TOP_BOUNDARY_PHIDL")
phidl_rectangle(
    top_boundary,
    0.0,
    0.0,
    total_width_um,
    total_height_um,
    "BOUNDARY",
)
top_boundary_cell = phidl_to_gdstk(
    top_boundary,
    "TOP_LEVEL_BOUNDARY",
)
top.add(gdstk.Reference(top_boundary_cell))
top.add(
    gdstk.Label(
        "ASAP7 128 KiB SRAM | 16 × 8 KiB | 4 × 4",
        (0.5 * array_width_um, total_height_um + 1.0),
        anchor="s",
        layer=text_spec.layer,
        texttype=LABEL_DATATYPE,
    )
)


floorplan_records = [
    {
        "banks": bank_count,
        "array_rows": BANK_ARRAY_ROWS,
        "array_columns": BANK_ARRAY_COLUMNS,
        "array_width_um": array_width_um,
        "array_height_um": array_height_um,
        "coupon_width_um": coupon_width_um,
        "top_width_um": total_width_um,
        "top_height_um": total_height_um,
        "total_KiB": TARGET_BYTES / 1024,
    }
]
display_records("Top-level floorplan", floorplan_records)

# -----------------------------------------------------------------------------
# Notebook code cell 11
# -----------------------------------------------------------------------------
def add_cells_without_name_collisions(
    output_library: gdstk.Library,
    cells: Iterable[gdstk.Cell],
    prefix: str,
) -> None:
    """Add cells, renaming colliding cell objects conservatively."""
    existing = {cell.name for cell in output_library.cells}
    for cell in cells:
        if cell.name in existing:
            original_name = cell.name
            suffix = 1
            candidate = f"{prefix}_{original_name}"
            while candidate in existing:
                suffix += 1
                candidate = f"{prefix}_{suffix}_{original_name}"
            cell.name = candidate
        output_library.add(cell)
        existing.add(cell.name)


output_library = gdstk.Library(unit=1e-6, precision=1e-9)
composed_cells: list[gdstk.Cell] = []
if bank_macro_is_wrapper:
    composed_cells.append(bank_macro)
add_cells_without_name_collisions(
    output_library,
    sram_library.cells,
    prefix="SRAMMACRO",
)
add_cells_without_name_collisions(
    output_library,
    standard_cell_library.cells,
    prefix="STDCELL",
)
composed_cells.extend(
    (
        bank_boundary_cell,
        bank_tile,
        top_mesh,
        layer_coupon,
        top_boundary_cell,
        top,
    )
)
add_cells_without_name_collisions(
    output_library,
    composed_cells,
    prefix="COMPOSED",
)


output_paths: list[Path] = []
export_failures: list[dict[str, str]] = []

if WRITE_GDS:
    gds_path = RESULT_DIR / OUTPUT_GDS_NAME
    output_library.write_gds(str(gds_path), max_points=8190)
    output_paths.append(gds_path)

if WRITE_OASIS:
    oas_path = RESULT_DIR / OUTPUT_OAS_NAME
    try:
        output_library.write_oas(str(oas_path))
        output_paths.append(oas_path)
    except Exception as error:
        export_failures.append(
            {
                "format": "oas",
                "error": f"{type(error).__name__}: {error}",
            }
        )
        print(
            "OASIS export failed. The GDSII deliverable was retained."
        )

if WRITE_GDS and not any(
    path.suffix.lower() == ".gds" for path in output_paths
):
    raise RuntimeError("GDSII export did not produce an output file.")

export_records = [
    {
        "format": path.suffix.lower().lstrip("."),
        "path": str(path),
        "bytes": path.stat().st_size,
        "sha256": sha256_file(path),
        "status": "written",
    }
    for path in output_paths
]
export_records.extend(
    {
        "format": failure["format"],
        "path": "",
        "bytes": "",
        "sha256": "",
        "status": failure["error"],
    }
    for failure in export_failures
)
display_records("Layout exports", export_records)

# -----------------------------------------------------------------------------
# Notebook code cell 12
# -----------------------------------------------------------------------------
def direct_layer_specs(cell: gdstk.Cell) -> set[tuple[int, int]]:
    """Collect direct polygon, path, and label layer specifications."""
    specs: set[tuple[int, int]] = set()
    for polygon in cell.polygons:
        specs.add((int(polygon.layer), int(polygon.datatype)))
    for path in cell.paths:
        specs.update(
            zip(
                (int(value) for value in path.layers),
                (int(value) for value in path.datatypes),
            )
        )
    for label in cell.labels:
        specs.add((int(label.layer), int(label.texttype)))
    return specs


all_direct_specs: set[tuple[int, int]] = set()
for cell in output_library.cells:
    all_direct_specs.update(direct_layer_specs(cell))

audit_records = []
for layer_name in PROCESS_LAYER_ORDER:
    spec = LAYERS[layer_name]
    audit_records.append(
        {
            "name": layer_name,
            "layer": spec.layer,
            "datatype": spec.datatype,
            "role": spec.role,
            "drawing_present": (
                spec.layer,
                spec.datatype,
            )
            in all_direct_specs,
            "pin_datatype_present": (
                spec.layer,
                PIN_DATATYPE,
            )
            in all_direct_specs,
            "label_datatype_present": (
                spec.layer,
                LABEL_DATATYPE,
            )
            in all_direct_specs,
        }
    )

missing_layers = [
    str(record["name"])
    for record in audit_records
    if not bool(record["drawing_present"])
]
if missing_layers:
    raise AssertionError(f"Missing drawing layers: {missing_layers}")

display_records("ASAP7 process-layer audit", audit_records)

# -----------------------------------------------------------------------------
# Notebook code cell 13
# -----------------------------------------------------------------------------
def polygon_points(polygon: object) -> np.ndarray:
    """Return polygon points from a GDSTK polygon-like object."""
    if hasattr(polygon, "points"):
        return np.asarray(polygon.points, dtype=float)
    return np.asarray(polygon, dtype=float)


def get_polygons_for_spec(
    cell: gdstk.Cell,
    spec: LayerSpec,
    depth: int | None,
) -> list[np.ndarray]:
    """Extract transformed polygons for one layer specification."""
    polygons = cell.get_polygons(
        apply_repetitions=True,
        include_paths=True,
        depth=depth,
        layer=spec.layer,
        datatype=spec.datatype,
    )
    return [polygon_points(polygon) for polygon in polygons]


def render_cell_layers(
    cell: gdstk.Cell,
    layer_names: Sequence[str],
    title: str,
    depth: int | None,
    figure_size: tuple[float, float],
    alpha: float = PLOT_LAYER_ALPHA,
) -> None:
    """Render selected hierarchy layers in memory at the configured DPI."""
    figure = plt.figure(figsize=figure_size, dpi=PREVIEW_DPI)
    axes = figure.add_axes((0.07, 0.08, 0.72, 0.84))
    color_cycle = itertools.cycle(
        plt.rcParams["axes.prop_cycle"].by_key()["color"]
    )
    legend_handles: list[Patch] = []
    polygon_total = 0

    for layer_name in layer_names:
        spec = LAYERS[layer_name]
        polygons = get_polygons_for_spec(cell, spec, depth)
        polygon_total += len(polygons)
        if polygon_total > MAX_RENDERED_POLYGONS:
            raise RuntimeError(
                "Render polygon limit exceeded. Reduce hierarchy depth "
                "or the selected layer set."
            )
        if not polygons:
            continue

        color = next(color_cycle)
        collection = PolyCollection(
            polygons,
            facecolors=color,
            edgecolors=color,
            linewidths=0.10,
            alpha=alpha,
            rasterized=True,
        )
        axes.add_collection(collection)
        legend_handles.append(
            Patch(
                facecolor=color,
                edgecolor=color,
                alpha=alpha,
                label=layer_name,
            )
        )

    bounds = bounding_box_array(cell)
    span = bounds[1] - bounds[0]
    x_span = float(span[0])
    y_span = float(span[1])
    margin = 0.02 * max(x_span, y_span)
    axes.set_xlim(bounds[0, 0] - margin, bounds[1, 0] + margin)
    axes.set_ylim(bounds[0, 1] - margin, bounds[1, 1] + margin)
    axes.set_aspect("equal", adjustable="box")
    axes.set_xlabel("x (µm)")
    axes.set_ylabel("y (µm)")
    axes.set_title(f"{title}\n{polygon_total:,} polygons rendered")

    if legend_handles:
        axes.legend(
            handles=legend_handles,
            loc="upper left",
            bbox_to_anchor=(1.01, 1.0),
            frameon=False,
        )

    plt.show()

# -----------------------------------------------------------------------------
# Notebook code cell 14
# -----------------------------------------------------------------------------
render_cell_layers(
    top,
    FULL_VIEW_LAYERS,
    title=(
        "ASAP7 hierarchical 128 KiB SRAM "
        "(16 × srambank_256x4x64_6t122)"
    ),
    depth=2,
    figure_size=(12.0, 9.0),
    alpha=PLOT_LAYER_ALPHA,
)

# -----------------------------------------------------------------------------
# Notebook code cell 15
# -----------------------------------------------------------------------------
def reference_histogram(library: gdstk.Library) -> Counter[str]:
    """Count direct references to every cell in a library."""
    counts: Counter[str] = Counter()
    for cell in library.cells:
        for reference in cell.references:
            target = reference.cell
            if isinstance(target, gdstk.Cell):
                counts[target.name] += 1
    return counts


BITCELL_POSITIVE_TOKENS = (
    "bitcell",
    "bit_cell",
    "sramcell",
    "sram_cell",
    "6t122",
    "6t123",
    "6t112",
    "6t111",
)

BITCELL_EXCLUDED_TOKENS = (
    "tapcell",
    "welltap",
    "tap_",
    "_tap",
    "dummy",
    "filler",
    "fill_",
    "_fill",
    "blank",
    "decap",
    "boundary",
    "marker",
    "endcap",
    "spacer",
    "strap",
    "precharge",
    "sense",
    "driver",
    "decoder",
    "column_mux",
    "row_mux",
    "control",
    "ctrl",
    "clock",
    "buffer",
    "inverter",
    "nand",
    "nor",
    "latch",
    "flop",
)


def referenced_cells(
    cell: gdstk.Cell,
) -> list[gdstk.Cell]:
    """Return directly referenced GDSTK cells."""
    children: list[gdstk.Cell] = []
    for reference in cell.references:
        target = reference.cell
        if isinstance(target, gdstk.Cell):
            children.append(target)
    return children


def candidate_layer_numbers(
    cell: gdstk.Cell,
    max_depth: int = 2,
) -> set[int]:
    """Collect drawing layers from a cell and shallow descendants."""
    layers: set[int] = set()
    visited: set[str] = set()

    def visit(current: gdstk.Cell, depth: int) -> None:
        if current.name in visited:
            return
        visited.add(current.name)

        layers.update(
            int(polygon.layer)
            for polygon in current.polygons
        )
        for path in current.paths:
            layers.update(int(value) for value in path.layers)

        if depth >= max_depth:
            return

        for child in referenced_cells(current):
            visit(child, depth + 1)

    visit(cell, 0)
    return layers


def contains_any_token(
    name: str,
    tokens: Sequence[str],
) -> bool:
    """Return whether a lowercase name contains any token."""
    lowered = name.lower()
    return any(token in lowered for token in tokens)


def classify_zoom_cell(cell: gdstk.Cell) -> str:
    """Classify a selected zoom target as bitcell or support logic."""
    name_lower = cell.name.lower()

    if contains_any_token(
        name_lower,
        BITCELL_EXCLUDED_TOKENS,
    ):
        return "supporting cell"

    if contains_any_token(
        name_lower,
        BITCELL_POSITIVE_TOKENS,
    ):
        return "bitcell"

    return "supporting cell"


def select_bitcell_candidate(
    library: gdstk.Library,
) -> tuple[
    gdstk.Cell,
    str,
    str,
    list[dict[str, Any]],
]:
    """Select a genuine SRAM bitcell and reject helper structures."""
    reference_counts = reference_histogram(library)
    required_feol = {
        LAYERS["fin"].layer,
        LAYERS["Gate"].layer,
        LAYERS["Active"].layer,
    }

    scored: list[
        tuple[float, gdstk.Cell, str, str]
    ] = []

    for cell in library.cells:
        name_lower = cell.name.lower()

        if contains_any_token(
            name_lower,
            BITCELL_EXCLUDED_TOKENS,
        ):
            continue

        try:
            bounds = bounding_box_array(cell)
        except ValueError:
            continue

        span = bounds[1] - bounds[0]
        width = float(span[0])
        height = float(span[1])
        area = width * height

        if area < 1e-3 or area > 25.0:
            continue
        if width < 0.02 or height < 0.02:
            continue

        positive_tokens = [
            token
            for token in BITCELL_POSITIVE_TOKENS
            if token in name_lower
        ]
        name_score = 0.0

        token_weights = {
            "bitcell": 9000.0,
            "bit_cell": 9000.0,
            "sramcell": 8000.0,
            "sram_cell": 8000.0,
            "6t122": 5000.0,
            "6t123": 4500.0,
            "6t112": 4500.0,
            "6t111": 4000.0,
        }
        for token in positive_tokens:
            name_score += token_weights[token]

        layers = candidate_layer_numbers(cell, max_depth=2)
        has_feol = required_feol.issubset(layers)
        layer_score = 2500.0 if has_feol else 0.0

        if not positive_tokens:
            continue
        if not has_feol:
            continue

        repeat_count = reference_counts[cell.name]
        repeat_score = 35.0 * math.log1p(repeat_count)
        compactness_score = min(800.0, 0.5 / area)
        hierarchy_score = (
            250.0 if cell.references else 0.0
        )

        score = (
            name_score
            + layer_score
            + repeat_score
            + compactness_score
            + hierarchy_score
        )

        reason = (
            f"name tokens={positive_tokens}, "
            f"name score={name_score:.0f}, "
            f"FEOL score={layer_score:.0f}, "
            f"references={repeat_count}, "
            f"area={area:.6g} µm²"
        )
        scored.append(
            (
                score,
                cell,
                reason,
                "bitcell",
            )
        )

    scored.sort(key=lambda item: item[0], reverse=True)

    diagnostic_records = [
        {
            "rank": rank,
            "cell": cell.name,
            "score": score,
            "classification": classification,
            "bbox_um": bbox_size(cell),
            "reason": reason,
        }
        for rank, (
            score,
            cell,
            reason,
            classification,
        ) in enumerate(scored[:15], start=1)
    ]

    if not scored:
        fallback_reason = (
            "No non-helper SRAM bitcell with FEOL geometry was "
            "resolved. Using the representative SRAM-VT standard "
            "cell as an explicitly labeled supporting-cell fallback."
        )
        fallback_records = [
            {
                "rank": 1,
                "cell": representative_std.name,
                "score": "",
                "classification": "supporting cell",
                "bbox_um": bbox_size(representative_std),
                "reason": fallback_reason,
            }
        ]
        return (
            representative_std,
            fallback_reason,
            "supporting cell",
            fallback_records,
        )

    best_score, best_cell, best_reason, classification = scored[0]

    if best_score < 7000.0:
        fallback_reason = (
            f"Best bitcell score {best_score:.2f} was below the "
            "acceptance threshold. Using the representative SRAM-VT "
            "standard cell as a supporting-cell fallback."
        )
        diagnostic_records.insert(
            0,
            {
                "rank": 0,
                "cell": representative_std.name,
                "score": "",
                "classification": "supporting cell",
                "bbox_um": bbox_size(representative_std),
                "reason": fallback_reason,
            },
        )
        return (
            representative_std,
            fallback_reason,
            "supporting cell",
            diagnostic_records,
        )

    return (
        best_cell,
        best_reason,
        classification,
        diagnostic_records,
    )


(
    zoom_cell,
    zoom_selection_reason,
    zoom_cell_classification,
    zoom_candidate_records,
) = select_bitcell_candidate(sram_library)

display_records(
    "SRAM zoom-cell candidate audit",
    zoom_candidate_records,
)

print(f"Zoom cell: {zoom_cell.name}")
print(f"Classification: {zoom_cell_classification}")
print(f"Selection basis: {zoom_selection_reason}")
print(f"Bounding box: {bbox_size(zoom_cell)} µm")

# -----------------------------------------------------------------------------
# Notebook code cell 16
# -----------------------------------------------------------------------------
if zoom_cell in sram_library.cells:
    zoom_scale = bank_scale
else:
    zoom_scale = standard_cell_scale

zoom_lower, _, _ = cell_lower_left_and_size(
    zoom_cell,
    zoom_scale,
)
zoom_render_cell = gdstk.Cell("ZOOM_RENDER_CELL")
zoom_render_cell.add(
    gdstk.Reference(
        zoom_cell,
        origin=(-zoom_lower[0], -zoom_lower[1]),
        magnification=zoom_scale,
    )
)

if zoom_cell_classification == "bitcell":
    zoom_title = (
        f"ASAP7 SRAM bitcell zoom: {zoom_cell.name}"
    )
else:
    zoom_title = (
        "ASAP7 SRAM supporting-cell zoom: "
        f"{zoom_cell.name}"
    )

render_cell_layers(
    zoom_render_cell,
    ZOOM_VIEW_LAYERS,
    title=zoom_title,
    depth=None,
    figure_size=(10.0, 8.0),
    alpha=PLOT_LAYER_ALPHA,
)
```

```python
# =============================================================================
# GRAPHVIZ SRAM CONNECTIVITY DIAGRAMS
# =============================================================================

# -----------------------------------------------------------------------------
# CONTROL KNOBS
# -----------------------------------------------------------------------------
GRAPHVIZ_ENGINE = "dot"
GRAPHVIZ_OUTPUT_FORMAT = "png"
GRAPHVIZ_DPI = 300

GLOBAL_GRAPH_SPLINES = "polyline"
DETAILED_GRAPH_SPLINES = "polyline"

GLOBAL_GRAPH_SIZE_INCHES = "14,10!"
DETAILED_GRAPH_SIZE_INCHES = "13,9!"
NOTEBOOK_DISPLAY_WIDTH_PX = 1500

SHOW_GLOBAL_BANK_MAP = True
SHOW_DETAILED_BANK_PATH = True

# Use the existing SRAM notebook configuration, with standalone defaults.
GRAPH_BANK_ROWS = int(globals().get("BANK_ARRAY_ROWS", 4))
GRAPH_BANK_COLUMNS = int(globals().get("BANK_ARRAY_COLUMNS", 4))
GRAPH_WORD_ROWS = int(globals().get("BANK_WORD_ROWS", 256))
GRAPH_COLUMN_MUX = int(globals().get("BANK_COLUMN_MUX", 4))
GRAPH_WORD_BITS = int(globals().get("BANK_WORD_BITS", 64))
GRAPH_BANK_BYTES = int(
    globals().get(
        "bank_bytes",
        GRAPH_WORD_ROWS
        * GRAPH_COLUMN_MUX
        * GRAPH_WORD_BITS
        // 8,
    )
)

# -----------------------------------------------------------------------------
# INSTALL GRAPHVIZ
# -----------------------------------------------------------------------------
import math
import shutil
import struct
import subprocess
import sys
from typing import Final

if shutil.which("dot") is None:
    subprocess.run(
        ["apt-get", "update", "-qq"],
        check=True,
    )
    subprocess.run(
        ["apt-get", "install", "-y", "-qq", "graphviz"],
        check=True,
    )

subprocess.run(
    [sys.executable, "-m", "pip", "install", "--quiet", "uv"],
    check=True,
)
subprocess.run(
    [
        "uv",
        "pip",
        "install",
        "--python",
        sys.executable,
        "--quiet",
        "graphviz",
    ],
    check=True,
)

from graphviz import Digraph
from IPython.display import Image, display

# -----------------------------------------------------------------------------
# DERIVED MEMORY ORGANIZATION
# -----------------------------------------------------------------------------
GRAPH_BANK_COUNT: Final[int] = (
    GRAPH_BANK_ROWS * GRAPH_BANK_COLUMNS
)
GRAPH_WORDS_PER_BANK: Final[int] = (
    GRAPH_WORD_ROWS * GRAPH_COLUMN_MUX
)
GRAPH_BANK_SELECT_BITS: Final[int] = math.ceil(
    math.log2(GRAPH_BANK_COUNT)
)
GRAPH_ROW_ADDRESS_BITS: Final[int] = math.ceil(
    math.log2(GRAPH_WORD_ROWS)
)
GRAPH_COLUMN_ADDRESS_BITS: Final[int] = math.ceil(
    math.log2(GRAPH_COLUMN_MUX)
)
GRAPH_LOCAL_ADDRESS_BITS: Final[int] = (
    GRAPH_ROW_ADDRESS_BITS
    + GRAPH_COLUMN_ADDRESS_BITS
)
GRAPH_TOTAL_ADDRESS_BITS: Final[int] = (
    GRAPH_BANK_SELECT_BITS
    + GRAPH_LOCAL_ADDRESS_BITS
)

assert GRAPH_BANK_COUNT > 0
assert GRAPH_WORD_ROWS > 0
assert GRAPH_COLUMN_MUX > 0
assert GRAPH_WORD_BITS > 0
assert GRAPH_BANK_BYTES > 0
assert GRAPH_BANK_BYTES * GRAPH_BANK_COUNT == 128 * 1024


def make_bank_html_label(bank_index: int) -> str:
    """Return a compact HTML label for one SRAM bank."""
    bank_kib = GRAPH_BANK_BYTES / 1024

    return f"""<
    <TABLE BORDER="1"
           CELLBORDER="1"
           CELLSPACING="0"
           CELLPADDING="4">
        <TR>
            <TD COLSPAN="2">
                <B>BANK {bank_index:02d}</B><BR/>
                {bank_kib:g} KiB
            </TD>
        </TR>
        <TR>
            <TD PORT="in">
                A[{GRAPH_LOCAL_ADDRESS_BITS - 1}:0],
                SEL[{bank_index}], CTRL, CLK<BR/>
                DIN[{GRAPH_WORD_BITS - 1}:0], VDD/VSS
            </TD>
            <TD PORT="out">
                DOUT[{GRAPH_WORD_BITS - 1}:0]
            </TD>
        </TR>
        <TR>
            <TD COLSPAN="2">
                {GRAPH_WORD_ROWS} ×
                {GRAPH_COLUMN_MUX} ×
                {GRAPH_WORD_BITS} bits
            </TD>
        </TR>
    </TABLE>
    >"""


def make_global_input_label() -> str:
    """Return the global interface label."""
    return f"""<
    <TABLE BORDER="1"
           CELLBORDER="1"
           CELLSPACING="0"
           CELLPADDING="5">
        <TR>
            <TD COLSPAN="5"><B>GLOBAL SRAM INTERFACE</B></TD>
        </TR>
        <TR>
            <TD PORT="addr">
                ADDRESS<BR/>
                A[{GRAPH_TOTAL_ADDRESS_BITS - 1}:0]
            </TD>
            <TD PORT="ctrl">CE / WE / OE / BYTE EN</TD>
            <TD PORT="clk">CLK / RESET</TD>
            <TD PORT="din">
                DIN[{GRAPH_WORD_BITS - 1}:0]
            </TD>
            <TD PORT="pwr">VDD / VSS</TD>
        </TR>
    </TABLE>
    >"""


def build_global_bank_graph() -> Digraph:
    """Build a compact bank-level connectivity diagram."""
    graph = Digraph(
        name="ASAP7_SRAM_128KiB_Global_Connectivity",
        engine=GRAPHVIZ_ENGINE,
        format=GRAPHVIZ_OUTPUT_FORMAT,
    )

    graph.attr(
        rankdir="TB",
        splines=GLOBAL_GRAPH_SPLINES,
        compound="true",
        newrank="true",
        concentrate="false",
        dpi=str(GRAPHVIZ_DPI),
        size=GLOBAL_GRAPH_SIZE_INCHES,
        ratio="compress",
        nodesep="0.18",
        ranksep="0.34",
        pad="0.08",
        margin="0.02",
        bgcolor="white",
        label=(
            "ASAP7 128 KiB SRAM Connectivity\\n"
            f"{GRAPH_BANK_ROWS} × {GRAPH_BANK_COLUMNS} banks, "
            f"{GRAPH_BANK_BYTES // 1024} KiB per bank"
        ),
        labelloc="t",
        fontsize="17",
        fontname="DejaVu Sans",
    )

    graph.attr(
        "node",
        shape="box",
        style="rounded",
        fontsize="8.5",
        margin="0.08",
        fontname="DejaVu Sans",
    )
    graph.attr(
        "edge",
        fontsize="7.5",
        arrowsize="0.58",
        fontname="DejaVu Sans",
    )

    # -------------------------------------------------------------------------
    # GLOBAL INPUT AND DISTRIBUTION FABRIC
    # -------------------------------------------------------------------------
    graph.node(
        "global_inputs",
        label=make_global_input_label(),
        shape="plain",
    )
    graph.node(
        "bank_decoder",
        (
            "GLOBAL BANK DECODER\\n"
            f"{GRAPH_BANK_SELECT_BITS}-to-{GRAPH_BANK_COUNT}"
        ),
    )
    graph.node(
        "global_fabric",
        (
            "GLOBAL DISTRIBUTION FABRIC\\n"
            f"A[{GRAPH_LOCAL_ADDRESS_BITS - 1}:0], "
            "SEL[15:0], CTRL, CLK, "
            f"DIN[{GRAPH_WORD_BITS - 1}:0], VDD/VSS"
        ),
    )
    graph.node(
        "global_read_mux",
        (
            "GLOBAL READ MULTIPLEXER\\n"
            f"{GRAPH_BANK_COUNT}:1 × {GRAPH_WORD_BITS} bits"
        ),
    )
    graph.node(
        "read_data_bus",
        f"READ DATA BUS\\nDOUT[{GRAPH_WORD_BITS - 1}:0]",
    )

    graph.edge(
        "global_inputs:addr:s",
        "bank_decoder:n",
        label=(
            f"A[{GRAPH_TOTAL_ADDRESS_BITS - 1}:"
            f"{GRAPH_LOCAL_ADDRESS_BITS}]"
        ),
        weight="12",
    )
    graph.edge(
        "global_inputs:addr:s",
        "global_fabric:n",
        label=f"A[{GRAPH_LOCAL_ADDRESS_BITS - 1}:0]",
        weight="10",
    )
    graph.edge(
        "global_inputs:ctrl:s",
        "global_fabric:n",
        label="CTRL",
        weight="8",
    )
    graph.edge(
        "global_inputs:clk:s",
        "global_fabric:n",
        label="CLK / RESET",
        weight="8",
    )
    graph.edge(
        "global_inputs:din:s",
        "global_fabric:n",
        label=f"DIN[{GRAPH_WORD_BITS - 1}:0]",
        weight="8",
    )
    graph.edge(
        "global_inputs:pwr:s",
        "global_fabric:n",
        label="VDD / VSS",
        style="dashed",
        weight="4",
    )
    graph.edge(
        "bank_decoder:s",
        "global_fabric:n",
        label=f"SEL[{GRAPH_BANK_COUNT - 1}:0]",
        weight="12",
    )

    # -------------------------------------------------------------------------
    # FOUR PHYSICAL BANK ROWS
    # -------------------------------------------------------------------------
    row_fabrics: list[str] = []
    row_muxes: list[str] = []

    for row in range(GRAPH_BANK_ROWS):
        row_fabric = f"row_fabric_{row}"
        row_mux = f"row_read_mux_{row}"
        row_fabrics.append(row_fabric)
        row_muxes.append(row_mux)

        with graph.subgraph(name=f"cluster_bank_row_{row}") as row_graph:
            row_graph.attr(
                label=f"BANK ROW {row}",
                style="rounded",
                margin="8",
                fontsize="10",
                fontname="DejaVu Sans",
            )

            with row_graph.subgraph(
                name=f"rank_bank_row_{row}"
            ) as rank_graph:
                rank_graph.attr(rank="same")

                rank_graph.node(
                    row_fabric,
                    (
                        f"ROW {row} BUS TAP\\n"
                        "ADDR / SELECT / CTRL / CLK / "
                        "WRITE DATA / POWER"
                    ),
                )

                bank_ids: list[str] = []
                for column in range(GRAPH_BANK_COLUMNS):
                    bank_index = (
                        row * GRAPH_BANK_COLUMNS + column
                    )
                    bank_id = f"bank_{bank_index:02d}"
                    bank_ids.append(bank_id)

                    rank_graph.node(
                        bank_id,
                        label=make_bank_html_label(bank_index),
                        shape="plain",
                        group=f"bank_column_{column}",
                    )

                rank_graph.node(
                    row_mux,
                    (
                        f"ROW {row} READ MUX\\n"
                        f"{GRAPH_BANK_COLUMNS}:1 × "
                        f"{GRAPH_WORD_BITS} bits"
                    ),
                )

            for column, bank_id in enumerate(bank_ids):
                bank_index = (
                    row * GRAPH_BANK_COLUMNS + column
                )
                row_graph.edge(
                    row_fabric,
                    f"{bank_id}:in:w",
                    label=f"B{bank_index:02d} input bundle",
                    weight="8",
                )
                row_graph.edge(
                    f"{bank_id}:out:e",
                    row_mux,
                    label=f"B{bank_index:02d} DOUT",
                    weight="8",
                )

        # Rank placement is controlled by invisible vertical chains below.
        graph.edge(
            "global_fabric",
            row_fabric,
            label=f"row {row} distribution bundle",
            constraint="false",
        )
        graph.edge(
            row_mux,
            "global_read_mux",
            label=f"row {row} read data",
            constraint="false",
        )

    # Enforce a stable top-to-bottom row order without drawing extra lines.
    graph.edge(
        "global_fabric",
        row_fabrics[0],
        style="invis",
        weight="100",
    )
    for upper, lower in zip(
        row_fabrics[:-1],
        row_fabrics[1:],
    ):
        graph.edge(
            upper,
            lower,
            style="invis",
            weight="100",
        )
    graph.edge(
        row_muxes[-1],
        "global_read_mux",
        style="invis",
        weight="100",
    )

    graph.edge(
        "global_read_mux",
        "read_data_bus",
        label=f"DOUT[{GRAPH_WORD_BITS - 1}:0]",
        weight="12",
    )

    return graph


def build_detailed_bank_graph() -> Digraph:
    """Build a compact internal read/write path for one SRAM bank."""
    graph = Digraph(
        name="ASAP7_SRAM_8KiB_Bank_Internal_Connectivity",
        engine=GRAPHVIZ_ENGINE,
        format=GRAPHVIZ_OUTPUT_FORMAT,
    )

    graph.attr(
        rankdir="TB",
        splines=DETAILED_GRAPH_SPLINES,
        compound="true",
        newrank="true",
        concentrate="false",
        dpi=str(GRAPHVIZ_DPI),
        size=DETAILED_GRAPH_SIZE_INCHES,
        ratio="compress",
        nodesep="0.24",
        ranksep="0.42",
        pad="0.08",
        margin="0.02",
        bgcolor="white",
        label=(
            "Representative ASAP7 8 KiB SRAM Bank\\n"
            "Address, read, write, timing, storage, and power connections"
        ),
        labelloc="t",
        fontsize="17",
        fontname="DejaVu Sans",
    )

    graph.attr(
        "node",
        shape="box",
        style="rounded",
        fontsize="9",
        margin="0.09",
        fontname="DejaVu Sans",
    )
    graph.attr(
        "edge",
        fontsize="8",
        arrowsize="0.62",
        fontname="DejaVu Sans",
    )

    # -------------------------------------------------------------------------
    # EXTERNAL INTERFACE
    # -------------------------------------------------------------------------
    with graph.subgraph(
        name="cluster_external_interface"
    ) as external:
        external.attr(
            label="External bank interface",
            style="rounded",
            margin="8",
            fontsize="10",
        )

        with external.subgraph(
            name="rank_external_interface"
        ) as external_rank:
            external_rank.attr(rank="same")
            external_rank.node("bank_select", "BANK SELECT")
            external_rank.node(
                "local_address",
                (
                    "LOCAL ADDRESS\\n"
                    f"A[{GRAPH_LOCAL_ADDRESS_BITS - 1}:0]"
                ),
            )
            external_rank.node(
                "control",
                "CONTROL\\nCE / WE / OE / BYTE EN",
            )
            external_rank.node("clock", "CLK / RESET")
            external_rank.node(
                "data_in",
                f"WRITE DATA\\nDIN[{GRAPH_WORD_BITS - 1}:0]",
            )

    # -------------------------------------------------------------------------
    # ADDRESS AND TIMING DECODE
    # -------------------------------------------------------------------------
    with graph.subgraph(
        name="cluster_decode"
    ) as decode:
        decode.attr(
            label="Address and timing decode",
            style="rounded",
            margin="8",
            fontsize="10",
        )

        decode.node(
            "bank_enable",
            "BANK ENABLE\\nchip-select qualification",
        )
        decode.node(
            "row_decoder",
            (
                "ROW DECODER\\n"
                f"{GRAPH_ROW_ADDRESS_BITS}-to-{GRAPH_WORD_ROWS}"
            ),
        )
        decode.node(
            "column_decoder",
            (
                "COLUMN DECODER\\n"
                f"{GRAPH_COLUMN_ADDRESS_BITS}-to-"
                f"{GRAPH_COLUMN_MUX}"
            ),
        )
        decode.node(
            "timing_control",
            (
                "LOCAL TIMING CONTROL\\n"
                "precharge / evaluate / write pulse"
            ),
        )
        decode.node(
            "wordline_driver",
            (
                "WORDLINE DRIVERS\\n"
                f"WL[{GRAPH_WORD_ROWS - 1}:0]"
            ),
        )

    # -------------------------------------------------------------------------
    # STORAGE CORE
    # -------------------------------------------------------------------------
    with graph.subgraph(
        name="cluster_storage"
    ) as storage:
        storage.attr(
            label="6T122 storage core",
            style="rounded",
            margin="8",
            fontsize="10",
        )

        with storage.subgraph(
            name="rank_storage_core"
        ) as storage_rank:
            storage_rank.attr(rank="same")
            storage_rank.node(
                "precharge",
                "BITLINE PRECHARGE\\nBL / BLB equalization",
            )
            storage_rank.node(
                "bitcell_array",
                (
                    "6T122 BITCELL ARRAY\\n"
                    f"{GRAPH_WORD_ROWS} rows × "
                    f"{GRAPH_WORD_BITS * GRAPH_COLUMN_MUX} columns\\n"
                    f"{GRAPH_BANK_BYTES // 1024} KiB"
                ),
            )
            storage_rank.node(
                "column_mux",
                (
                    f"{GRAPH_COLUMN_MUX}:1 COLUMN MUX\\n"
                    f"{GRAPH_WORD_BITS}-bit selected word"
                ),
            )

    # -------------------------------------------------------------------------
    # READ AND WRITE PERIPHERY
    # -------------------------------------------------------------------------
    with graph.subgraph(
        name="cluster_read_write"
    ) as read_write:
        read_write.attr(
            label="Read and write periphery",
            style="rounded",
            margin="8",
            fontsize="10",
        )

        with read_write.subgraph(
            name="rank_read_write"
        ) as rw_rank:
            rw_rank.attr(rank="same")
            rw_rank.node(
                "write_mask",
                "WRITE MASK\\nbyte enable",
            )
            rw_rank.node(
                "write_driver",
                (
                    "WRITE DRIVERS\\n"
                    f"{GRAPH_WORD_BITS}-bit differential drive"
                ),
            )
            rw_rank.node(
                "sense_amplifier",
                (
                    "SENSE AMPLIFIERS\\n"
                    f"{GRAPH_WORD_BITS}-bit differential sensing"
                ),
            )
            rw_rank.node(
                "output_register",
                (
                    "OUTPUT REGISTER\\n"
                    f"{GRAPH_WORD_BITS}-bit"
                ),
            )
            rw_rank.node(
                "data_out",
                f"READ DATA\\nDOUT[{GRAPH_WORD_BITS - 1}:0]",
            )

    # -------------------------------------------------------------------------
    # POWER DISTRIBUTION
    # -------------------------------------------------------------------------
    with graph.subgraph(
        name="cluster_power"
    ) as power:
        power.attr(
            label="Power distribution",
            style="rounded",
            margin="8",
            fontsize="10",
        )

        with power.subgraph(
            name="rank_power_distribution"
        ) as power_rank:
            power_rank.attr(rank="same")
            power_rank.node("power_input", "VDD / VSS")
            power_rank.node(
                "decode_power_tap",
                "DECODE POWER TAP",
            )
            power_rank.node(
                "array_power_tap",
                "ARRAY POWER TAP",
            )
            power_rank.node(
                "io_power_tap",
                "I/O POWER TAP",
            )

    # -------------------------------------------------------------------------
    # ADDRESS AND CONTROL CONNECTIONS
    # -------------------------------------------------------------------------
    graph.edge(
        "bank_select",
        "bank_enable",
        label="SEL",
        weight="10",
    )
    graph.edge(
        "control",
        "bank_enable",
        label="CE",
        weight="8",
    )
    graph.edge(
        "local_address",
        "row_decoder",
        label=(
            f"row A[{GRAPH_LOCAL_ADDRESS_BITS - 1}:"
            f"{GRAPH_COLUMN_ADDRESS_BITS}]"
        ),
        weight="10",
    )
    graph.edge(
        "local_address",
        "column_decoder",
        label=(
            f"column A[{GRAPH_COLUMN_ADDRESS_BITS - 1}:0]"
        ),
        weight="10",
    )
    graph.edge(
        "clock",
        "timing_control",
        label="CLK / RESET",
        weight="8",
    )
    graph.edge(
        "control",
        "timing_control",
        label="WE / OE",
        weight="8",
    )
    graph.edge(
        "bank_enable",
        "row_decoder",
        label="enable",
        weight="8",
    )
    graph.edge(
        "bank_enable",
        "column_decoder",
        label="enable",
        weight="8",
    )
    graph.edge(
        "bank_enable",
        "timing_control",
        label="enable",
        weight="8",
    )
    graph.edge(
        "row_decoder",
        "wordline_driver",
        label="one-hot row select",
        weight="10",
    )
    graph.edge(
        "wordline_driver",
        "bitcell_array",
        label=f"WL[{GRAPH_WORD_ROWS - 1}:0]",
        weight="10",
    )
    graph.edge(
        "column_decoder",
        "column_mux",
        label="column select",
        weight="10",
    )

    # -------------------------------------------------------------------------
    # READ PATH
    # -------------------------------------------------------------------------
    graph.edge(
        "timing_control",
        "precharge",
        label="precharge enable",
        weight="8",
    )
    graph.edge(
        "precharge",
        "bitcell_array",
        label="BL / BLB precharge",
        weight="10",
    )
    graph.edge(
        "bitcell_array",
        "column_mux",
        label="BL / BLB read path",
        weight="12",
    )
    graph.edge(
        "column_mux",
        "sense_amplifier",
        label=f"{GRAPH_WORD_BITS} differential pairs",
        weight="12",
    )
    graph.edge(
        "sense_amplifier",
        "output_register",
        label=f"Q[{GRAPH_WORD_BITS - 1}:0]",
        weight="10",
    )
    graph.edge(
        "output_register",
        "data_out",
        label=f"DOUT[{GRAPH_WORD_BITS - 1}:0]",
        weight="10",
    )

    # -------------------------------------------------------------------------
    # WRITE PATH
    # -------------------------------------------------------------------------
    graph.edge(
        "data_in",
        "write_mask",
        label=f"DIN[{GRAPH_WORD_BITS - 1}:0]",
        weight="10",
    )
    graph.edge(
        "control",
        "write_mask",
        label="byte enables",
        weight="8",
    )
    graph.edge(
        "write_mask",
        "write_driver",
        label="masked write data",
        weight="10",
    )
    graph.edge(
        "timing_control",
        "write_driver",
        label="write pulse",
        weight="8",
    )
    graph.edge(
        "write_driver",
        "column_mux",
        label="BL / BLB write drive",
        weight="10",
    )
    graph.edge(
        "column_mux",
        "bitcell_array",
        label="selected write columns",
        constraint="false",
        minlen="1",
    )

    # -------------------------------------------------------------------------
    # COMPACT POWER TREE
    # -------------------------------------------------------------------------
    graph.edge(
        "power_input",
        "decode_power_tap",
        style="dashed",
        arrowhead="none",
        weight="6",
    )
    graph.edge(
        "power_input",
        "array_power_tap",
        style="dashed",
        arrowhead="none",
        weight="6",
    )
    graph.edge(
        "power_input",
        "io_power_tap",
        style="dashed",
        arrowhead="none",
        weight="6",
    )

    for powered_block in (
        "bank_enable",
        "row_decoder",
        "column_decoder",
        "timing_control",
        "wordline_driver",
    ):
        graph.edge(
            "decode_power_tap",
            powered_block,
            style="dashed",
            arrowhead="none",
            constraint="false",
        )

    for powered_block in (
        "precharge",
        "bitcell_array",
        "column_mux",
    ):
        graph.edge(
            "array_power_tap",
            powered_block,
            style="dashed",
            arrowhead="none",
            constraint="false",
        )

    for powered_block in (
        "write_mask",
        "write_driver",
        "sense_amplifier",
        "output_register",
    ):
        graph.edge(
            "io_power_tap",
            powered_block,
            style="dashed",
            arrowhead="none",
            constraint="false",
        )

    # Force the power-distribution cluster beneath the signal path.
    graph.edge(
        "data_out",
        "power_input",
        style="invis",
        weight="100",
        minlen="1",
    )

    return graph


def png_dimensions(png_bytes: bytes) -> tuple[int, int]:
    """Return the width and height encoded in a PNG byte stream."""
    png_signature = b"\x89PNG\r\n\x1a\n"
    if not png_bytes.startswith(png_signature):
        raise ValueError("The byte stream is not a valid PNG.")
    return struct.unpack(">II", png_bytes[16:24])


def display_graphviz_png(
    graph: Digraph,
    display_width_px: int = NOTEBOOK_DISPLAY_WIDTH_PX,
) -> bytes:
    """Render and display a 300 DPI Graphviz PNG entirely in memory."""
    graph.graph_attr["dpi"] = str(GRAPHVIZ_DPI)
    graph.graph_attr["bgcolor"] = "white"

    png_bytes = graph.pipe(
        format=GRAPHVIZ_OUTPUT_FORMAT,
        engine=GRAPHVIZ_ENGINE,
    )

    width_px, height_px = png_dimensions(png_bytes)

    display(
        Image(
            data=png_bytes,
            format="png",
            width=min(display_width_px, width_px),
        )
    )

    print(
        f"PNG raster: {width_px} × {height_px} pixels "
        f"at {GRAPHVIZ_DPI} DPI; "
        f"{len(png_bytes):,} bytes retained in memory."
    )

    return png_bytes


# -----------------------------------------------------------------------------
# RENDER BOTH 300 DPI PNG DIAGRAMS DIRECTLY IN NOTEBOOK MEMORY
# -----------------------------------------------------------------------------
global_bank_png: bytes | None = None
detailed_bank_png: bytes | None = None

if SHOW_GLOBAL_BANK_MAP:
    global_bank_graph = build_global_bank_graph()
    global_bank_png = display_graphviz_png(global_bank_graph)

if SHOW_DETAILED_BANK_PATH:
    detailed_bank_graph = build_detailed_bank_graph()
    detailed_bank_png = display_graphviz_png(detailed_bank_graph)

rendered_diagram_count = sum(
    png_data is not None
    for png_data in (
        global_bank_png,
        detailed_bank_png,
    )
)

print(
    "Rendered "
    f"{rendered_diagram_count} in-memory Graphviz PNG diagram(s): "
    f"{GRAPH_BANK_COUNT} banks, "
    f"{GRAPH_BANK_BYTES // 1024} KiB per bank, "
    f"{GRAPH_WORD_BITS}-bit data path."
)
```

```python
"""Build and render a hierarchical 128 KiB ASAP7 SRAM array.

The zoom selector rejects tap cells, dummy cells, fillers, decaps, and other helper structures before identifying a genuine SRAM bitcell or an explicitly labeled supporting-cell fallback.
"""

# -----------------------------------------------------------------------------
# Notebook code cell 1
# -----------------------------------------------------------------------------
# Install only the geometry packages.
#
# NumPy and Matplotlib are intentionally left at the versions supplied by
# the active Colab runtime. PHIDL 1.7.2 supports NumPy 2.x.

import subprocess
import sys

GEOMETRY_REQUIREMENTS = (
    "gdstk==1.0.0",
    "gdspy==1.6.13",
    "phidl==1.7.2",
)

subprocess.run(
    [sys.executable, "-m", "pip", "install", "--quiet", "uv"],
    check=True,
)
subprocess.run(
    [
        "uv",
        "pip",
        "install",
        "--python",
        sys.executable,
        "--upgrade",
        "--no-deps",
        *GEOMETRY_REQUIREMENTS,
    ],
    check=True,
)

probe_code = '''
import gdspy
import gdstk
import matplotlib
import numpy
import phidl
from phidl import Device

print("numpy", numpy.__version__)
print("matplotlib", matplotlib.__version__)
print("gdstk", gdstk.__version__)
print("gdspy", gdspy.__version__)
print("phidl", getattr(phidl, "__version__", "unknown"))
print("PHIDL Device", Device.__name__)
'''

probe = subprocess.run(
    [sys.executable, "-c", probe_code],
    check=True,
    capture_output=True,
    text=True,
)
print(probe.stdout)

# -----------------------------------------------------------------------------
# Notebook code cell 2
# -----------------------------------------------------------------------------
# -------------------------------------------------------------------------
# CAPACITY AND FLOORPLAN
# -------------------------------------------------------------------------
TARGET_BYTES = 128 * 1024
BANK_MACRO_NAME = "srambank_256x4x64_6t122"
BANK_WORD_ROWS = 256
BANK_COLUMN_MUX = 4
BANK_WORD_BITS = 64

BANK_ARRAY_ROWS = 4
BANK_ARRAY_COLUMNS = 4
INTERBANK_GAP_UM = 8.0
BANK_MARGIN_UM = 2.0
PERIPHERY_ROWS = 8

# -------------------------------------------------------------------------
# SOURCE ASSETS
# -------------------------------------------------------------------------
DOWNLOAD_DIRECTORY = "/content/asap7_sources"
OUTPUT_DIRECTORY = "/content/asap7_sram_128kb_output"

SRAM_GDS_URL = (
    "https://raw.githubusercontent.com/The-OpenROAD-Project/"
    "asap7_sram_0p0/main/gds/srambank_64b.gds"
)
SRAM_MACRO_LEF_URL = (
    "https://raw.githubusercontent.com/The-OpenROAD-Project/"
    "asap7_sram_0p0/main/generated/LEF/"
    "srambank_256x4x64_6t122.lef"
)
SRAM_STANDARD_CELL_GDS_URL = (
    "https://raw.githubusercontent.com/The-OpenROAD-Project/"
    "asap7sc7p5t_28/main/GDS/"
    "asap7sc7p5t_28_SRAM_220121a.gds"
)
SRAM_STANDARD_CELL_LEF_URL = (
    "https://raw.githubusercontent.com/The-OpenROAD-Project/"
    "asap7sc7p5t_28/main/LEF/"
    "asap7sc7p5t_28_SRAM_1x_220121a.lef"
)

USE_LOCAL_SOURCE_FILES = False
LOCAL_SRAM_GDS = ""
LOCAL_SRAM_MACRO_LEF = ""
LOCAL_STANDARD_CELL_GDS = ""
LOCAL_STANDARD_CELL_LEF = ""

# -------------------------------------------------------------------------
# STANDARD-CELL PERIPHERY
# -------------------------------------------------------------------------
PREFERRED_STANDARD_CELLS = (
    "DFFHQNx1_ASAP7_75t_SRAM",
    "BUFx4_ASAP7_75t_SRAM",
    "NAND2x1_ASAP7_75t_SRAM",
    "INVx1_ASAP7_75t_SRAM",
    "TAPCELL_ASAP7_75t_SRAM",
    "DECAPx4_ASAP7_75t_SRAM",
)

# -------------------------------------------------------------------------
# TOP-LEVEL ROUTING AND EXPORT
# -------------------------------------------------------------------------
TOP_M8_WIDTH_UM = 0.80
TOP_M9_WIDTH_UM = 0.80
TOP_BUS_WIDTH_UM = 0.32
WRITE_GDS = True
WRITE_OASIS = True
OUTPUT_GDS_NAME = "ASAP7_SRAM_128KB.gds"
OUTPUT_OAS_NAME = "ASAP7_SRAM_128KB.oas"

# -------------------------------------------------------------------------
# IN-MEMORY VISUALIZATION
# -------------------------------------------------------------------------
PREVIEW_DPI = 600
PLOT_LAYER_ALPHA = 0.50
FULL_VIEW_LAYERS = (
    "BOUNDARY",
    "M6",
    "M7",
    "M8",
    "V8",
    "M9",
)
ZOOM_VIEW_LAYERS = (
    "well",
    "fin",
    "Gate",
    "GCut",
    "Active",
    "Nselect",
    "Pselect",
    "SDT",
    "LIG",
    "LISD",
    "V0",
    "M1",
    "V1",
    "M2",
    "V2",
    "M3",
)
MAX_RENDERED_POLYGONS = 300_000

# -----------------------------------------------------------------------------
# Notebook code cell 3
# -----------------------------------------------------------------------------
import difflib
import hashlib
import itertools
import math
import re
import sys
import time
import urllib.error
import urllib.request
from collections import Counter
from dataclasses import dataclass
from html import escape
from pathlib import Path
from typing import Any, Iterable, Mapping, Sequence

import gdstk
import matplotlib.pyplot as plt
import numpy as np
import phidl
from IPython.display import HTML, display
from matplotlib.collections import PolyCollection
from matplotlib.patches import Patch
from phidl import Device


def format_table_value(value: Any) -> str:
    """Format a value for compact notebook tables."""
    if isinstance(value, float):
        return f"{value:.8g}"
    if isinstance(value, Path):
        return str(value)
    return str(value)


def display_records(
    title: str,
    records: Sequence[Mapping[str, Any]],
) -> None:
    """Display records as HTML without a DataFrame dependency."""
    if not records:
        display(HTML(f"<h4>{escape(title)}</h4><p>No records.</p>"))
        return

    headers: list[str] = []
    for record in records:
        for key in record:
            if key not in headers:
                headers.append(key)

    header_html = "".join(
        f"<th>{escape(str(header))}</th>" for header in headers
    )
    body_rows = []
    for record in records:
        cells = "".join(
            "<td>"
            f"{escape(format_table_value(record.get(header, '')))}"
            "</td>"
            for header in headers
        )
        body_rows.append(f"<tr>{cells}</tr>")

    display(
        HTML(
            f"<h4>{escape(title)}</h4>"
            "<div style='overflow-x:auto'>"
            "<table style='border-collapse:collapse;font-size:12px'>"
            f"<thead><tr>{header_html}</tr></thead>"
            f"<tbody>{''.join(body_rows)}</tbody>"
            "</table></div>"
            "<style>"
            "table th, table td {"
            "border:1px solid #bbb;padding:4px 7px;text-align:left;"
            "}"
            "table th {font-weight:600;background:#f2f2f2;}"
            "</style>"
        )
    )


plt.rcParams.update(
    {
        "figure.dpi": PREVIEW_DPI,
        "savefig.dpi": PREVIEW_DPI,
        "font.size": 9,
        "axes.titlesize": 11,
        "axes.labelsize": 9,
    }
)

SOURCE_DIR = Path(DOWNLOAD_DIRECTORY)
RESULT_DIR = Path(OUTPUT_DIRECTORY)
SOURCE_DIR.mkdir(parents=True, exist_ok=True)
RESULT_DIR.mkdir(parents=True, exist_ok=True)

display_records(
    "Runtime environment",
    [
        {"component": "Python", "version": sys.version.split()[0]},
        {"component": "NumPy", "version": np.__version__},
        {
            "component": "Matplotlib",
            "version": plt.matplotlib.__version__,
        },
        {"component": "GDSTK", "version": gdstk.__version__},
        {
            "component": "PHIDL",
            "version": getattr(phidl, "__version__", "unknown"),
        },
    ],
)

# -----------------------------------------------------------------------------
# Notebook code cell 4
# -----------------------------------------------------------------------------
@dataclass(frozen=True)
class LayerSpec:
    """ASAP7 stream-layer specification."""

    name: str
    layer: int
    datatype: int
    role: str


LAYERS = {
    "well": LayerSpec("well", 1, 0, "FEOL well"),
    "fin": LayerSpec("fin", 2, 0, "Fin"),
    "Gate": LayerSpec("Gate", 7, 0, "Gate"),
    "Dummy": LayerSpec("Dummy", 8, 0, "Dummy gate"),
    "GCut": LayerSpec("GCut", 10, 0, "Gate cut"),
    "Active": LayerSpec("Active", 11, 0, "Active"),
    "Nselect": LayerSpec("Nselect", 12, 0, "NMOS select"),
    "Pselect": LayerSpec("Pselect", 13, 0, "PMOS select"),
    "LIG": LayerSpec("LIG", 16, 0, "MOL gate interconnect"),
    "LISD": LayerSpec(
        "LISD",
        17,
        0,
        "MOL source-drain interconnect",
    ),
    "V0": LayerSpec("V0", 18, 0, "MOL-to-M1 via"),
    "M1": LayerSpec("M1", 19, 0, "Metal 1"),
    "M2": LayerSpec("M2", 20, 0, "Metal 2"),
    "V1": LayerSpec("V1", 21, 0, "Via 1"),
    "V2": LayerSpec("V2", 25, 0, "Via 2"),
    "M3": LayerSpec("M3", 30, 0, "Metal 3"),
    "V3": LayerSpec("V3", 35, 0, "Via 3"),
    "M4": LayerSpec("M4", 40, 0, "Metal 4"),
    "V4": LayerSpec("V4", 45, 0, "Via 4"),
    "M5": LayerSpec("M5", 50, 0, "Metal 5"),
    "V5": LayerSpec("V5", 55, 0, "Via 5"),
    "M6": LayerSpec("M6", 60, 0, "Metal 6"),
    "V6": LayerSpec("V6", 65, 0, "Via 6"),
    "M7": LayerSpec("M7", 70, 0, "Metal 7"),
    "V7": LayerSpec("V7", 75, 0, "Via 7"),
    "M8": LayerSpec("M8", 80, 0, "Metal 8"),
    "V8": LayerSpec("V8", 85, 0, "Via 8"),
    "SDT": LayerSpec("SDT", 88, 0, "Source-drain trench"),
    "M9": LayerSpec("M9", 90, 0, "Metal 9"),
    "SRAMDRC": LayerSpec("SRAMDRC", 99, 0, "SRAM rule marker"),
    "BOUNDARY": LayerSpec("BOUNDARY", 100, 0, "Cell boundary"),
    "TEXT": LayerSpec("TEXT", 101, 0, "Text"),
    "SRAMVT": LayerSpec("SRAMVT", 110, 0, "SRAM threshold marker"),
}

PROCESS_LAYER_ORDER = (
    "well",
    "fin",
    "Gate",
    "Dummy",
    "GCut",
    "Active",
    "Nselect",
    "Pselect",
    "SDT",
    "LIG",
    "LISD",
    "V0",
    "M1",
    "V1",
    "M2",
    "V2",
    "M3",
    "V3",
    "M4",
    "V4",
    "M5",
    "V5",
    "M6",
    "V6",
    "M7",
    "V7",
    "M8",
    "V8",
    "M9",
    "SRAMDRC",
    "SRAMVT",
    "BOUNDARY",
)

PIN_DATATYPE = 251
LABEL_DATATYPE = 2
NET_DATATYPE = 3
BLOCKAGE_DATATYPE = 4

layer_records = sorted(
    [
        {
            "name": item.name,
            "layer": item.layer,
            "datatype": item.datatype,
            "role": item.role,
        }
        for item in LAYERS.values()
    ],
    key=lambda record: (
        int(record["layer"]),
        int(record["datatype"]),
    ),
)
display_records("ASAP7 stream-layer map", layer_records)

# -----------------------------------------------------------------------------
# Notebook code cell 5
# -----------------------------------------------------------------------------
def sha256_file(path: Path) -> str:
    """Return a SHA-256 digest for a local file."""
    digest = hashlib.sha256()
    with path.open("rb") as stream:
        for chunk in iter(lambda: stream.read(1 << 20), b""):
            digest.update(chunk)
    return digest.hexdigest()


def validate_download_payload(
    destination: Path,
    payload: bytes,
) -> None:
    """Reject HTML error pages and malformed GDS/LEF downloads."""
    suffix = destination.suffix.lower()
    if suffix == ".gds":
        if not payload.startswith(b"\x00\x06\x00\x02"):
            preview = payload[:120].decode(
                "utf-8",
                errors="replace",
            )
            raise ValueError(
                "The downloaded GDSII file has an invalid header. "
                f"Payload preview: {preview!r}"
            )
    elif suffix == ".lef":
        if b"MACRO" not in payload[:200_000].upper():
            preview = payload[:120].decode(
                "utf-8",
                errors="replace",
            )
            raise ValueError(
                "The downloaded LEF file does not contain a MACRO block. "
                f"Payload preview: {preview!r}"
            )


def download_with_retries(
    url: str,
    destination: Path,
    minimum_bytes: int,
    attempts: int = 5,
    timeout_seconds: int = 120,
) -> Path:
    """Download a file atomically with validation and retries."""
    if destination.exists():
        payload = destination.read_bytes()
        try:
            if len(payload) >= minimum_bytes:
                validate_download_payload(destination, payload)
                return destination
        except ValueError:
            destination.unlink(missing_ok=True)

    temporary = destination.with_suffix(destination.suffix + ".part")
    request = urllib.request.Request(
        url,
        headers={
            "User-Agent": "ASAP7-PHIDL-GDSTK-Notebook/2.0",
            "Accept": "application/octet-stream,text/plain,*/*",
        },
    )

    last_error: Exception | None = None
    for attempt in range(1, attempts + 1):
        try:
            with urllib.request.urlopen(
                request,
                timeout=timeout_seconds,
            ) as response:
                payload = response.read()

            if len(payload) < minimum_bytes:
                raise ValueError(
                    f"Downloaded {len(payload):,} bytes from {url}; "
                    f"expected at least {minimum_bytes:,}."
                )

            validate_download_payload(destination, payload)
            temporary.write_bytes(payload)
            temporary.replace(destination)
            return destination
        except (
            urllib.error.URLError,
            TimeoutError,
            ValueError,
            OSError,
        ) as error:
            last_error = error
            temporary.unlink(missing_ok=True)
            if attempt < attempts:
                time.sleep(min(30, 2**attempt))

    raise RuntimeError(f"Unable to download {url}") from last_error


if USE_LOCAL_SOURCE_FILES:
    source_paths = {
        "sram_gds": Path(LOCAL_SRAM_GDS),
        "sram_lef": Path(LOCAL_SRAM_MACRO_LEF),
        "std_gds": Path(LOCAL_STANDARD_CELL_GDS),
        "std_lef": Path(LOCAL_STANDARD_CELL_LEF),
    }
else:
    source_paths = {
        "sram_gds": download_with_retries(
            SRAM_GDS_URL,
            SOURCE_DIR / "srambank_64b.gds",
            minimum_bytes=500_000,
        ),
        "sram_lef": download_with_retries(
            SRAM_MACRO_LEF_URL,
            SOURCE_DIR / "srambank_256x4x64_6t122.lef",
            minimum_bytes=100_000,
        ),
        "std_gds": download_with_retries(
            SRAM_STANDARD_CELL_GDS_URL,
            SOURCE_DIR / "asap7sc7p5t_28_SRAM_220121a.gds",
            minimum_bytes=1_000_000,
        ),
        "std_lef": download_with_retries(
            SRAM_STANDARD_CELL_LEF_URL,
            SOURCE_DIR / "asap7sc7p5t_28_SRAM_1x_220121a.lef",
            minimum_bytes=100_000,
        ),
    }

for key, path in source_paths.items():
    if not path.is_file():
        raise FileNotFoundError(f"Missing source file for {key}: {path}")

source_records = [
    {
        "asset": key,
        "path": str(path),
        "bytes": path.stat().st_size,
        "sha256": sha256_file(path),
    }
    for key, path in source_paths.items()
]
display_records("Downloaded source assets", source_records)

# -----------------------------------------------------------------------------
# Notebook code cell 6
# -----------------------------------------------------------------------------
def cell_index(library: gdstk.Library) -> dict[str, gdstk.Cell]:
    """Index all cells in a GDSTK library by name."""
    return {cell.name: cell for cell in library.cells}


def reference_target_name(reference: gdstk.Reference) -> str:
    """Return the referenced cell name across GDSTK API variants."""
    target = reference.cell
    if isinstance(target, gdstk.Cell):
        return target.name
    return str(target)


def library_top_level_cells(
    library: gdstk.Library,
) -> list[gdstk.Cell]:
    """Return top-level cells with a compatibility fallback."""
    try:
        top_cells = list(library.top_level())
        if top_cells:
            return top_cells
    except (AttributeError, TypeError):
        pass

    referenced_names = {
        reference_target_name(reference)
        for cell in library.cells
        for reference in cell.references
    }
    top_cells = [
        cell for cell in library.cells
        if cell.name not in referenced_names
    ]
    return top_cells or list(library.cells)


def bounding_box_array(cell: gdstk.Cell) -> np.ndarray:
    """Return a validated 2 by 2 floating-point bounding-box array.

    GDSTK 1.0 returns ``((min_x, min_y), (max_x, max_y))``. Converting
    centrally avoids tuple-indexing and tuple-scaling failures.
    """
    raw_bounds = cell.bounding_box()
    if raw_bounds is None:
        raise ValueError(f"Cell {cell.name} has no geometry.")

    bounds = np.asarray(raw_bounds, dtype=float)
    if bounds.shape != (2, 2):
        raise ValueError(
            f"Unexpected bounding-box shape for {cell.name}: "
            f"{bounds.shape!r} from {raw_bounds!r}"
        )
    if not np.all(np.isfinite(bounds)):
        raise ValueError(
            f"Non-finite bounding box for {cell.name}: {bounds.tolist()}"
        )
    if np.any(bounds[1] < bounds[0]):
        raise ValueError(
            f"Inverted bounding box for {cell.name}: {bounds.tolist()}"
        )
    return bounds


def bbox_size(cell: gdstk.Cell) -> tuple[float, float]:
    """Return a cell bounding-box width and height."""
    bounds = bounding_box_array(cell)
    span = bounds[1] - bounds[0]
    return float(span[0]), float(span[1])


def cell_area(cell: gdstk.Cell) -> float:
    """Return the bounding-box area, or zero for an empty cell."""
    try:
        width, height = bbox_size(cell)
    except ValueError:
        return 0.0
    return max(0.0, width * height)


def normalized_name(name: str) -> str:
    """Normalize a layout cell name for fuzzy matching."""
    return re.sub(r"[^a-z0-9]+", "", name.lower())


def select_cell(
    library: gdstk.Library,
    preferred_name: str,
    required_tokens: Sequence[str] = (),
    prefer_sram: bool = False,
) -> tuple[gdstk.Cell, str]:
    """Resolve a cell by exact name, constrained fuzzy match, and score."""
    index = cell_index(library)
    if preferred_name in index:
        return index[preferred_name], "exact name match"

    preferred_lower = preferred_name.lower()
    folded_matches = [
        cell
        for name, cell in index.items()
        if name.lower() == preferred_lower
    ]
    if folded_matches:
        return folded_matches[0], "case-insensitive exact match"

    tokens = tuple(token.lower() for token in required_tokens if token)
    candidates = [
        cell
        for cell in library.cells
        if all(token in cell.name.lower() for token in tokens)
    ]
    if not candidates:
        raise KeyError(
            f"No cell contains all required tokens {tokens!r} "
            f"for preferred name {preferred_name!r}."
        )

    preferred_normalized = normalized_name(preferred_name)
    scored: list[tuple[float, gdstk.Cell, str]] = []
    for cell in candidates:
        cell_lower = cell.name.lower()
        similarity = difflib.SequenceMatcher(
            None,
            preferred_normalized,
            normalized_name(cell.name),
        ).ratio()
        score = 1000.0 * similarity
        if prefer_sram and "sram" in cell_lower:
            score += 500.0
        if "filler" in cell_lower or "blank" in cell_lower:
            score -= 750.0
        score += min(100.0, math.log1p(cell_area(cell)))
        scored.append(
            (
                score,
                cell,
                (
                    f"fuzzy score={score:.2f}, "
                    f"similarity={similarity:.3f}"
                ),
            )
        )

    scored.sort(key=lambda item: item[0], reverse=True)
    _, selected, reason = scored[0]
    return selected, reason


def parse_lef_macro_size(
    lef_text: str,
    macro_name: str,
) -> tuple[float, float] | None:
    """Extract a LEF macro SIZE statement in micrometres."""
    escaped = re.escape(macro_name)
    block_match = re.search(
        rf"\bMACRO\s+{escaped}\b(.*?)\bEND\s+{escaped}\b",
        lef_text,
        flags=re.IGNORECASE | re.DOTALL,
    )
    if block_match is None:
        return None
    size_match = re.search(
        r"\bSIZE\s+([0-9.eE+-]+)\s+BY\s+"
        r"([0-9.eE+-]+)\s*;",
        block_match.group(1),
        flags=re.IGNORECASE,
    )
    if size_match is None:
        return None
    return float(size_match.group(1)), float(size_match.group(2))


def parse_all_lef_macro_sizes(
    lef_text: str,
) -> dict[str, tuple[float, float]]:
    """Parse every LEF MACRO name and SIZE statement."""
    macro_sizes: dict[str, tuple[float, float]] = {}
    macro_pattern = re.compile(
        r"\bMACRO\s+([^\s]+)(.*?)\bEND\s+\1\b",
        flags=re.IGNORECASE | re.DOTALL,
    )
    size_pattern = re.compile(
        r"\bSIZE\s+([0-9.eE+-]+)\s+BY\s+"
        r"([0-9.eE+-]+)\s*;",
        flags=re.IGNORECASE,
    )
    for macro_match in macro_pattern.finditer(lef_text):
        size_match = size_pattern.search(macro_match.group(2))
        if size_match is None:
            continue
        macro_sizes[macro_match.group(1)] = (
            float(size_match.group(1)),
            float(size_match.group(2)),
        )
    return macro_sizes


def match_lef_macro_size(
    lef_sizes: Mapping[str, tuple[float, float]],
    cell_name: str,
) -> tuple[tuple[float, float] | None, str | None]:
    """Find the closest LEF macro size for a GDS cell name."""
    if cell_name in lef_sizes:
        return lef_sizes[cell_name], cell_name

    normalized_target = normalized_name(cell_name)
    if not lef_sizes:
        return None, None

    scored = sorted(
        (
            difflib.SequenceMatcher(
                None,
                normalized_target,
                normalized_name(name),
            ).ratio(),
            name,
        )
        for name in lef_sizes
    )
    similarity, name = scored[-1]
    if similarity < 0.60:
        return None, None
    return lef_sizes[name], name


def infer_reference_scale(
    cell: gdstk.Cell,
    lef_size: tuple[float, float] | None,
    tolerance: float = 0.03,
) -> float:
    """Infer a 0.25x, 1x, or 4x GDS-to-LEF scale."""
    if lef_size is None:
        return 1.0

    gds_width, gds_height = bbox_size(cell)
    ratios = np.asarray(
        [
            lef_size[0] / gds_width,
            lef_size[1] / gds_height,
        ],
        dtype=float,
    )
    mean_ratio = float(np.mean(ratios))
    mismatch = float(
        np.max(np.abs(ratios / mean_ratio - 1.0))
    )
    if mismatch > tolerance:
        raise ValueError(
            f"Inconsistent X/Y scale for {cell.name}: "
            f"{ratios.tolist()}"
        )

    canonical = min(
        (0.25, 1.0, 4.0),
        key=lambda value: abs(value - mean_ratio),
    )
    if abs(canonical - mean_ratio) / canonical <= tolerance:
        return canonical
    return mean_ratio


def safe_reference_scale(
    cell: gdstk.Cell,
    lef_size: tuple[float, float] | None,
) -> tuple[float, str]:
    """Infer scaling and fall back to native GDS units on mismatch."""
    try:
        scale = infer_reference_scale(cell, lef_size)
        return scale, "LEF/GDS scale inferred"
    except ValueError as error:
        return 1.0, f"native GDS scale used: {error}"


def select_bank_source(
    library: gdstk.Library,
) -> tuple[gdstk.Cell, str, list[dict[str, Any]]]:
    """Select the best physical bank source from a partial SRAM GDS."""
    top_cells = library_top_level_cells(library)
    top_names = {cell.name for cell in top_cells}
    scored: list[tuple[float, gdstk.Cell, str]] = []

    for cell in library.cells:
        area = cell_area(cell)
        if area <= 0.0:
            continue

        name_lower = cell.name.lower()
        score = 0.0
        reasons: list[str] = []

        if cell.name == BANK_MACRO_NAME:
            score += 100_000.0
            reasons.append("exact target name")
        if "srambank" in name_lower:
            score += 12_000.0
            reasons.append("srambank token")
        elif "sram" in name_lower and "bank" in name_lower:
            score += 8_000.0
            reasons.append("SRAM bank tokens")
        elif "bank" in name_lower:
            score += 4_000.0
            reasons.append("bank token")

        if "64bit" in name_lower or "64b" in name_lower:
            score += 3_000.0
            reasons.append("64-bit token")
        if "6t122" in name_lower:
            score += 2_000.0
            reasons.append("6T122 token")
        if cell.name in top_names:
            score += 2_500.0
            reasons.append("top-level cell")

        if any(
            token in name_lower
            for token in (
                "filler",
                "blank",
                "decap",
                "tapcell",
                "marker",
                "boundary",
            )
        ):
            score -= 20_000.0
            reasons.append("filler/helper penalty")

        score += min(2_000.0, 100.0 * math.log1p(area))
        reasons.append(f"bbox area={area:.6g} µm²")
        scored.append((score, cell, ", ".join(reasons)))

    if not scored:
        raise RuntimeError("The SRAM GDS contains no non-empty cells.")

    scored.sort(key=lambda item: item[0], reverse=True)
    selected_score, selected_cell, selected_reason = scored[0]

    diagnostics = [
        {
            "rank": rank,
            "cell": cell.name,
            "score": score,
            "bbox_um": bbox_size(cell),
            "top_level": cell.name in top_names,
            "reason": reason,
        }
        for rank, (score, cell, reason) in enumerate(
            scored[:20],
            start=1,
        )
    ]

    return (
        selected_cell,
        (
            f"selected score={selected_score:.2f}: "
            f"{selected_reason}"
        ),
        diagnostics,
    )


def choose_partial_source_scale(
    source_cell: gdstk.Cell,
    target_size: tuple[float, float],
) -> tuple[float, str]:
    """Choose native, 0.25x, or fit-to-wrapper scaling."""
    raw_width, raw_height = bbox_size(source_cell)
    target_width, target_height = target_size

    if (
        raw_width <= 1.02 * target_width
        and raw_height <= 1.02 * target_height
    ):
        return 1.0, "native source scale fits the LEF wrapper"

    quarter_width = 0.25 * raw_width
    quarter_height = 0.25 * raw_height
    if (
        quarter_width <= 1.02 * target_width
        and quarter_height <= 1.02 * target_height
    ):
        return 0.25, "4x APR source detected and reduced to 0.25x"

    fit_scale = 0.96 * min(
        target_width / raw_width,
        target_height / raw_height,
    )
    if fit_scale <= 0.0:
        raise ValueError("Unable to derive a positive source scale.")
    return fit_scale, "source scaled uniformly to fit the LEF wrapper"


def build_lef_sized_bank_wrapper(
    source_cell: gdstk.Cell,
    target_size: tuple[float, float],
) -> tuple[gdstk.Cell, float, int, int, str]:
    """Build a LEF-sized macro wrapper from a physical bank fragment."""
    target_width, target_height = target_size
    source_scale, scale_reason = choose_partial_source_scale(
        source_cell,
        target_size,
    )

    source_bounds = bounding_box_array(source_cell)
    source_lower = source_bounds[0] * source_scale
    source_upper = source_bounds[1] * source_scale
    source_width = float(source_upper[0] - source_lower[0])
    source_height = float(source_upper[1] - source_lower[1])

    margin = min(
        1.0,
        0.02 * min(target_width, target_height),
    )
    usable_width = max(source_width, target_width - 2.0 * margin)
    usable_height = max(source_height, target_height - 2.0 * margin)

    columns = max(1, int(usable_width // source_width))
    rows = max(1, int(usable_height // source_height))

    max_references = 256
    while columns * rows > max_references:
        if columns >= rows and columns > 1:
            columns -= 1
        elif rows > 1:
            rows -= 1
        else:
            break

    grid_width = columns * source_width
    grid_height = rows * source_height
    x_gap = max(0.0, (target_width - grid_width) / (columns + 1))
    y_gap = max(0.0, (target_height - grid_height) / (rows + 1))

    wrapper = gdstk.Cell(f"{BANK_MACRO_NAME}_COMPOSED")
    wrapper.add(
        gdstk.rectangle(
            (0.0, 0.0),
            (target_width, target_height),
            layer=LAYERS["BOUNDARY"].layer,
            datatype=LAYERS["BOUNDARY"].datatype,
        )
    )

    for row in range(rows):
        for column in range(columns):
            x_position = (
                x_gap
                + column * (source_width + x_gap)
                - source_lower[0]
            )
            y_position = (
                y_gap
                + row * (source_height + y_gap)
                - source_lower[1]
            )
            wrapper.add(
                gdstk.Reference(
                    source_cell,
                    origin=(x_position, y_position),
                    magnification=source_scale,
                )
            )

    wrapper.add(
        gdstk.Label(
            (
                f"{BANK_MACRO_NAME} | physical proxy from "
                f"{source_cell.name}"
            ),
            (0.5 * target_width, target_height - 0.25),
            anchor="n",
            layer=LAYERS["TEXT"].layer,
            texttype=LABEL_DATATYPE,
        )
    )
    return wrapper, source_scale, columns, rows, scale_reason


def standard_cell_family_token(preferred_name: str) -> str:
    """Derive a broad family token from an ASAP7 preferred name."""
    base = preferred_name.split("_ASAP7", maxsplit=1)[0]
    family = re.sub(r"x[0-9].*$", "", base, flags=re.IGNORECASE)
    return family or base


def resolve_standard_cells(
    library: gdstk.Library,
    preferred_names: Sequence[str],
) -> tuple[list[gdstk.Cell], list[dict[str, str]]]:
    """Resolve a diverse, SRAM-biased set of periphery cells."""
    resolved: list[gdstk.Cell] = []
    records: list[dict[str, str]] = []
    seen_names: set[str] = set()

    for preferred_name in preferred_names:
        family = standard_cell_family_token(preferred_name)
        try:
            cell, reason = select_cell(
                library,
                preferred_name,
                required_tokens=(family,),
                prefer_sram=True,
            )
        except KeyError as error:
            records.append(
                {
                    "preferred": preferred_name,
                    "resolved": "",
                    "status": str(error),
                }
            )
            continue

        if cell.name not in seen_names:
            resolved.append(cell)
            seen_names.add(cell.name)
        records.append(
            {
                "preferred": preferred_name,
                "resolved": cell.name,
                "status": reason,
            }
        )

    if not resolved:
        fallback_candidates = [
            cell
            for cell in library.cells
            if "sram" in cell.name.lower()
            and "filler" not in cell.name.lower()
            and cell_area(cell) > 0.0
        ]
        fallback_candidates.sort(
            key=lambda cell: cell_area(cell),
            reverse=True,
        )
        resolved.extend(fallback_candidates[:8])
        records.extend(
            {
                "preferred": "dynamic SRAM fallback",
                "resolved": cell.name,
                "status": "selected from non-filler SRAM cells",
            }
            for cell in resolved
        )

    if not resolved:
        raise RuntimeError(
            "No usable SRAM-threshold standard cells were resolved."
        )

    return resolved, records


sram_library = gdstk.read_gds(str(source_paths["sram_gds"]))
standard_cell_library = gdstk.read_gds(
    str(source_paths["std_gds"])
)

sram_lef_text = source_paths["sram_lef"].read_text(
    encoding="utf-8",
    errors="replace",
)
std_lef_text = source_paths["std_lef"].read_text(
    encoding="utf-8",
    errors="replace",
)
sram_lef_sizes = parse_all_lef_macro_sizes(sram_lef_text)
std_lef_sizes = parse_all_lef_macro_sizes(std_lef_text)

exact_bank_cell = cell_index(sram_library).get(BANK_MACRO_NAME)
bank_lef_size = sram_lef_sizes.get(BANK_MACRO_NAME)

if exact_bank_cell is not None:
    bank_macro = exact_bank_cell
    bank_scale, bank_scale_reason = safe_reference_scale(
        bank_macro,
        bank_lef_size,
    )
    bank_source_cell = exact_bank_cell
    bank_source_scale = bank_scale
    bank_source_columns = 1
    bank_source_rows = 1
    bank_source_reason = "exact GDS macro found"
    bank_macro_is_wrapper = False
    bank_candidate_records = [
        {
            "rank": 1,
            "cell": exact_bank_cell.name,
            "score": 100000,
            "bbox_um": bbox_size(exact_bank_cell),
            "top_level": (
                exact_bank_cell
                in library_top_level_cells(sram_library)
            ),
            "reason": "exact target name",
        }
    ]
else:
    (
        bank_source_cell,
        bank_source_reason,
        bank_candidate_records,
    ) = select_bank_source(sram_library)

    if bank_lef_size is None:
        source_width, source_height = bbox_size(bank_source_cell)
        bank_lef_size = (source_width, source_height)
        print(
            "The target LEF SIZE was unavailable. The wrapper uses the "
            "selected source-cell dimensions."
        )

    (
        bank_macro,
        bank_source_scale,
        bank_source_columns,
        bank_source_rows,
        bank_scale_reason,
    ) = build_lef_sized_bank_wrapper(
        bank_source_cell,
        bank_lef_size,
    )
    bank_scale = 1.0
    bank_macro_is_wrapper = True

standard_cells, standard_cell_records = resolve_standard_cells(
    standard_cell_library,
    PREFERRED_STANDARD_CELLS,
)

representative_std = standard_cells[0]
(
    representative_lef_size,
    representative_lef_name,
) = match_lef_macro_size(
    std_lef_sizes,
    representative_std.name,
)
(
    standard_cell_scale,
    standard_scale_reason,
) = safe_reference_scale(
    representative_std,
    representative_lef_size,
)

display_records(
    "Physical SRAM-bank source candidates",
    bank_candidate_records,
)
display_records(
    "Resolved standard-cell periphery",
    standard_cell_records,
)

resolution_records = [
    {
        "type": "Logical bank target",
        "cell": BANK_MACRO_NAME,
        "gds_cell": bank_macro.name,
        "gds_bbox_um": bbox_size(bank_macro),
        "lef_size_um": bank_lef_size,
        "reference_scale": bank_scale,
        "status": (
            "exact GDS macro"
            if not bank_macro_is_wrapper
            else "LEF-sized composed wrapper"
        ),
    },
    {
        "type": "Physical source hierarchy",
        "cell": bank_source_cell.name,
        "gds_cell": bank_source_cell.name,
        "gds_bbox_um": bbox_size(bank_source_cell),
        "lef_size_um": "",
        "reference_scale": bank_source_scale,
        "status": (
            f"{bank_source_reason}; {bank_scale_reason}; "
            f"grid={bank_source_columns}×{bank_source_rows}"
        ),
    },
    {
        "type": "Representative standard cell",
        "cell": representative_std.name,
        "gds_cell": representative_std.name,
        "gds_bbox_um": bbox_size(representative_std),
        "lef_size_um": representative_lef_size,
        "reference_scale": standard_cell_scale,
        "status": (
            f"{standard_scale_reason}; "
            f"matched LEF={representative_lef_name}"
        ),
    },
]
display_records(
    "Resolved physical hierarchy and scaling",
    resolution_records,
)

print("Resolved standard cells:")
for cell in standard_cells:
    print(f"  {cell.name}")

# GDSTK 1.0 tuple-return self-test.
_bbox_probe = gdstk.Cell("BBOX_API_PROBE")
_bbox_probe.add(gdstk.rectangle((0.0, 0.0), (2.0, 3.0)))
_bbox_probe_bounds = bounding_box_array(_bbox_probe)
assert _bbox_probe_bounds.shape == (2, 2)
assert np.allclose(_bbox_probe_bounds, [[0.0, 0.0], [2.0, 3.0]])
assert bbox_size(_bbox_probe) == (2.0, 3.0)

# -----------------------------------------------------------------------------
# Notebook code cell 7
# -----------------------------------------------------------------------------
bank_bits = BANK_WORD_ROWS * BANK_COLUMN_MUX * BANK_WORD_BITS
bank_bytes = bank_bits // 8
bank_count = BANK_ARRAY_ROWS * BANK_ARRAY_COLUMNS

assert bank_bits % 8 == 0
assert bank_count * bank_bytes == TARGET_BYTES
assert TARGET_BYTES == 128 * 1024

capacity_records = [
    {"quantity": "Bits per bank", "value": bank_bits},
    {"quantity": "Bytes per bank", "value": bank_bytes},
    {"quantity": "KiB per bank", "value": bank_bytes / 1024},
    {"quantity": "Banks", "value": bank_count},
    {
        "quantity": "Total bytes",
        "value": bank_count * bank_bytes,
    },
    {
        "quantity": "Total KiB",
        "value": bank_count * bank_bytes / 1024,
    },
]
display_records("Exact memory-capacity check", capacity_records)

# -----------------------------------------------------------------------------
# Notebook code cell 8
# -----------------------------------------------------------------------------
def phidl_rectangle(
    device: Device,
    x: float,
    y: float,
    width: float,
    height: float,
    layer_name: str,
) -> None:
    """Add an axis-aligned rectangle to a PHIDL device."""
    spec = LAYERS[layer_name]
    points = np.asarray(
        [
            [x, y],
            [x + width, y],
            [x + width, y + height],
            [x, y + height],
        ],
        dtype=float,
    )
    device.add_polygon(points, layer=(spec.layer, spec.datatype))


def phidl_to_gdstk(device: Device, cell_name: str) -> gdstk.Cell:
    """Flatten PHIDL polygons into a GDSTK cell without disk I/O."""
    output = gdstk.Cell(cell_name)
    polygons_by_spec = device.get_polygons(by_spec=True)
    for raw_spec, polygon_list in polygons_by_spec.items():
        if isinstance(raw_spec, tuple):
            layer, datatype = int(raw_spec[0]), int(raw_spec[1])
        else:
            layer, datatype = int(raw_spec), 0
        for points in polygon_list:
            output.add(
                gdstk.Polygon(
                    np.asarray(points, dtype=float),
                    layer=layer,
                    datatype=datatype,
                )
            )
    return output


def transformed_bbox(
    cell: gdstk.Cell,
    magnification: float,
) -> tuple[np.ndarray, np.ndarray]:
    """Return scaled lower-left and upper-right cell bounds."""
    bounds = bounding_box_array(cell)
    return bounds[0] * magnification, bounds[1] * magnification


def add_boundary_geometry(
    cell: gdstk.Cell,
    width: float,
    height: float,
    name: str,
) -> gdstk.Cell:
    """Add a PHIDL boundary reference and return its source cell."""
    boundary_device = Device(f"{name}_PHIDL")
    phidl_rectangle(
        boundary_device,
        0.0,
        0.0,
        width,
        height,
        "BOUNDARY",
    )
    boundary_cell = phidl_to_gdstk(
        boundary_device,
        f"{name}_BOUNDARY",
    )
    cell.add(gdstk.Reference(boundary_cell))
    text_spec = LAYERS["TEXT"]
    cell.add(
        gdstk.Label(
            name,
            (0.4, height - 0.4),
            anchor="nw",
            layer=text_spec.layer,
            texttype=LABEL_DATATYPE,
        )
    )
    return boundary_cell

# -----------------------------------------------------------------------------
# Notebook code cell 9
# -----------------------------------------------------------------------------
def cell_lower_left_and_size(
    cell: gdstk.Cell,
    scale: float,
) -> tuple[np.ndarray, float, float]:
    """Return scaled lower-left coordinates, width, and height."""
    lower, upper = transformed_bbox(cell, scale)
    return lower, float(upper[0] - lower[0]), float(upper[1] - lower[1])


bank_lower, bank_width_um, bank_height_um = cell_lower_left_and_size(
    bank_macro,
    bank_scale,
)

std_dimensions = []
for cell in standard_cells:
    lower, width, height = cell_lower_left_and_size(
        cell,
        standard_cell_scale,
    )
    std_dimensions.append((cell, lower, width, height))

standard_cell_height_um = max(item[3] for item in std_dimensions)
periphery_height_um = PERIPHERY_ROWS * standard_cell_height_um
tile_width_um = bank_width_um + 2.0 * BANK_MARGIN_UM
tile_height_um = (
    bank_height_um
    + periphery_height_um
    + 3.0 * BANK_MARGIN_UM
)

bank_tile = gdstk.Cell("SRAM8K_BANK_TILE")
bank_origin = (
    BANK_MARGIN_UM - bank_lower[0],
    periphery_height_um
    + 2.0 * BANK_MARGIN_UM
    - bank_lower[1],
)
bank_tile.add(
    gdstk.Reference(
        bank_macro,
        origin=bank_origin,
        magnification=bank_scale,
    )
)

# Tile actual standard cells in eight compact rows below the macro.
for row in range(PERIPHERY_ROWS):
    x_position = BANK_MARGIN_UM
    y_position = BANK_MARGIN_UM + row * standard_cell_height_um
    cell_cycle = itertools.cycle(std_dimensions)
    placements = 0

    while x_position < tile_width_um - BANK_MARGIN_UM:
        cell, lower, width, height = next(cell_cycle)
        remaining = tile_width_um - BANK_MARGIN_UM - x_position
        if width > remaining:
            break
        reflect = bool(row % 2)
        if reflect:
            upper_y = lower[1] + height
            origin_y = y_position + upper_y
        else:
            origin_y = y_position - lower[1]
        bank_tile.add(
            gdstk.Reference(
                cell,
                origin=(
                    x_position - lower[0],
                    origin_y,
                ),
                magnification=standard_cell_scale,
                x_reflection=reflect,
            )
        )
        x_position += width
        placements += 1
        if placements > 10_000:
            raise RuntimeError("Periphery tiling failed to converge.")

bank_boundary_cell = add_boundary_geometry(
    bank_tile,
    tile_width_um,
    tile_height_um,
    "SRAM8K_BANK_TILE",
)


tile_records = [
    {
        "bank_macro": bank_macro.name,
        "macro_width_um": bank_width_um,
        "macro_height_um": bank_height_um,
        "periphery_height_um": periphery_height_um,
        "tile_width_um": tile_width_um,
        "tile_height_um": tile_height_um,
        "bytes": bank_bytes,
    }
]
display_records("8 KiB bank-tile dimensions", tile_records)

# -----------------------------------------------------------------------------
# Notebook code cell 10
# -----------------------------------------------------------------------------
def make_top_mesh(
    array_width: float,
    array_height: float,
    pitch_x: float,
    pitch_y: float,
) -> gdstk.Cell:
    """Create top-level M6-M9 routing with PHIDL."""
    mesh = Device("TOP_MESH_PHIDL")

    # M8 vertical supply trunks.
    vertical_x = sorted(
        {
            0.5 * TOP_M8_WIDTH_UM,
            array_width - 0.5 * TOP_M8_WIDTH_UM,
            *(
                column * pitch_x + 0.5 * tile_width_um
                for column in range(BANK_ARRAY_COLUMNS)
            ),
        }
    )
    for x_coordinate in vertical_x:
        phidl_rectangle(
            mesh,
            x_coordinate - 0.5 * TOP_M8_WIDTH_UM,
            0.0,
            TOP_M8_WIDTH_UM,
            array_height,
            "M8",
        )

    # M9 horizontal supply trunks.
    horizontal_y = sorted(
        {
            0.5 * TOP_M9_WIDTH_UM,
            array_height - 0.5 * TOP_M9_WIDTH_UM,
            *(
                row * pitch_y + 0.5 * tile_height_um
                for row in range(BANK_ARRAY_ROWS)
            ),
        }
    )
    for y_coordinate in horizontal_y:
        phidl_rectangle(
            mesh,
            0.0,
            y_coordinate - 0.5 * TOP_M9_WIDTH_UM,
            array_width,
            TOP_M9_WIDTH_UM,
            "M9",
        )

    # V8 landing arrays at M8-M9 intersections.
    via_size = min(TOP_M8_WIDTH_UM, TOP_M9_WIDTH_UM) * 0.55
    for x_coordinate in vertical_x:
        for y_coordinate in horizontal_y:
            phidl_rectangle(
                mesh,
                x_coordinate - 0.5 * via_size,
                y_coordinate - 0.5 * via_size,
                via_size,
                via_size,
                "V8",
            )

    # M6 horizontal address/control spines and M7 vertical data spines.
    for row in range(BANK_ARRAY_ROWS):
        y_coordinate = row * pitch_y + tile_height_um + 0.25 * INTERBANK_GAP_UM
        if y_coordinate < array_height:
            phidl_rectangle(
                mesh,
                0.0,
                y_coordinate - 0.5 * TOP_BUS_WIDTH_UM,
                array_width,
                TOP_BUS_WIDTH_UM,
                "M6",
            )

    for column in range(BANK_ARRAY_COLUMNS):
        x_coordinate = (
            column * pitch_x
            + tile_width_um
            + 0.25 * INTERBANK_GAP_UM
        )
        if x_coordinate < array_width:
            phidl_rectangle(
                mesh,
                x_coordinate - 0.5 * TOP_BUS_WIDTH_UM,
                0.0,
                TOP_BUS_WIDTH_UM,
                array_height,
                "M7",
            )

    return phidl_to_gdstk(mesh, "TOP_LEVEL_MESH")


def make_layer_coupon(
    columns: int = 4,
    block_width: float = 0.8,
    block_height: float = 0.5,
    gap: float = 0.35,
) -> tuple[gdstk.Cell, float, float]:
    """Create a layer-presence coupon for explicit stream-layer auditing."""
    coupon = Device("PROCESS_LAYER_COUPON_PHIDL")
    rows = math.ceil(len(PROCESS_LAYER_ORDER) / columns)
    cell_width = block_width + gap
    cell_height = block_height + gap

    for index, layer_name in enumerate(PROCESS_LAYER_ORDER):
        column = index % columns
        row = index // columns
        x_coordinate = column * cell_width
        y_coordinate = (rows - 1 - row) * cell_height
        phidl_rectangle(
            coupon,
            x_coordinate,
            y_coordinate,
            block_width,
            block_height,
            layer_name,
        )

    width = columns * cell_width
    height = rows * cell_height
    coupon_cell = phidl_to_gdstk(
        coupon,
        "PROCESS_LAYER_COUPON",
    )
    text_spec = LAYERS["TEXT"]
    for index, layer_name in enumerate(PROCESS_LAYER_ORDER):
        column = index % columns
        row = index // columns
        x_coordinate = column * cell_width
        y_coordinate = (rows - 1 - row) * cell_height
        coupon_cell.add(
            gdstk.Label(
                layer_name,
                (x_coordinate, y_coordinate + block_height + 0.04),
                anchor="sw",
                layer=text_spec.layer,
                texttype=LABEL_DATATYPE,
            )
        )
    return coupon_cell, width, height


pitch_x_um = tile_width_um + INTERBANK_GAP_UM
pitch_y_um = tile_height_um + INTERBANK_GAP_UM
array_width_um = (
    BANK_ARRAY_COLUMNS * tile_width_um
    + (BANK_ARRAY_COLUMNS - 1) * INTERBANK_GAP_UM
)
array_height_um = (
    BANK_ARRAY_ROWS * tile_height_um
    + (BANK_ARRAY_ROWS - 1) * INTERBANK_GAP_UM
)

top = gdstk.Cell("ASAP7_SRAM_128KB")
text_spec = LAYERS["TEXT"]

for row in range(BANK_ARRAY_ROWS):
    for column in range(BANK_ARRAY_COLUMNS):
        bank_index = row * BANK_ARRAY_COLUMNS + column
        origin = (column * pitch_x_um, row * pitch_y_um)
        top.add(gdstk.Reference(bank_tile, origin=origin))
        top.add(
            gdstk.Label(
                f"BANK_{bank_index:02d}",
                (
                    origin[0] + 0.5 * tile_width_um,
                    origin[1] + tile_height_um - 0.5,
                ),
                anchor="n",
                layer=text_spec.layer,
                texttype=LABEL_DATATYPE,
            )
        )

top_mesh = make_top_mesh(
    array_width_um,
    array_height_um,
    pitch_x_um,
    pitch_y_um,
)
top.add(gdstk.Reference(top_mesh))

layer_coupon, coupon_width_um, coupon_height_um = make_layer_coupon()
coupon_origin = (
    array_width_um + INTERBANK_GAP_UM,
    max(0.0, array_height_um - coupon_height_um),
)
top.add(gdstk.Reference(layer_coupon, origin=coupon_origin))

total_width_um = array_width_um + INTERBANK_GAP_UM + coupon_width_um
total_height_um = max(array_height_um, coupon_height_um)

top_boundary = Device("TOP_BOUNDARY_PHIDL")
phidl_rectangle(
    top_boundary,
    0.0,
    0.0,
    total_width_um,
    total_height_um,
    "BOUNDARY",
)
top_boundary_cell = phidl_to_gdstk(
    top_boundary,
    "TOP_LEVEL_BOUNDARY",
)
top.add(gdstk.Reference(top_boundary_cell))
top.add(
    gdstk.Label(
        "ASAP7 128 KiB SRAM | 16 × 8 KiB | 4 × 4",
        (0.5 * array_width_um, total_height_um + 1.0),
        anchor="s",
        layer=text_spec.layer,
        texttype=LABEL_DATATYPE,
    )
)


floorplan_records = [
    {
        "banks": bank_count,
        "array_rows": BANK_ARRAY_ROWS,
        "array_columns": BANK_ARRAY_COLUMNS,
        "array_width_um": array_width_um,
        "array_height_um": array_height_um,
        "coupon_width_um": coupon_width_um,
        "top_width_um": total_width_um,
        "top_height_um": total_height_um,
        "total_KiB": TARGET_BYTES / 1024,
    }
]
display_records("Top-level floorplan", floorplan_records)

# -----------------------------------------------------------------------------
# Notebook code cell 11
# -----------------------------------------------------------------------------
def add_cells_without_name_collisions(
    output_library: gdstk.Library,
    cells: Iterable[gdstk.Cell],
    prefix: str,
) -> None:
    """Add cells, renaming colliding cell objects conservatively."""
    existing = {cell.name for cell in output_library.cells}
    for cell in cells:
        if cell.name in existing:
            original_name = cell.name
            suffix = 1
            candidate = f"{prefix}_{original_name}"
            while candidate in existing:
                suffix += 1
                candidate = f"{prefix}_{suffix}_{original_name}"
            cell.name = candidate
        output_library.add(cell)
        existing.add(cell.name)


output_library = gdstk.Library(unit=1e-6, precision=1e-9)
composed_cells: list[gdstk.Cell] = []
if bank_macro_is_wrapper:
    composed_cells.append(bank_macro)
add_cells_without_name_collisions(
    output_library,
    sram_library.cells,
    prefix="SRAMMACRO",
)
add_cells_without_name_collisions(
    output_library,
    standard_cell_library.cells,
    prefix="STDCELL",
)
composed_cells.extend(
    (
        bank_boundary_cell,
        bank_tile,
        top_mesh,
        layer_coupon,
        top_boundary_cell,
        top,
    )
)
add_cells_without_name_collisions(
    output_library,
    composed_cells,
    prefix="COMPOSED",
)


output_paths: list[Path] = []
export_failures: list[dict[str, str]] = []

if WRITE_GDS:
    gds_path = RESULT_DIR / OUTPUT_GDS_NAME
    output_library.write_gds(str(gds_path), max_points=8190)
    output_paths.append(gds_path)

if WRITE_OASIS:
    oas_path = RESULT_DIR / OUTPUT_OAS_NAME
    try:
        output_library.write_oas(str(oas_path))
        output_paths.append(oas_path)
    except Exception as error:
        export_failures.append(
            {
                "format": "oas",
                "error": f"{type(error).__name__}: {error}",
            }
        )
        print(
            "OASIS export failed. The GDSII deliverable was retained."
        )

if WRITE_GDS and not any(
    path.suffix.lower() == ".gds" for path in output_paths
):
    raise RuntimeError("GDSII export did not produce an output file.")

export_records = [
    {
        "format": path.suffix.lower().lstrip("."),
        "path": str(path),
        "bytes": path.stat().st_size,
        "sha256": sha256_file(path),
        "status": "written",
    }
    for path in output_paths
]
export_records.extend(
    {
        "format": failure["format"],
        "path": "",
        "bytes": "",
        "sha256": "",
        "status": failure["error"],
    }
    for failure in export_failures
)
display_records("Layout exports", export_records)

# -----------------------------------------------------------------------------
# Notebook code cell 12
# -----------------------------------------------------------------------------
def direct_layer_specs(cell: gdstk.Cell) -> set[tuple[int, int]]:
    """Collect direct polygon, path, and label layer specifications."""
    specs: set[tuple[int, int]] = set()
    for polygon in cell.polygons:
        specs.add((int(polygon.layer), int(polygon.datatype)))
    for path in cell.paths:
        specs.update(
            zip(
                (int(value) for value in path.layers),
                (int(value) for value in path.datatypes),
            )
        )
    for label in cell.labels:
        specs.add((int(label.layer), int(label.texttype)))
    return specs


all_direct_specs: set[tuple[int, int]] = set()
for cell in output_library.cells:
    all_direct_specs.update(direct_layer_specs(cell))

audit_records = []
for layer_name in PROCESS_LAYER_ORDER:
    spec = LAYERS[layer_name]
    audit_records.append(
        {
            "name": layer_name,
            "layer": spec.layer,
            "datatype": spec.datatype,
            "role": spec.role,
            "drawing_present": (
                spec.layer,
                spec.datatype,
            )
            in all_direct_specs,
            "pin_datatype_present": (
                spec.layer,
                PIN_DATATYPE,
            )
            in all_direct_specs,
            "label_datatype_present": (
                spec.layer,
                LABEL_DATATYPE,
            )
            in all_direct_specs,
        }
    )

missing_layers = [
    str(record["name"])
    for record in audit_records
    if not bool(record["drawing_present"])
]
if missing_layers:
    raise AssertionError(f"Missing drawing layers: {missing_layers}")

display_records("ASAP7 process-layer audit", audit_records)

# -----------------------------------------------------------------------------
# Notebook code cell 13
# -----------------------------------------------------------------------------
def polygon_points(polygon: object) -> np.ndarray:
    """Return polygon points from a GDSTK polygon-like object."""
    if hasattr(polygon, "points"):
        return np.asarray(polygon.points, dtype=float)
    return np.asarray(polygon, dtype=float)


def get_polygons_for_spec(
    cell: gdstk.Cell,
    spec: LayerSpec,
    depth: int | None,
) -> list[np.ndarray]:
    """Extract transformed polygons for one layer specification."""
    polygons = cell.get_polygons(
        apply_repetitions=True,
        include_paths=True,
        depth=depth,
        layer=spec.layer,
        datatype=spec.datatype,
    )
    return [polygon_points(polygon) for polygon in polygons]


def render_cell_layers(
    cell: gdstk.Cell,
    layer_names: Sequence[str],
    title: str,
    depth: int | None,
    figure_size: tuple[float, float],
    alpha: float = PLOT_LAYER_ALPHA,
) -> None:
    """Render selected hierarchy layers in memory at the configured DPI."""
    figure = plt.figure(figsize=figure_size, dpi=PREVIEW_DPI)
    axes = figure.add_axes((0.07, 0.08, 0.72, 0.84))
    color_cycle = itertools.cycle(
        plt.rcParams["axes.prop_cycle"].by_key()["color"]
    )
    legend_handles: list[Patch] = []
    polygon_total = 0

    for layer_name in layer_names:
        spec = LAYERS[layer_name]
        polygons = get_polygons_for_spec(cell, spec, depth)
        polygon_total += len(polygons)
        if polygon_total > MAX_RENDERED_POLYGONS:
            raise RuntimeError(
                "Render polygon limit exceeded. Reduce hierarchy depth "
                "or the selected layer set."
            )
        if not polygons:
            continue

        color = next(color_cycle)
        collection = PolyCollection(
            polygons,
            facecolors=color,
            edgecolors=color,
            linewidths=0.10,
            alpha=alpha,
            rasterized=True,
        )
        axes.add_collection(collection)
        legend_handles.append(
            Patch(
                facecolor=color,
                edgecolor=color,
                alpha=alpha,
                label=layer_name,
            )
        )

    bounds = bounding_box_array(cell)
    span = bounds[1] - bounds[0]
    x_span = float(span[0])
    y_span = float(span[1])
    margin = 0.02 * max(x_span, y_span)
    axes.set_xlim(bounds[0, 0] - margin, bounds[1, 0] + margin)
    axes.set_ylim(bounds[0, 1] - margin, bounds[1, 1] + margin)
    axes.set_aspect("equal", adjustable="box")
    axes.set_xlabel("x (µm)")
    axes.set_ylabel("y (µm)")
    axes.set_title(f"{title}\n{polygon_total:,} polygons rendered")

    if legend_handles:
        axes.legend(
            handles=legend_handles,
            loc="upper left",
            bbox_to_anchor=(1.01, 1.0),
            frameon=False,
        )

    plt.show()

# -----------------------------------------------------------------------------
# Notebook code cell 14
# -----------------------------------------------------------------------------
render_cell_layers(
    top,
    FULL_VIEW_LAYERS,
    title=(
        "ASAP7 hierarchical 128 KiB SRAM "
        "(16 × srambank_256x4x64_6t122)"
    ),
    depth=2,
    figure_size=(12.0, 9.0),
    alpha=PLOT_LAYER_ALPHA,
)

# -----------------------------------------------------------------------------
# Notebook code cell 15
# -----------------------------------------------------------------------------
def reference_histogram(library: gdstk.Library) -> Counter[str]:
    """Count direct references to every cell in a library."""
    counts: Counter[str] = Counter()
    for cell in library.cells:
        for reference in cell.references:
            target = reference.cell
            if isinstance(target, gdstk.Cell):
                counts[target.name] += 1
    return counts


BITCELL_POSITIVE_TOKENS = (
    "bitcell",
    "bit_cell",
    "sramcell",
    "sram_cell",
    "6t122",
    "6t123",
    "6t112",
    "6t111",
)

BITCELL_EXCLUDED_TOKENS = (
    "tapcell",
    "welltap",
    "tap_",
    "_tap",
    "dummy",
    "filler",
    "fill_",
    "_fill",
    "blank",
    "decap",
    "boundary",
    "marker",
    "endcap",
    "spacer",
    "strap",
    "precharge",
    "sense",
    "driver",
    "decoder",
    "column_mux",
    "row_mux",
    "control",
    "ctrl",
    "clock",
    "buffer",
    "inverter",
    "nand",
    "nor",
    "latch",
    "flop",
)


def referenced_cells(
    cell: gdstk.Cell,
) -> list[gdstk.Cell]:
    """Return directly referenced GDSTK cells."""
    children: list[gdstk.Cell] = []
    for reference in cell.references:
        target = reference.cell
        if isinstance(target, gdstk.Cell):
            children.append(target)
    return children


def candidate_layer_numbers(
    cell: gdstk.Cell,
    max_depth: int = 2,
) -> set[int]:
    """Collect drawing layers from a cell and shallow descendants."""
    layers: set[int] = set()
    visited: set[str] = set()

    def visit(current: gdstk.Cell, depth: int) -> None:
        if current.name in visited:
            return
        visited.add(current.name)

        layers.update(
            int(polygon.layer)
            for polygon in current.polygons
        )
        for path in current.paths:
            layers.update(int(value) for value in path.layers)

        if depth >= max_depth:
            return

        for child in referenced_cells(current):
            visit(child, depth + 1)

    visit(cell, 0)
    return layers


def contains_any_token(
    name: str,
    tokens: Sequence[str],
) -> bool:
    """Return whether a lowercase name contains any token."""
    lowered = name.lower()
    return any(token in lowered for token in tokens)


def classify_zoom_cell(cell: gdstk.Cell) -> str:
    """Classify a selected zoom target as bitcell or support logic."""
    name_lower = cell.name.lower()

    if contains_any_token(
        name_lower,
        BITCELL_EXCLUDED_TOKENS,
    ):
        return "supporting cell"

    if contains_any_token(
        name_lower,
        BITCELL_POSITIVE_TOKENS,
    ):
        return "bitcell"

    return "supporting cell"


def select_bitcell_candidate(
    library: gdstk.Library,
) -> tuple[
    gdstk.Cell,
    str,
    str,
    list[dict[str, Any]],
]:
    """Select a genuine SRAM bitcell and reject helper structures."""
    reference_counts = reference_histogram(library)
    required_feol = {
        LAYERS["fin"].layer,
        LAYERS["Gate"].layer,
        LAYERS["Active"].layer,
    }

    scored: list[
        tuple[float, gdstk.Cell, str, str]
    ] = []

    for cell in library.cells:
        name_lower = cell.name.lower()

        if contains_any_token(
            name_lower,
            BITCELL_EXCLUDED_TOKENS,
        ):
            continue

        try:
            bounds = bounding_box_array(cell)
        except ValueError:
            continue

        span = bounds[1] - bounds[0]
        width = float(span[0])
        height = float(span[1])
        area = width * height

        if area < 1e-3 or area > 25.0:
            continue
        if width < 0.02 or height < 0.02:
            continue

        positive_tokens = [
            token
            for token in BITCELL_POSITIVE_TOKENS
            if token in name_lower
        ]
        name_score = 0.0

        token_weights = {
            "bitcell": 9000.0,
            "bit_cell": 9000.0,
            "sramcell": 8000.0,
            "sram_cell": 8000.0,
            "6t122": 5000.0,
            "6t123": 4500.0,
            "6t112": 4500.0,
            "6t111": 4000.0,
        }
        for token in positive_tokens:
            name_score += token_weights[token]

        layers = candidate_layer_numbers(cell, max_depth=2)
        has_feol = required_feol.issubset(layers)
        layer_score = 2500.0 if has_feol else 0.0

        if not positive_tokens:
            continue
        if not has_feol:
            continue

        repeat_count = reference_counts[cell.name]
        repeat_score = 35.0 * math.log1p(repeat_count)
        compactness_score = min(800.0, 0.5 / area)
        hierarchy_score = (
            250.0 if cell.references else 0.0
        )

        score = (
            name_score
            + layer_score
            + repeat_score
            + compactness_score
            + hierarchy_score
        )

        reason = (
            f"name tokens={positive_tokens}, "
            f"name score={name_score:.0f}, "
            f"FEOL score={layer_score:.0f}, "
            f"references={repeat_count}, "
            f"area={area:.6g} µm²"
        )
        scored.append(
            (
                score,
                cell,
                reason,
                "bitcell",
            )
        )

    scored.sort(key=lambda item: item[0], reverse=True)

    diagnostic_records = [
        {
            "rank": rank,
            "cell": cell.name,
            "score": score,
            "classification": classification,
            "bbox_um": bbox_size(cell),
            "reason": reason,
        }
        for rank, (
            score,
            cell,
            reason,
            classification,
        ) in enumerate(scored[:15], start=1)
    ]

    if not scored:
        fallback_reason = (
            "No non-helper SRAM bitcell with FEOL geometry was "
            "resolved. Using the representative SRAM-VT standard "
            "cell as an explicitly labeled supporting-cell fallback."
        )
        fallback_records = [
            {
                "rank": 1,
                "cell": representative_std.name,
                "score": "",
                "classification": "supporting cell",
                "bbox_um": bbox_size(representative_std),
                "reason": fallback_reason,
            }
        ]
        return (
            representative_std,
            fallback_reason,
            "supporting cell",
            fallback_records,
        )

    best_score, best_cell, best_reason, classification = scored[0]

    if best_score < 7000.0:
        fallback_reason = (
            f"Best bitcell score {best_score:.2f} was below the "
            "acceptance threshold. Using the representative SRAM-VT "
            "standard cell as a supporting-cell fallback."
        )
        diagnostic_records.insert(
            0,
            {
                "rank": 0,
                "cell": representative_std.name,
                "score": "",
                "classification": "supporting cell",
                "bbox_um": bbox_size(representative_std),
                "reason": fallback_reason,
            },
        )
        return (
            representative_std,
            fallback_reason,
            "supporting cell",
            diagnostic_records,
        )

    return (
        best_cell,
        best_reason,
        classification,
        diagnostic_records,
    )


(
    zoom_cell,
    zoom_selection_reason,
    zoom_cell_classification,
    zoom_candidate_records,
) = select_bitcell_candidate(sram_library)

display_records(
    "SRAM zoom-cell candidate audit",
    zoom_candidate_records,
)

print(f"Zoom cell: {zoom_cell.name}")
print(f"Classification: {zoom_cell_classification}")
print(f"Selection basis: {zoom_selection_reason}")
print(f"Bounding box: {bbox_size(zoom_cell)} µm")

# -----------------------------------------------------------------------------
# Notebook code cell 16
# -----------------------------------------------------------------------------
if zoom_cell in sram_library.cells:
    zoom_scale = bank_scale
else:
    zoom_scale = standard_cell_scale

zoom_lower, _, _ = cell_lower_left_and_size(
    zoom_cell,
    zoom_scale,
)
zoom_render_cell = gdstk.Cell("ZOOM_RENDER_CELL")
zoom_render_cell.add(
    gdstk.Reference(
        zoom_cell,
        origin=(-zoom_lower[0], -zoom_lower[1]),
        magnification=zoom_scale,
    )
)

if zoom_cell_classification == "bitcell":
    zoom_title = (
        f"ASAP7 SRAM bitcell zoom: {zoom_cell.name}"
    )
else:
    zoom_title = (
        "ASAP7 SRAM supporting-cell zoom: "
        f"{zoom_cell.name}"
    )

render_cell_layers(
    zoom_render_cell,
    ZOOM_VIEW_LAYERS,
    title=zoom_title,
    depth=None,
    figure_size=(10.0, 8.0),
    alpha=PLOT_LAYER_ALPHA,
)

# -----------------------------------------------------------------------------
# Notebook code cell 17
# -----------------------------------------------------------------------------
# PRODUCE A TRANSISTOR + INTERCONNECT + VIA ONLY PHYSICAL VIEW
#
# This cell flattens selected process layers from the chosen hierarchy,
# renders them in memory, and exports a self-contained filtered GDS/OASIS.
# It intentionally excludes boundaries, text, SRAM markers, and unrelated
# layers from the filtered physical layout.

# -------------------------------------------------------------------------
# CONTROL KNOBS
# -------------------------------------------------------------------------
FILTER_SOURCE = "bitcell"
# Allowed values:
#   "bitcell"    -> selected SRAM bitcell zoom
#   "bank"       -> one composed 8 KiB bank tile
#   "full_array" -> complete 128 KiB hierarchy

INCLUDE_WELL_AND_SELECT = True
INCLUDE_DUMMY_GATE = False
INCLUDE_MOL_INTERCONNECT = True
MAX_METAL_LEVEL = 3
# MAX_METAL_LEVEL=3 includes:
#   V0, M1, V1, M2, V2, M3
# Set 1 through 9 to extend the retained routing stack.

FILTERED_PREVIEW_DPI = 600
FILTERED_LAYER_ALPHA = PLOT_LAYER_ALPHA
FILTERED_FIGURE_SIZE = (10.0, 8.0)
MAX_FILTERED_POLYGONS = 500_000

RENDER_FILTERED_LAYOUT = True
RENDER_SEPARATE_GROUPS = False
EXPORT_FILTERED_GDS = True
EXPORT_FILTERED_OASIS = True

FILTERED_OUTPUT_STEM = "ASAP7_SRAM_transistors_interconnects_vias"


def sanitize_cell_name(name: str) -> str:
    """Return a GDS-safe cell-name fragment."""
    cleaned = re.sub(r"[^A-Za-z0-9_]+", "_", name)
    cleaned = re.sub(r"_+", "_", cleaned).strip("_")
    return cleaned[:96] or "FILTERED_LAYOUT"


def build_filtered_layer_list(
    include_well_and_select: bool,
    include_dummy_gate: bool,
    include_mol: bool,
    max_metal_level: int,
) -> tuple[
    tuple[str, ...],
    tuple[str, ...],
    tuple[str, ...],
    tuple[str, ...],
]:
    """Build FEOL, MOL, metal, and via layer selections."""
    if not 1 <= max_metal_level <= 9:
        raise ValueError("MAX_METAL_LEVEL must be between 1 and 9.")

    transistor_layers = ["fin", "Gate", "GCut", "Active"]

    if include_well_and_select:
        transistor_layers = [
            "well",
            *transistor_layers,
            "Nselect",
            "Pselect",
        ]

    if include_dummy_gate:
        gate_index = transistor_layers.index("Gate") + 1
        transistor_layers.insert(gate_index, "Dummy")

    mol_layers = (
        ["SDT", "LIG", "LISD"]
        if include_mol
        else []
    )

    metal_layers = [
        f"M{level}"
        for level in range(1, max_metal_level + 1)
    ]

    via_layers = ["V0"]
    via_layers.extend(
        f"V{level}"
        for level in range(1, max_metal_level)
    )

    ordered_layers: list[str] = [
        *transistor_layers,
        *mol_layers,
        "V0",
    ]

    for level in range(1, max_metal_level + 1):
        ordered_layers.append(f"M{level}")
        if level < max_metal_level:
            ordered_layers.append(f"V{level}")

    missing = [
        layer_name
        for layer_name in ordered_layers
        if layer_name not in LAYERS
    ]
    if missing:
        raise KeyError(
            f"Selected layers are absent from LAYERS: {missing}"
        )

    return (
        tuple(transistor_layers),
        tuple(mol_layers),
        tuple(metal_layers),
        tuple(via_layers),
    )


def resolve_filter_source(
    source_name: str,
) -> tuple[gdstk.Cell, str]:
    """Resolve the hierarchy selected for layer filtering."""
    source_map = {
        "bitcell": (
            zoom_render_cell,
            f"SRAM bitcell {zoom_cell.name}",
        ),
        "bank": (
            bank_tile,
            "one composed 8 KiB SRAM bank",
        ),
        "full_array": (
            top,
            "complete 128 KiB SRAM array",
        ),
    }

    if source_name not in source_map:
        raise ValueError(
            "FILTER_SOURCE must be 'bitcell', 'bank', or "
            "'full_array'."
        )

    return source_map[source_name]


def copy_selected_layers_flat(
    source_cell: gdstk.Cell,
    layer_names: Sequence[str],
    output_name: str,
    polygon_limit: int,
) -> tuple[
    gdstk.Cell,
    list[dict[str, Any]],
]:
    """Flatten selected hierarchy layers into a self-contained cell."""
    filtered_cell = gdstk.Cell(output_name)
    audit_records: list[dict[str, Any]] = []
    polygon_total = 0

    for layer_name in layer_names:
        spec = LAYERS[layer_name]
        polygons = get_polygons_for_spec(
            source_cell,
            spec,
            depth=None,
        )

        polygon_total += len(polygons)
        if polygon_total > polygon_limit:
            raise RuntimeError(
                "Filtered polygon limit exceeded. Reduce "
                "MAX_METAL_LEVEL, change FILTER_SOURCE, or increase "
                "MAX_FILTERED_POLYGONS."
            )

        for points in polygons:
            filtered_cell.add(
                gdstk.Polygon(
                    np.asarray(points, dtype=float),
                    layer=spec.layer,
                    datatype=spec.datatype,
                )
            )

        audit_records.append(
            {
                "layer_name": layer_name,
                "layer": spec.layer,
                "datatype": spec.datatype,
                "role": spec.role,
                "polygon_count": len(polygons),
            }
        )

    if polygon_total == 0:
        raise RuntimeError(
            "No polygons were found on the selected transistor, "
            "interconnect, metal, and via layers."
        )

    return filtered_cell, audit_records


def write_filtered_layout(
    filtered_cell: gdstk.Cell,
    output_stem: str,
) -> list[dict[str, Any]]:
    """Write the filtered flat cell to GDSII and optional OASIS."""
    filtered_library = gdstk.Library(
        unit=1e-6,
        precision=1e-9,
    )
    filtered_library.add(filtered_cell)

    export_records: list[dict[str, Any]] = []

    if EXPORT_FILTERED_GDS:
        gds_path = RESULT_DIR / f"{output_stem}.gds"
        filtered_library.write_gds(
            str(gds_path),
            max_points=8190,
        )
        export_records.append(
            {
                "format": "gds",
                "path": str(gds_path),
                "bytes": gds_path.stat().st_size,
                "sha256": sha256_file(gds_path),
                "status": "written",
            }
        )

    if EXPORT_FILTERED_OASIS:
        oas_path = RESULT_DIR / f"{output_stem}.oas"
        try:
            filtered_library.write_oas(str(oas_path))
            export_records.append(
                {
                    "format": "oas",
                    "path": str(oas_path),
                    "bytes": oas_path.stat().st_size,
                    "sha256": sha256_file(oas_path),
                    "status": "written",
                }
            )
        except Exception as error:
            export_records.append(
                {
                    "format": "oas",
                    "path": "",
                    "bytes": "",
                    "sha256": "",
                    "status": (
                        f"{type(error).__name__}: {error}"
                    ),
                }
            )

    return export_records


(
    FILTERED_TRANSISTOR_LAYERS,
    FILTERED_MOL_LAYERS,
    FILTERED_METAL_LAYERS,
    FILTERED_VIA_LAYERS,
) = build_filtered_layer_list(
    include_well_and_select=INCLUDE_WELL_AND_SELECT,
    include_dummy_gate=INCLUDE_DUMMY_GATE,
    include_mol=INCLUDE_MOL_INTERCONNECT,
    max_metal_level=MAX_METAL_LEVEL,
)

FILTERED_LAYER_NAMES = (
    *FILTERED_TRANSISTOR_LAYERS,
    *FILTERED_MOL_LAYERS,
    "V0",
    *tuple(
        layer_name
        for level in range(1, MAX_METAL_LEVEL + 1)
        for layer_name in (
            (f"M{level}", f"V{level}")
            if level < MAX_METAL_LEVEL
            else (f"M{level}",)
        )
    ),
)

filter_source_cell, filter_source_description = (
    resolve_filter_source(FILTER_SOURCE)
)

filtered_cell_name = sanitize_cell_name(
    f"{FILTER_SOURCE}_TRANSISTORS_INTERCONNECTS_VIAS"
)

(
    transistor_interconnect_cell,
    filtered_layer_audit,
) = copy_selected_layers_flat(
    source_cell=filter_source_cell,
    layer_names=FILTERED_LAYER_NAMES,
    output_name=filtered_cell_name,
    polygon_limit=MAX_FILTERED_POLYGONS,
)

filtered_polygon_total = sum(
    int(record["polygon_count"])
    for record in filtered_layer_audit
)

display_records(
    "Filtered transistor/interconnect layer audit",
    filtered_layer_audit,
)

print(
    "Filtered source:",
    filter_source_description,
)
print(
    "Retained layers:",
    ", ".join(FILTERED_LAYER_NAMES),
)
print(
    "Filtered polygon count:",
    f"{filtered_polygon_total:,}",
)

if RENDER_FILTERED_LAYOUT:
    original_preview_dpi = PREVIEW_DPI
    PREVIEW_DPI = FILTERED_PREVIEW_DPI
    try:
        render_cell_layers(
            transistor_interconnect_cell,
            FILTERED_LAYER_NAMES,
            title=(
                "ASAP7 transistor layers, interconnects, and vias\n"
                f"{filter_source_description}"
            ),
            depth=None,
            figure_size=FILTERED_FIGURE_SIZE,
            alpha=FILTERED_LAYER_ALPHA,
        )
    finally:
        PREVIEW_DPI = original_preview_dpi

if RENDER_SEPARATE_GROUPS:
    render_cell_layers(
        transistor_interconnect_cell,
        FILTERED_TRANSISTOR_LAYERS,
        title=(
            "ASAP7 transistor-forming FEOL layers\n"
            f"{filter_source_description}"
        ),
        depth=None,
        figure_size=FILTERED_FIGURE_SIZE,
        alpha=FILTERED_LAYER_ALPHA,
    )

    interconnect_only_layers = (
        *FILTERED_MOL_LAYERS,
        *FILTERED_VIA_LAYERS,
        *FILTERED_METAL_LAYERS,
    )
    render_cell_layers(
        transistor_interconnect_cell,
        interconnect_only_layers,
        title=(
            "ASAP7 local interconnect, metals, and vias\n"
            f"{filter_source_description}"
        ),
        depth=None,
        figure_size=FILTERED_FIGURE_SIZE,
        alpha=FILTERED_LAYER_ALPHA,
    )

filtered_export_records: list[dict[str, Any]] = []
if EXPORT_FILTERED_GDS or EXPORT_FILTERED_OASIS:
    source_suffix = sanitize_cell_name(
        FILTER_SOURCE.lower()
    )
    filtered_output_stem = (
        f"{FILTERED_OUTPUT_STEM}_{source_suffix}_M"
        f"{MAX_METAL_LEVEL}"
    )
    filtered_export_records = write_filtered_layout(
        transistor_interconnect_cell,
        filtered_output_stem,
    )
    display_records(
        "Filtered physical-layout exports",
        filtered_export_records,
    )
```
