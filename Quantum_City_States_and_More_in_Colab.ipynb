## Foundational Primer: Density Operators, Pure States, and Mixed States

Quantum mechanics represents the physical state of an isolated, deterministic system as a normalized state vector $|\psi\rangle$ residing within a complex Hilbert space $\mathcal{H}$ of finite dimension $d = 2^N$, where $N$ denotes the number of two-level quantum systems, or qubits. Realistic experimental architectures inevitably experience classical preparation uncertainty and environmental entanglement, which necessitates replacing the single state vector with the density operator $\hat{\rho}$ acting on $\mathcal{H}$. Expanding $\hat{\rho}$ within a fixed orthonormal computational basis $\{|0\rangle, |1\rangle, \dots, |d-1\rangle\}$ yields a $d \times d$ Hermitian matrix $\boldsymbol{\rho}$ with complex entries $\rho_{jk} = \langle j|\hat{\rho}|k\rangle$, where $j$ and $k$ index the row and column basis states from $0$ to $d-1$. Positive semi-definiteness of $\hat{\rho}$ ensures real, non-negative eigenvalues that sum to unity under the trace normalization condition $\mathrm{Tr}(\hat{\rho}) = 1$. Visualizing this matrix via a Three-Dimensional (3D) state city plot maps the discrete indices $(j, k)$ onto a Two-Dimensional (2D) horizontal planar grid and extrudes vertical rectangular prisms proportional to the real component $\mathrm{Re}(\rho_{jk})$ or imaginary component $\mathrm{Im}(\rho_{jk})$.

A pure quantum state characterizes a physical system prepared in a coherent quantum superposition $|\psi\rangle = \sum_{j=0}^{d-1} c_j |j\rangle$, where each expansion coefficient $c_j = |c_j|e^{i\theta_j} \in \mathbb{C}$ specifies the complex probability amplitude with modulus $|c_j|$ and phase angle $\theta_j$. Constructing the outer product of the ket vector $|\psi\rangle$ with its conjugate transpose bra vector $\langle\psi| = \sum_{k=0}^{d-1} c_k^* \langle k|$ produces the rank-one projection operator $\hat{\rho}_{\mathrm{pure}}$, where $c_k^*$ denotes the complex conjugate of $c_k$:

$$\hat{\rho}_{\mathrm{pure}} = |\psi\rangle\langle\psi| = \sum_{j=0}^{d-1} \sum_{k=0}^{d-1} c_j c_k^* |j\rangle\langle k|, \quad \text{valid for } \sum_{j=0}^{d-1} |c_j|^2 = 1 \text{ and } \hat{\rho}_{\mathrm{pure}}^2 = \hat{\rho}_{\mathrm{pure}}$$

Idempotence of the pure density operator ($\hat{\rho}_{\mathrm{pure}}^2 = \hat{\rho}_{\mathrm{pure}}$) guarantees that every off-diagonal element satisfies the factorization identity $|\rho_{jk}| = |c_j||c_k| = \sqrt{\rho_{jj}\rho_{kk}}$, which locks the magnitude of every off-diagonal coherence tower in a state city plot to the geometric mean of its corresponding diagonal population towers. Evaluating the state purity $\gamma$, defined as the trace of the squared density operator, provides an essential basis-independent scalar metric that distinguishes pure quantum states from incoherent statistical mixtures. Across any $d$-dimensional Hilbert space $\mathcal{H}$, the purity $\gamma$ obeys the bounded normalization relation:

$$\gamma \equiv \mathrm{Tr}\left(\hat{\rho}^2\right) = \sum_{j=0}^{d-1} \sum_{k=0}^{d-1} |\rho_{jk}|^2, \quad \text{valid for } \frac{1}{d} \le \gamma \le 1$$

Statistical ensembles arise under partial decoherence, thermal equilibration, or stochastic state preparation, producing a mixed quantum state $\hat{\rho}_{\mathrm{mixed}}$ formulated as a convex combination of $M$ pure projection operators $|\psi_m\rangle\langle\psi_m|$ weighted by classical probabilities $p_m$. Each weighting coefficient $p_m$ represents the classical statistical fraction of the ensemble occupying pure sub-state $|\psi_m\rangle$, which breaks idempotence and distributes spectral weight across multiple non-zero eigenvalues:

$$\hat{\rho}_{\mathrm{mixed}} = \sum_{m=1}^{M} p_m |\psi_m\rangle\langle\psi_m|, \quad \text{valid for } \sum_{m=1}^{M} p_m = 1, \quad 0 \lt p_m \lt 1, \quad \text{and } \hat{\rho}_{\mathrm{mixed}}^2 \neq \hat{\rho}_{\mathrm{mixed}}$$

Decoherence and classical ensemble averaging damp the off-diagonal matrix elements below the geometric mean bound ($|\rho_{jk}|$ less than $\sqrt{\rho_{jj}\rho_{kk}}$), which visibly attenuates or eliminates the off-diagonal towers in the state city plot. Under Markovian phase relaxation characterized by transverse dephasing time $T_2$ over elapsed evolution time $t \ge 0$, the off-diagonal coherence amplitude obeys the asymptotic scaling relation:

$$|\rho_{jk}(t)| \propto |\rho_{jk}(0)| \exp\left(-\frac{t}{T_2}\right), \quad \text{valid for } j \neq k \text{ and } t \ge 0$$

At the same time, the purity $\gamma = \mathrm{Tr}(\hat{\rho}_{\mathrm{mixed}}^2)$ drops below unity and approaches the lower bound of $1/d$ for a maximally mixed state $\hat{\rho}_{\mathrm{max}} = \frac{1}{d}\hat{I}_d$, where $\hat{I}_d$ denotes the $d \times d$ identity operator. Complementary to purity, the von Neumann entropy $S(\hat{\rho}) = -\mathrm{Tr}(\hat{\rho}\ln\hat{\rho})$ vanishes for a pure state ($S = 0$) and reaches its maximum $\ln(d)$ for a maximally mixed distribution.

## Anatomical Topology of the State City Plot Corners

Geometric inspection of the $d \times d$ state city grid reveals that the four extreme perimeter corners anchor the physical interpretation of both single-qubit ($d = 2$) and multi-qubit ($d = 2^N$) quantum states in Quantum State Tomography (QST). Located at the back-left origin index $(0, 0)$, the North-West corner element $\rho_{0,0} = \langle 0|\hat{\rho}|0\rangle$ quantifies the classical occupation probability of the ground computational basis state $|0\rangle$ (or $|00\dots 0\rangle$ in an $N$-qubit register). Situated at the opposite extremity of the main diagonal at index $(d-1, d-1)$, the South-East corner element $\rho_{d-1,d-1} = \langle d-1|\hat{\rho}|d-1\rangle$ measures the classical population of the highest-index excited basis state $|d-1\rangle$ (or $|11\dots 1\rangle$). Hermiticity of the density operator ($\hat{\rho}^\dagger = \hat{\rho}$) forces both diagonal corners to exhibit zero imaginary height ($\mathrm{Im}(\rho_{0,0}) = \mathrm{Im}(\rho_{d-1,d-1}) = 0$) and non-negative real height bounded between zero and one.

Positioned at the back-right perimeter coordinate $(0, d-1)$ and front-left perimeter coordinate $(d-1, 0)$, the North-East corner $\rho_{0,d-1} = \langle 0|\hat{\rho}|d-1\rangle$ and South-West corner $\rho_{d-1,0} = \langle d-1|\hat{\rho}|0\rangle$ occupy the extremities of the anti-diagonal. These two anti-diagonal corners govern the maximal-span quantum coherence, capturing the phase relationship and interference amplitude between the ground state $|0\rangle$ and the highest excited state $|d-1\rangle$. Conjugate symmetry ($\rho_{d-1,0} = \rho_{0,d-1}^*$) forces the North-East and South-West corner towers to display identical vertical heights in the real-component state city plot ($\mathrm{Re}(\rho_{0,d-1}) = \mathrm{Re}(\rho_{d-1,0})$). Conversely, the imaginary-component state city plot displays these two anti-diagonal corners with equal magnitudes and opposite algebraic polarity ($\mathrm{Im}(\rho_{0,d-1}) = -\mathrm{Im}(\rho_{d-1,0})$), where a positive tower on one corner pairs with a negative trough of equal depth on the opposite corner.

## Correlation Across Density Matrix Representations

Correlating these four corners across standard density matrix parameterizations clarifies how state geometry transforms across Hilbert space dimensions. Within the single-qubit Bloch representation ($d = 2$), the $2 \times 2$ grid contains exclusively the four corner cells, parameterized by the 3D Bloch vector $\mathbf{r} = (r_x, r_y, r_z) \in \mathbb{R}^3$ and the Pauli spin matrices $\hat{\sigma}_x, \hat{\sigma}_y, \hat{\sigma}_z$:

$$\boldsymbol{\rho}_{\mathrm{1Q}}(\mathbf{r}) = \frac{1}{2}\left(\hat{I}_2 + r_x \hat{\sigma}_x + r_y \hat{\sigma}_y + r_z \hat{\sigma}_z\right) = \frac{1}{2}\begin{pmatrix} 1 + r_z & r_x - i r_y \\ r_x + i r_y & 1 - r_z \end{pmatrix}, \quad \text{valid for } \|\mathbf{r}\|_2 = \sqrt{r_x^2 + r_y^2 + r_z^2} \le 1$$

Longitudinal spin polarization $r_z = \mathrm{Tr}(\hat{\rho}\hat{\sigma}_z)$ directly controls the height differential between the North-West diagonal corner $\rho_{00} = (1 + r_z)/2$ and the South-East diagonal corner $\rho_{11} = (1 - r_z)/2$. Transverse equatorial coherence components $r_x = \mathrm{Tr}(\hat{\rho}\hat{\sigma}_x)$ and $r_y = \mathrm{Tr}(\hat{\rho}\hat{\sigma}_y)$ map directly onto the anti-diagonal corners, setting the real state city corner heights to $\mathrm{Re}(\rho_{01}) = \mathrm{Re}(\rho_{10}) = r_x/2$ and the imaginary corner heights to $\mathrm{Im}(\rho_{10}) = -\mathrm{Im}(\rho_{01}) = r_y/2$. Pure single-qubit states reside on the Bloch sphere surface ($\Vert{}\mathbf{r}\Vert{}_2 = 1$) and sustain full-amplitude anti-diagonal corner towers for equatorial superpositions. Conversely, phase damping shrinks the transverse vector components ($r_x, r_y \to 0$), which collapses the anti-diagonal corners without altering the diagonal populations.

Scaling the Hilbert space to a two-qubit register ($d = 4$) expands the state city grid to $16$ elements indexed across $\{|00\rangle, |01\rangle, |10\rangle, |11\rangle\}$, where the four perimeter corners isolate nonlocal bipartite entanglement in Bell and Greenberger-Horne-Zeilinger (GHZ) states. Preparing a pure Bell state $|\Phi(\phi)\rangle = \frac{1}{\sqrt{2}}(|00\rangle + e^{i\phi}|11\rangle)$ with relative phase $\phi$ concentrates all matrix weight exclusively onto the four perimeter corners: $\rho_{0,0} = \rho_{3,3} = 1/2$ and $\rho_{0,3} = \rho_{3,0}^* = \frac{1}{2}e^{-i\phi}$, leaving the twelve interior cells completely flat at zero. Subjecting this pure entangled state to isotropic depolarizing noise (parameterized by visibility $p \in [0, 1]$) alongside pure phase damping (parameterized by dephasing fraction $\gamma_d \in [0, 1]$) yields the generalized Werner-dephased density matrix $\boldsymbol{\rho}_{\mathrm{2Q}}(p, \gamma_d, \phi)$:

$$\boldsymbol{\rho}_{\mathrm{2Q}}(p, \gamma_d, \phi) = \begin{pmatrix} \frac{1 + p}{4} & 0 & 0 & \frac{p(1 - \gamma_d)}{2}e^{-i\phi} \\ 0 & \frac{1 - p}{4} & 0 & 0 \\ 0 & 0 & \frac{1 - p}{4} & 0 \\ \frac{p(1 - \gamma_d)}{2}e^{i\phi} & 0 & 0 & \frac{1 + p}{4} \end{pmatrix}, \quad \text{valid for } 0 \le p, \gamma_d \le 1 \text{ and } \phi \in [0, 2\pi)$$

Decreasing the visibility parameter $p$ below unity or increasing the dephasing parameter $\gamma_d$ above zero induces distinct topological shifts across the $4 \times 4$ state city skyline and reduces the exact analytical purity $\gamma_{\mathrm{2Q}}(p, \gamma_d) = \frac{1 + p^2}{4} + \frac{p^2(1 - \gamma_d)^2}{2}$. First, the anti-diagonal coherence corners $\rho_{0,3}$ and $\rho_{3,0}$ attenuate proportionally to $p(1 - \gamma_d)/2$, directly reflecting the decay of nonlocal quantum entanglement. Second, isotropic depolarizing noise ($p$ less than $1$) redistributes probability out of the outer diagonal corners $\rho_{0,0} = \rho_{3,3} = (1 + p)/4$ and elevates the previously vacant interior diagonal towers $\rho_{1,1} = \langle 01|\hat{\rho}|01\rangle$ and $\rho_{2,2} = \langle 10|\hat{\rho}|10\rangle$ to height $(1 - p)/4$. In this way, comparing the ratio of the anti-diagonal corner amplitudes to the diagonal corner and interior amplitudes facilitates an immediate visual diagnostic that separates pure phase damping (which preserves diagonal corners at $1/2$ and leaves interior diagonals at $0$) from depolarizing thermalization (which equalizes the entire diagonal ridge toward $1/4$).

## Symbolic and Numerical State City Implementation in Google Colab

Symbolic and numerical implementation of these state city comparisons unites SymPy for exact algebraic matrix construction, QuTiP for operator-level state manipulation and tomography metrics, and Matplotlib for high-resolution 3D bar rendering. SymPy constructs the exact parameterized density matrices and verifies the symbolic trace, purity, and corner identities in closed analytical form. QuTiP instantiates the corresponding `Qobj` state objects, evaluates numerical purity $\gamma$ and von Neumann entropy $S(\hat{\rho})$, and interfaces with `matrix_histogram` alongside custom color-coded 3D bar projections that render side-by-side real and imaginary state city plots with direct corner callouts.

```
"""Compare pure and mixed quantum states using 3D state-city plots.

SymPy derives analytical density matrices. QuTiP constructs physical density
operators and evaluates their purity and von Neumann entropy. Matplotlib draws
projection-aware, collision-free corner plots alongside QuTiP's native
matrix histograms.

Install dependencies in a terminal before running this Python file:
    uv pip install qutip sympy matplotlib numpy

In Google Colab, use a separate setup cell:
    !uv pip install --quiet qutip sympy matplotlib numpy
"""

from __future__ import annotations

import dataclasses
from itertools import permutations
from typing import Any

import matplotlib as mpl
import matplotlib.patches as mpatches
import matplotlib.pyplot as plt
from matplotlib.text import Annotation
from matplotlib.cm import ScalarMappable
from mpl_toolkits.mplot3d import proj3d
import numpy as np
import qutip as qt
import sympy as sp


# =============================================================================
# CONTROL KNOBS
# =============================================================================
@dataclasses.dataclass(frozen=True)
class SimulationControlKnobs:
    """Configure the quantum states, rendering, and manual 3D subplot spacing."""

    num_qubits: int = 2
    relative_phase_rad: float = np.pi / 4.0
    werner_visibility_p: float = 0.60
    dephasing_fraction_gamma: float = 0.25
    single_qubit_bloch_pure: tuple[float, float, float] = (
        1.0 / np.sqrt(2.0),
        -1.0 / np.sqrt(2.0),
        0.0,
    )
    single_qubit_bloch_mixed: tuple[float, float, float] = (0.30, -0.25, 0.20)
    figure_dpi: int = 250
    view_elevation_deg: float = 28.0
    view_azimuth_deg: float = -42.0
    bar_width: float = 0.56
    corner_label_fontsize: float = 8.3
    corner_label_inset: float = 0.035
    corner_label_top: float = 0.87
    corner_label_bottom: float = 0.17
    corner_leader_linewidth: float = 1.2

    # Manual figure geometry. Avoid tight_layout() for 3D axes and colorbars.
    figure1_size: tuple[float, float] = (14.5, 11.5)
    figure2_size: tuple[float, float] = (16.0, 7.6)
    figure1_wspace: float = 0.20
    figure1_hspace: float = 0.32
    histogram_colormap: str = "coolwarm"
    histogram_left_axes: tuple[float, float, float, float] = (
        0.035, 0.235, 0.425, 0.575
    )
    histogram_right_axes: tuple[float, float, float, float] = (
        0.535, 0.235, 0.425, 0.575
    )
    histogram_colorbar_axes: tuple[float, float, float, float] = (
        0.285, 0.075, 0.430, 0.028
    )


KNOBS = SimulationControlKnobs()
mpl.rcParams["figure.dpi"] = KNOBS.figure_dpi
mpl.rcParams["font.sans-serif"] = ["DejaVu Sans"]
mpl.rcParams["font.weight"] = "normal"
mpl.rcParams["axes.unicode_minus"] = False
mpl.rcParams["figure.autolayout"] = False
mpl.rcParams["figure.constrained_layout.use"] = False


# =============================================================================
# SYMBOLIC DENSITY MATRIX AND CORNER DERIVATIONS
# =============================================================================
def derive_symbolic_representations(knobs: SimulationControlKnobs) -> None:
    """Print the analytical 1Q and 2Q density matrices and their purities."""
    r_x, r_y, r_z = sp.symbols("r_x r_y r_z", real=True)
    p_sym, gamma_d, phi_sym = sp.symbols("p gamma_d phi", real=True)

    rho_1q = sp.Matrix(
        [
            [(1 + r_z) / 2, (r_x - sp.I * r_y) / 2],
            [(r_x + sp.I * r_y) / 2, (1 - r_z) / 2],
        ]
    )
    purity_1q = sp.simplify(sp.trace(rho_1q * rho_1q))

    ket_bell = sp.Matrix([1, 0, 0, sp.exp(sp.I * phi_sym)]) / sp.sqrt(2)
    rho_2q_pure = sp.simplify(ket_bell * ket_bell.H)
    purity_2q_pure = sp.simplify(sp.trace(rho_2q_pure * rho_2q_pure))

    rho_2q_mixed = sp.Matrix(
        p_sym * rho_2q_pure + (1 - p_sym) * sp.eye(4) / 4
    )
    rho_2q_mixed[0, 3] *= 1 - gamma_d
    rho_2q_mixed[3, 0] *= 1 - gamma_d
    rho_2q_mixed = sp.simplify(rho_2q_mixed)
    purity_2q_mixed = sp.simplify(sp.trace(rho_2q_mixed * rho_2q_mixed))

    print("=" * 78)
    print("SYMPY SYMBOLIC DENSITY MATRIX AND CORNER ANALYSIS")
    print("=" * 78)
    print("\n[1] Single-qubit Bloch density matrix:")
    sp.pprint(rho_1q, use_unicode=True)
    print(f"Purity Tr(rho^2): {purity_1q}")
    print(f"  NW (0,0), population <0|rho|0>: {rho_1q[0, 0]}")
    print(f"  SE (1,1), population <1|rho|1>: {rho_1q[1, 1]}")
    print(f"  NE (0,1), coherence <0|rho|1>: {rho_1q[0, 1]}")
    print(f"  SW (1,0), coherence <1|rho|0>: {rho_1q[1, 0]}")

    print("\n[2] Pure two-qubit Bell density matrix:")
    sp.pprint(rho_2q_pure, use_unicode=True)
    print(f"Purity Tr(rho^2): {purity_2q_pure}")

    print("\n[3] Depolarized and dephased two-qubit density matrix:")
    sp.pprint(rho_2q_mixed, use_unicode=True)
    print(f"Purity Tr(rho^2): {purity_2q_mixed}")
    print(f"  NW (0,0), population <00|rho|00>: {rho_2q_mixed[0, 0]}")
    print(f"  SE (3,3), population <11|rho|11>: {rho_2q_mixed[3, 3]}")
    print(f"  NE (0,3), coherence <00|rho|11>: {rho_2q_mixed[0, 3]}")
    print(f"  SW (3,0), coherence <11|rho|00>: {rho_2q_mixed[3, 0]}")
    print(f"  Interior populations (1,1) and (2,2): {rho_2q_mixed[1, 1]}")

    substitutions = {
        p_sym: knobs.werner_visibility_p,
        gamma_d: knobs.dephasing_fraction_gamma,
        phi_sym: knobs.relative_phase_rad,
    }
    evaluated_purity = float(sp.re(purity_2q_mixed.evalf(subs=substitutions)))
    print(f"\nEvaluated two-qubit mixed-state purity: {evaluated_purity:.6f}")
    print("=" * 78 + "\n")


# =============================================================================
# QUTIP STATE CONSTRUCTION AND VALIDATION
# =============================================================================
def _validate_density_matrix(rho: qt.Qobj, name: str) -> None:
    """Check Hermiticity, unit trace, and positive semidefiniteness."""
    matrix = rho.full()
    if not np.allclose(matrix, matrix.conj().T, atol=1e-11):
        raise ValueError(f"{name} is not Hermitian.")
    if not np.isclose(np.trace(matrix), 1.0, atol=1e-11):
        raise ValueError(f"{name} does not have unit trace.")
    if np.min(np.linalg.eigvalsh(matrix)) < -1e-10:
        raise ValueError(f"{name} is not positive semidefinite.")


def build_qutip_states(knobs: SimulationControlKnobs) -> dict[str, Any]:
    """Build pure and mixed density operators with numerical diagnostics."""
    if knobs.num_qubits not in (1, 2):
        raise ValueError("num_qubits must equal 1 or 2.")
    if not 0.0 <= knobs.werner_visibility_p <= 1.0:
        raise ValueError("werner_visibility_p must lie in [0, 1].")
    if not 0.0 <= knobs.dephasing_fraction_gamma <= 1.0:
        raise ValueError("dephasing_fraction_gamma must lie in [0, 1].")
    if not 0.0 < knobs.bar_width <= 1.0:
        raise ValueError("bar_width must lie in (0, 1].")

    if knobs.num_qubits == 1:
        basis_labels = [r"$|0\rangle$", r"$|1\rangle$"]
        pure_vector = np.asarray(knobs.single_qubit_bloch_pure, dtype=float)
        mixed_vector = np.asarray(knobs.single_qubit_bloch_mixed, dtype=float)
        pure_norm = np.linalg.norm(pure_vector)
        if pure_norm < 1e-12:
            raise ValueError("The pure-state Bloch vector must be nonzero.")
        if np.linalg.norm(mixed_vector) >= 1.0:
            raise ValueError("The mixed-state Bloch vector must have norm < 1.")
        pure_vector /= pure_norm
        sigma = (qt.sigmax(), qt.sigmay(), qt.sigmaz())
        rho_pure = 0.5 * (
            qt.qeye(2)
            + pure_vector[0] * sigma[0]
            + pure_vector[1] * sigma[1]
            + pure_vector[2] * sigma[2]
        )
        rho_mixed = 0.5 * (
            qt.qeye(2)
            + mixed_vector[0] * sigma[0]
            + mixed_vector[1] * sigma[1]
            + mixed_vector[2] * sigma[2]
        )
    else:
        basis_labels = [
            r"$|00\rangle$", r"$|01\rangle$",
            r"$|10\rangle$", r"$|11\rangle$",
        ]
        ket_00 = qt.tensor(qt.basis(2, 0), qt.basis(2, 0))
        ket_11 = qt.tensor(qt.basis(2, 1), qt.basis(2, 1))
        phase = np.exp(1j * knobs.relative_phase_rad)
        ket_bell = (ket_00 + phase * ket_11).unit()
        rho_pure = qt.ket2dm(ket_bell)

        identity_2q = qt.tensor(qt.qeye(2), qt.qeye(2))
        rho_werner = (
            knobs.werner_visibility_p * rho_pure
            + (1 - knobs.werner_visibility_p) * identity_2q / 4.0
        )
        mixed_array = rho_werner.full().copy()
        mixed_array[0, 3] *= 1.0 - knobs.dephasing_fraction_gamma
        mixed_array[3, 0] *= 1.0 - knobs.dephasing_fraction_gamma
        rho_mixed = qt.Qobj(mixed_array, dims=[[2, 2], [2, 2]])

    _validate_density_matrix(rho_pure, "Pure state")
    _validate_density_matrix(rho_mixed, "Mixed state")
    metrics = {
        "rho_pure": rho_pure,
        "rho_mixed": rho_mixed,
        "basis_labels": basis_labels,
        "purity_pure": float(np.real(rho_pure.purity())),
        "purity_mixed": float(np.real(rho_mixed.purity())),
        "entropy_pure": max(0.0, float(np.real(qt.entropy_vn(rho_pure)))),
        "entropy_mixed": max(0.0, float(np.real(qt.entropy_vn(rho_mixed)))),
    }
    print("=" * 78)
    print(f"QUTIP VERIFICATION ({knobs.num_qubits}-QUBIT SYSTEM)")
    print("=" * 78)
    for state_name in ("pure", "mixed"):
        operator = metrics[f"rho_{state_name}"]
        print(
            f"{state_name.title():5s} | Trace: {operator.tr().real:.4f} | "
            f"Purity: {metrics[f'purity_{state_name}']:.4f} | "
            f"Entropy (nats): {metrics[f'entropy_{state_name}']:.4f}"
        )
    print("=" * 78 + "\n")
    return metrics


# =============================================================================
# PROJECTION-AWARE CALLOUT PLACEMENT
# =============================================================================
class ProjectedCornerAnnotation(Annotation):
    """Draw a screen-facing callout with a leader tied to a 3D data point."""

    def __init__(
        self,
        text: str,
        anchor_3d: tuple[float, float, float],
        **kwargs: Any,
    ) -> None:
        """Initialize an annotation whose anchor follows the 3D camera."""
        super().__init__(text, xy=(0.0, 0.0), xycoords="data", **kwargs)
        self.anchor_3d = anchor_3d

    def draw(self, renderer: Any) -> None:
        """Update the 2D anchor for the current 3D projection before drawing."""
        self.xy = proj3d.proj_transform(
            *self.anchor_3d,
            self.axes.get_proj(),
        )[:2]
        super().draw(renderer)


def add_collision_free_corner_callouts(
    ax: plt.Axes,
    matrix_slice: np.ndarray,
    knobs: SimulationControlKnobs,
) -> None:
    """Assign four annotations to exclusive screen-space docking regions.

    Dock assignment minimizes projected leader length across all 24 possible
    one-to-one assignments. Nonoverlapping docks and foreground drawing prevent
    collisions and bar occlusion, while 3D anchors track camera rotation.
    """
    dimension = matrix_slice.shape[0]
    inboard = knobs.corner_label_inset
    slots = [
        ((inboard, knobs.corner_label_top), "left", "top"),
        ((1.0 - inboard, knobs.corner_label_top), "right", "top"),
        ((inboard, knobs.corner_label_bottom), "left", "bottom"),
        ((1.0 - inboard, knobs.corner_label_bottom), "right", "bottom"),
    ]
    corners = [
        (0, 0, "NW"),
        (0, dimension - 1, "NE"),
        (dimension - 1, 0, "SW"),
        (dimension - 1, dimension - 1, "SE"),
    ]

    anchors = []
    for row, col, _ in corners:
        value = float(matrix_slice[row, col])
        elevation = value if abs(value) >= 1e-12 else 0.0
        xyz = (float(row), float(col), float(elevation))
        projected = proj3d.proj_transform(*xyz, ax.get_proj())[:2]
        point_pixels = ax.transData.transform(projected)
        anchor_axes = ax.transAxes.inverted().transform(point_pixels)
        anchors.append((xyz, anchor_axes))

    slot_pixels = [
        ax.transAxes.transform(position) for position, _, _ in slots
    ]
    anchor_pixels = [
        ax.transAxes.transform(position) for _, position in anchors
    ]
    cost = np.array([
        [np.linalg.norm(anchor - destination) for destination in slot_pixels]
        for anchor in anchor_pixels
    ])
    assignment = min(
        permutations(range(4)),
        key=lambda order: sum(cost[idx, position] for idx, position in enumerate(order)),
    )

    for (row, col, corner_name), (xyz, _), dock_index in zip(
        corners, anchors, assignment
    ):
        position, horizontal, vertical = slots[dock_index]
        value = float(matrix_slice[row, col])
        is_population = row == col
        line_color = "#1D4ED8" if is_population else "#B91C1C"
        if not is_population and value < -1e-12:
            line_color = "#B45309"

        label = ProjectedCornerAnnotation(
            f"{corner_name} ({row},{col})\n{value:+.3f}",
            anchor_3d=xyz,
            xytext=position,
            textcoords="axes fraction",
            ha=horizontal,
            va=vertical,
            fontsize=knobs.corner_label_fontsize,
            fontweight="semibold",
            linespacing=1.12,
            color=line_color,
            bbox={
                "boxstyle": "round,pad=0.30",
                "facecolor": "white",
                "edgecolor": line_color,
                "linewidth": 1.0,
                "alpha": 0.98,
            },
            arrowprops={
                "arrowstyle": "-",
                "color": line_color,
                "linewidth": knobs.corner_leader_linewidth,
                "shrinkA": 4,
                "shrinkB": 3,
                "connectionstyle": "arc3,rad=0.06",
            },
            zorder=100,
            annotation_clip=False,
        )
        ax.add_artist(label)


# =============================================================================
# MATPLOTLIB 3D STATE-CITY PLOTS
# =============================================================================
def render_corner_annotated_state_city(
    ax: plt.Axes,
    matrix_slice: np.ndarray,
    basis_labels: list[str],
    subplot_title: str,
    z_label: str,
    knobs: SimulationControlKnobs,
) -> None:
    """Draw a signed matrix histogram with screen-facing corner callouts."""
    dimension = matrix_slice.shape[0]
    half_width = knobs.bar_width / 2.0
    grid_x, grid_y = np.meshgrid(
        np.linspace(-0.5, dimension - 0.5, 2),
        np.linspace(-0.5, dimension - 0.5, 2),
    )
    ax.plot_surface(
        grid_x, grid_y, np.zeros_like(grid_x),
        color="#E2E8F0", alpha=0.30, shade=False,
    )

    for row_index in range(dimension):
        for col_index in range(dimension):
            value = float(matrix_slice[row_index, col_index])
            if abs(value) < 1e-12:
                continue

            diagonal_corner = (row_index, col_index) in {
                (0, 0), (dimension - 1, dimension - 1)
            }
            opposite_corner = (row_index, col_index) in {
                (0, dimension - 1), (dimension - 1, 0)
            }
            interior_population = row_index == col_index and not diagonal_corner

            if diagonal_corner:
                color, alpha = "#1D4ED8", 0.92
            elif opposite_corner:
                color, alpha = ("#DC2626" if value >= 0 else "#D97706"), 0.92
            elif interior_population:
                color, alpha = "#0D9488", 0.85
            else:
                color, alpha = "#94A3B8", 0.60

            ax.bar3d(
                row_index - half_width,
                col_index - half_width,
                min(0.0, value),
                knobs.bar_width,
                knobs.bar_width,
                abs(value),
                color=color,
                edgecolor="#0F172A",
                linewidth=0.6,
                alpha=alpha,
                shade=True,
            )

    max_amplitude = max(0.50, float(np.max(np.abs(matrix_slice))))
    ax.set_zlim(-max_amplitude - 0.20, max_amplitude + 0.26)
    ax.set_xlim(-0.5, dimension - 0.5)
    ax.set_ylim(-0.5, dimension - 0.5)
    ax.set_xticks(np.arange(dimension))
    ax.set_xticklabels(basis_labels, fontsize=8)
    ax.set_yticks(np.arange(dimension))
    ax.set_yticklabels(basis_labels, fontsize=8)
    ax.set_xlabel(r"Row basis $\langle j|$", fontsize=8.5, labelpad=5)
    ax.set_ylabel(r"Column basis $|k\rangle$", fontsize=8.5, labelpad=5)
    ax.set_zlabel(z_label, fontsize=8.5, labelpad=4)
    ax.set_title(subplot_title, fontsize=10.0, pad=12)
    ax.view_init(elev=knobs.view_elevation_deg, azim=knobs.view_azimuth_deg)
    add_collision_free_corner_callouts(ax, matrix_slice, knobs)


def plot_state_city_comparison(
    metrics: dict[str, Any],
    knobs: SimulationControlKnobs,
) -> None:
    """Show the four annotated state cities and two native QuTiP histograms."""
    pure_array = metrics["rho_pure"].full()
    mixed_array = metrics["rho_mixed"].full()
    basis_labels = metrics["basis_labels"]

    # Figure 1: manual margins reserve space for the title and legend.
    fig1 = plt.figure(figsize=knobs.figure1_size, dpi=knobs.figure_dpi)
    fig1.suptitle(
        "Pure and Mixed Quantum States | 3D Density-Matrix Corners "
        f"| {knobs.num_qubits} Qubit(s)",
        fontsize=13,
        y=0.975,
    )

    configurations = [
        (
            np.real(pure_array),
            "Pure State | Real Part\n"
            f"Purity = {metrics['purity_pure']:.3f} | "
            f"S = {metrics['entropy_pure']:.3f} nats",
            r"$\mathrm{Re}(\rho_{jk})$",
        ),
        (
            np.imag(pure_array),
            "Pure State | Imaginary Part\n"
            f"Relative phase = {knobs.relative_phase_rad:.3f} rad "
            "(2Q only)",
            r"$\mathrm{Im}(\rho_{jk})$",
        ),
        (
            np.real(mixed_array),
            "Mixed State | Real Part\n"
            f"Purity = {metrics['purity_mixed']:.3f} | "
            f"S = {metrics['entropy_mixed']:.3f} nats",
            r"$\mathrm{Re}(\rho_{jk})$",
        ),
        (
            np.imag(mixed_array),
            "Mixed State | Imaginary Part\n"
            f"Depolarization p = {knobs.werner_visibility_p:.2f}, "
            f"dephasing = {knobs.dephasing_fraction_gamma:.2f} (2Q only)",
            r"$\mathrm{Im}(\rho_{jk})$",
        ),
    ]

    for position, (matrix_slice, title, z_label) in enumerate(configurations, 1):
        ax = fig1.add_subplot(2, 2, position, projection="3d")
        render_corner_annotated_state_city(
            ax=ax,
            matrix_slice=matrix_slice,
            basis_labels=basis_labels,
            subplot_title=title,
            z_label=z_label,
            knobs=knobs,
        )

    legend_handles = [
        mpatches.Patch(
            color="#1D4ED8",
            label="NW/SE diagonal corners: basis-state populations",
        ),
        mpatches.Patch(
            color="#DC2626",
            label="NE/SW anti-diagonal corners: positive coherence",
        ),
        mpatches.Patch(
            color="#D97706",
            label="NE/SW anti-diagonal corners: negative coherence",
        ),
        mpatches.Patch(
            color="#0D9488",
            label="Interior diagonal: intermediate basis populations (2Q)",
        ),
    ]
    fig1.legend(
        handles=legend_handles,
        loc="lower center",
        ncol=2,
        fontsize=8.1,
        frameon=True,
        bbox_to_anchor=(0.5, 0.018),
    )
    fig1.subplots_adjust(
        left=0.04,
        right=0.96,
        bottom=0.14,
        top=0.90,
        wspace=knobs.figure1_wspace,
        hspace=knobs.figure1_hspace,
    )
    plt.show()

    # Figure 2: two native QuTiP histograms and one shared horizontal colorbar.
    # Independent fixed-size axes avoid automatic subplot resizing and prevent
    # the colorbar from covering 3D z-axis ticks, titles, and matrix values.
    fig2 = plt.figure(figsize=knobs.figure2_size, dpi=knobs.figure_dpi)
    fig2.suptitle(
        "QuTiP Matrix Histograms | Same Color Scale for Both States",
        fontsize=13.5,
        y=0.970,
    )

    real_pure = np.real(pure_array)
    real_mixed = np.real(mixed_array)
    global_low = min(0.0, float(real_pure.min()), float(real_mixed.min()))
    global_high = max(0.0, float(real_pure.max()), float(real_mixed.max()))
    maximum_absolute_value = max(abs(global_low), abs(global_high), 1e-8)
    symmetric_color_limit = 1.05 * maximum_absolute_value
    z_minimum = global_low * 1.08 if global_low < 0.0 else 0.0
    z_maximum = global_high * 1.08 if global_high > 0.0 else 0.1

    color_limits = [-symmetric_color_limit, symmetric_color_limit]
    norm = mpl.colors.Normalize(*color_limits)
    cmap = mpl.colormaps[knobs.histogram_colormap]

    histogram_axes = (
        knobs.histogram_left_axes,
        knobs.histogram_right_axes,
    )
    for (matrix, state_name), bounds in zip(
        ((real_pure, "pure"), (real_mixed, "mixed")),
        histogram_axes,
    ):
        ax = fig2.add_axes(bounds, projection="3d")
        qt.matrix_histogram(
            matrix,
            limits=[z_minimum, z_maximum],
            color_limits=color_limits,
            cmap=cmap,
            fig=fig2,
            ax=ax,
            colorbar=False,
        )
        ax.set_title(
            f"{state_name.title()} State | Re($\\rho$)\n"
            f"Purity = {metrics[f'purity_{state_name}']:.4f}",
            fontsize=11,
            pad=14,
        )
        ax.set_xticks(np.arange(len(basis_labels)))
        ax.set_xticklabels(basis_labels, fontsize=9)
        ax.set_yticks(np.arange(len(basis_labels)))
        ax.set_yticklabels(basis_labels, fontsize=9)
        ax.set_zlabel(r"$\mathrm{Re}(\rho_{jk})$", fontsize=9.5, labelpad=8)
        ax.tick_params(axis="z", labelsize=9, pad=5)
        ax.view_init(
            elev=knobs.view_elevation_deg,
            azim=knobs.view_azimuth_deg,
        )

    # The shared color scale is placed in its own axes, below both plots.
    colorbar_ax = fig2.add_axes(knobs.histogram_colorbar_axes)
    scalar_mappable = ScalarMappable(norm=norm, cmap=cmap)
    scalar_mappable.set_array([])
    colorbar = fig2.colorbar(
        scalar_mappable,
        cax=colorbar_ax,
        orientation="horizontal",
    )
    colorbar.set_label(r"Real density-matrix element $\mathrm{Re}(\rho_{jk})$", fontsize=10)
    colorbar.ax.tick_params(labelsize=9)
    plt.show()


def main() -> None:
    """Derive symbolic expressions and display the numerical comparisons."""
    derive_symbolic_representations(KNOBS)
    state_metrics = build_qutip_states(KNOBS)
    plot_state_city_comparison(state_metrics, KNOBS)


if __name__ == "__main__":
    main()
```

In a state city plot, a pure quantum state exhibits tall, phase-coherent off-diagonal towers whose magnitudes match the geometric mean of the corresponding diagonal population towers ($|\rho_{jk}| = \sqrt{\rho_{jj}\rho_{kk}}$ and purity $\mathrm{Tr}(\hat{\rho}^2) = 1$). Conversely, a mixed quantum state exhibits attenuated or vanishing off-diagonal towers relative to the diagonal populations ($|\rho_{jk}|$ less than $\sqrt{\rho_{jj}\rho_{kk}}$ and purity $\mathrm{Tr}(\hat{\rho}^2)$ less than $1$). Across all density matrix representations, the North-West corner $\rho_{0,0}$ and South-East corner $\rho_{d-1,d-1}$ along the main diagonal represent the real, non-negative classical occupation probabilities of the ground state $|0\dots 0\rangle$ and highest excited state $|1\dots 1\rangle$. Completing the perimeter, the North-East corner $\rho_{0,d-1}$ and South-West corner $\rho_{d-1,0}$ on the anti-diagonal represent the complex-conjugate maximal-span quantum coherences ($\rho_{d-1,0} = \rho_{0,d-1}^*$) that facilitate equatorial Bloch polarization in single-qubit systems and promote nonlocal bipartite or multipartite entanglement in Bell and GHZ states.
