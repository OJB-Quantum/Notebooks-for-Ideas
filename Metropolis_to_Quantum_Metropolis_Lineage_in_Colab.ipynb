```python
"""The graphical representation below maps the algorithmic evolution extending from the
foundational concepts jointly established by Nicholas Metropolis, Stanislaw Ulam,
and John von Neumann through the classical Metropolis Algorithm to advanced
contemporary paradigms, including Quantum Metropolis Sampling."""

import matplotlib.pyplot as plt
import graphviz
from IPython.display import display, Image

# ----------------------------------------------------------------------
# Control knobs
# ----------------------------------------------------------------------
GRAPH_RESOLUTION_DPI: str = '250'
NODE_SHAPE_STYLE: str = 'box'
NODE_FILL_STYLE: str = 'filled, rounded'
NODE_BACKGROUND_COLOR: str = '#E8F0FE'
NODE_BORDER_COLOR: str = '#1A73E8'
EDGE_LINE_COLOR: str = '#5F6368'
GLOBAL_FONT_FAMILY: str = 'Tahoma'

# ----------------------------------------------------------------------
# Global Matplotlib configuration
# ----------------------------------------------------------------------
plt.rcParams['figure.dpi'] = int(GRAPH_RESOLUTION_DPI)
plt.rcParams['font.sans-serif'] = [GLOBAL_FONT_FAMILY, 'DejaVu Sans']
plt.rcParams['font.weight'] = 'normal'


def generate_algorithmic_lineage_graph() -> graphviz.Digraph:
    """Instantiate a Graphviz directed graph representing sampling algorithms.

    Returns:
        A graphviz.Digraph object populated with historical inheritance connections.
    """
    lineage_graph = graphviz.Digraph(
        name='Metropolis_Genealogy',
        graph_attr={
            'dpi': GRAPH_RESOLUTION_DPI,
            'rankdir': 'TB',      # Top‑to‑Bottom layout
            'fontname': GLOBAL_FONT_FAMILY,
            'nodesep': '0.5',
            'ranksep': '0.8'
        },
        node_attr={
            'shape': NODE_SHAPE_STYLE,
            'style': NODE_FILL_STYLE,
            'fillcolor': NODE_BACKGROUND_COLOR,
            'color': NODE_BORDER_COLOR,
            'fontname': GLOBAL_FONT_FAMILY,
            'fontsize': '12'
        },
        edge_attr={
            'color': EDGE_LINE_COLOR,
            'fontname': GLOBAL_FONT_FAMILY,
            'fontsize': '10',
            'arrowsize': '0.7'
        }
    )

    lineage_graph.node('NM', 'Nicholas Metropolis')
    lineage_graph.node('SU', 'Stanislaw Ulam')
    lineage_graph.node('JVN', 'John von Neumann')
    lineage_graph.node('MC', 'Monte Carlo Method\n(MC)')
    lineage_graph.node('MA', 'Metropolis Algorithm\n(MA)')
    lineage_graph.node('MCMC', 'Markov Chain Monte Carlo\n(MCMC)')
    lineage_graph.node('MHA', 'Metropolis‑Hastings Algorithm\n(MHA)')
    lineage_graph.node('SA', 'Simulated Annealing\n(SA)')
    lineage_graph.node('QMC', 'Quantum Monte Carlo\n(QMC)')
    lineage_graph.node('QPE', 'Quantum Phase Estimation\n(QPE)')
    lineage_graph.node('QMS', 'Quantum Metropolis Sampling\n(QMS)')

    lineage_graph.edge('NM', 'MC', label='Jointly Invented')
    lineage_graph.edge('SU', 'MC', label='Conceptual Originator')
    lineage_graph.edge('JVN', 'MC', label='Algorithmic Developer')
    lineage_graph.edge('NM', 'MA', label='Primary Architect')
    lineage_graph.edge('MC', 'MCMC', label='Foundational Class')
    lineage_graph.edge('MA', 'MCMC', label='Core Engine')
    lineage_graph.edge('MA', 'MHA', label='Generalized By Hastings')
    lineage_graph.edge('MA', 'SA', label='Heuristic Adaptation')
    lineage_graph.edge('MCMC', 'QMC', label='Quantum System Application')
    lineage_graph.edge('MCMC', 'QMS', label='Direct Quantum Analogue')
    lineage_graph.edge('QPE', 'QMS', label='Essential Subroutine')

    return lineage_graph


def display_graph_inline() -> None:
    """Render and display the Graphviz object as an inline PNG image.

    This function processes the graph entirely in memory, which purposefully
    prevents any persistent file generation on the local filesystem.
    """
    metropolis_graph = generate_algorithmic_lineage_graph()
    png_data = metropolis_graph.pipe(format='png')
    display(Image(data=png_data))


if __name__ == '__main__':
    display_graph_inline()
```
