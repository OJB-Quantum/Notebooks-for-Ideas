# Ti/Au top contact on a 20 nm magnetic tunnel junction

During an electrode deposition step, a relatively smaller magnetic tunnel junction (MTJ) collects fewer atoms in total, while the number of atoms deposited per exposed square nanometer stays the same at a fixed calibrated rate. Start with 100 Å titanium (Ti) at 1 Å/s followed by 1,000 Å gold (Au) at 3 Å/s in the same pumpdown.

The ångström (Å) (named after Anders Ångström) equals 0.1 nm. In electron-beam evaporation, electrons heat the source and the resulting vapor supplies the metal at the wafer. Maxwell's molecular-velocity distribution gives the residual-gas impact calculation [1, 2]. Vázquez-Mena and colleagues quantify geometric shadow-mask effects [3], and Todeschini and colleagues measure how Ti changes the structure of thin Au films [4].

## Vapor access to the cap

Trace a ray from the cap toward the vapor source. At an opening $H$ above the cap, a vapor angle $\theta$ shifts the ray laterally by $H\tan\theta$. Aligning that shift with the overlay error gives the worst projected offset $d$. Full exposure requires the projected cap disk to fit inside the opening [3].

$$
d=\Delta+H\tan\theta,\qquad W_{\min}=D+2d,\qquad W\ge W_{\min}\quad\text{for full initial exposure.}
$$

For several constrictions, apply the full-exposure test at each height using the corresponding width. For a long lead opening, use the overlay and projected vapor shift toward the nearest edge. Check both source pockets since they may have different vapor angles. The plot uses the exact overlap of two circular disks for one ray direction. In that model, the exposed fraction receives the full witness flux, and the shadowed fraction receives zero direct flux.

The geometry describes the start of deposition. Metal growth on a resist lip and recessed topography between the cap and lead affect coverage later in the run. The illustrative 50 nm opening is narrower than the 110 nm combined film thickness, so an opening of that size calls for a lift-off and lead-continuity check. Use the developed opening and the height to the constricting edge, which may differ from the drawn electrode width and nominal resist thickness.

## Metal and residual-gas flux

Calibrate the QCM for each material against a planar witness at the wafer location. At calibrated rate $v$, the metal adds mass per area at rate $\rho v$. Multiplying the atomic flux by the exposed cap area gives the atom count per second, which scales with diameter squared for a fixed exposure fraction.

$$
\Gamma_{\rm metal}=\frac{\rho N_A}{M}v,\qquad t=\frac{h}{v},\qquad \dot N_{\rm cap}=fA_{\rm cap}\Gamma_{\rm metal}\propto D^2\quad\text{at fixed }f.
$$

For a thermal gas, the normal molecular velocities directed into the surface give the incident flux [1, 2]. The calculator uses a 300 K nitrogen-equivalent gas at the entered pressure to illustrate impacts. Actual uptake depends on gas partial pressures and sticking probabilities. At a fixed pressure, slowing a layer of fixed thickness increases its pressure–time exposure in inverse proportion to the deposition rate.

$$
\Gamma_i=\frac{P_i}{\sqrt{2\pi m_i k_BT_g}},\qquad P_it=\frac{P_ih}{v}\propto v^{-1}.
$$

If process measurements supply an allowable number $\eta_{\max}$ of reactive sticking events per deposited metal atom, the flux balance gives a conditional minimum wafer rate for a fully exposed surface. The required inputs are the gas partial pressures during deposition and effective sticking probabilities for the growing film.

$$
v_{\min}=\frac{M}{\rho N_A\eta_{\max}}\sum_i s_i\Gamma_i.
$$

## Electrical criterion

Titanium first forms the buried interface with the cap. Gold must form a connected overlayer and lead. Thin-film measurements show that Ti affects Au wetting and grain structure [4], while MTJ fabrication studies show how cap exposure and dielectric passivation affect top-contact formation [5].

At low read bias, a field-independent series resistance $R_s\simeq\rho_c/A_{\rm eff}+R_{\rm lead}$ increases the denominator of the measured TMR. The calculator takes the 20 nm geometric cap area as the electrical junction area for illustration. Substitute measured electrical area for $A_J$ if those areas differ.

$$
R_{\rm P,0}=\frac{RA_{\rm P}}{A_J},\qquad \frac{\mathrm{TMR}_{\rm meas}}{\mathrm{TMR}_0}=\frac{1}{1+R_s/R_{\rm P,0}},\qquad R_s\leq\frac{\epsilon}{1-\epsilon}R_{\rm P,0}.
$$

As the conducting cap and active junction scale with the same area, both interface and tunnel resistance scale as $D^{-2}$ and their ratio stays constant.

## Summary

Use 1 Å/s Ti and 3 Å/s Au for the first 20 nm fabrication split.

```python
"""This script models vapor access and compares Ti/Au deposition rates."""

# --- Control Knobs ---
CAP_DIAMETER_NM = 20.0
OPENING_DIAMETER_NM = 50.0  # Illustrative until measured.
HEIGHT_ABOVE_CAP_NM = 150.0  # Cap to limiting opening plane.
OVERLAY_ERROR_NM = 5.0  # Worst direction, aligned with the vapor shift.
INCIDENCE_DEG = 5.0  # Use the largest relevant angle for either source.

TI_THICKNESS_A = 100.0
AU_THICKNESS_A = 1000.0
TI_RATE_A_S = 1.0
AU_BASELINE_RATE_A_S = 3.0
AU_COMPARISON_RATES_A_S = (1.0, 2.0)
PRESSURE_TORR = 5e-6  # Use the pressure during deposition if measured.
GAS_TEMPERATURE_K = 300.0
RA_P_OHM_UM2 = 10.0  # Illustrative intrinsic resistance–area product.
TMR_LOSS_FRACTION = 0.01

# --- Imports and style ---
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams.update({
    "figure.dpi": 250,
    "font.family": "DejaVu Sans",
    "font.weight": "normal",
    "axes.titleweight": "normal",
})

# --- Initial cap exposure ---
cap_radius_nm = CAP_DIAMETER_NM / 2.0
cap_area_nm2 = np.pi * cap_radius_nm**2
angles_deg = np.union1d(
    np.linspace(0.0, max(12.0, INCIDENCE_DEG + 3.0), 241),
    INCIDENCE_DEG,
)
widths_nm = np.union1d(
    np.linspace(1.0, max(130.0, OPENING_DIAMETER_NM + 20.0), 260),
    OPENING_DIAMETER_NM,
)
offset_nm = OVERLAY_ERROR_NM + HEIGHT_ABOVE_CAP_NM * np.tan(
    np.deg2rad(angles_deg)
)
offset_grid, opening_radius = np.broadcast_arrays(
    offset_nm[:, None], widths_nm[None, :] / 2.0
)
exposed_fraction = np.zeros_like(offset_grid)
contained = offset_grid <= np.abs(opening_radius - cap_radius_nm)
exposed_fraction[contained] = (
    np.minimum(opening_radius[contained], cap_radius_nm) / cap_radius_nm
) ** 2

# Partial overlap is the sum of the two circular segment areas.
partial = ~contained & (offset_grid < opening_radius + cap_radius_nm)
distance = offset_grid[partial]
radius = opening_radius[partial]
cap_to_chord = (distance**2 + cap_radius_nm**2 - radius**2) / (2.0 * distance)
opening_to_chord = distance - cap_to_chord
alpha = np.arccos(np.clip(cap_to_chord / cap_radius_nm, -1.0, 1.0))
beta = np.arccos(np.clip(opening_to_chord / radius, -1.0, 1.0))
segment_area_nm2 = (
    cap_radius_nm**2 * (alpha - np.sin(alpha) * np.cos(alpha))
    + radius**2 * (beta - np.sin(beta) * np.cos(beta))
)
exposed_fraction[partial] = np.clip(segment_area_nm2 / cap_area_nm2, 0.0, 1.0)

angle_index = np.searchsorted(angles_deg, INCIDENCE_DEG)
width_index = np.searchsorted(widths_nm, OPENING_DIAMETER_NM)
example_fraction = exposed_fraction[angle_index, width_index]
full_opening_nm = CAP_DIAMETER_NM + 2.0 * offset_nm
margin_nm = (OPENING_DIAMETER_NM - CAP_DIAMETER_NM) / 2.0 - OVERLAY_ERROR_NM

print(f"Cap area: {cap_area_nm2:.1f} nm²")
print(f"Projected offset: {offset_nm[angle_index]:.2f} nm")
print(f"Initially exposed cap area: {example_fraction:.1%}")
print(f"Opening required for full exposure: {full_opening_nm[angle_index]:.2f} nm")
if margin_nm < 0.0:
    print("Full exposure requires a wider opening even at normal incidence.")
elif HEIGHT_ABOVE_CAP_NM > 0.0:
    angle_limit_deg = np.rad2deg(np.arctan(margin_nm / HEIGHT_ABOVE_CAP_NM))
    print(f"Maximum angle for full exposure: {angle_limit_deg:.2f}°")

# --- Metal and gas flux ---
avogadro_mol = 6.02214076e23
boltzmann_j_k = 1.380649e-23
pressure_pa = PRESSURE_TORR * (101325.0 / 760.0)
n2_mass_kg = 28.0134e-3 / avogadro_mol
gas_flux_nm2_s = (
    pressure_pa
    / np.sqrt(2.0 * np.pi * n2_mass_kg * boltzmann_j_k * GAS_TEMPERATURE_K)
    * 1e-18
)
ti_atoms_nm3 = 4.506e3 / 47.867e-3 * avogadro_mol * 1e-27
au_atoms_nm3 = 19.32e3 / 196.96657e-3 * avogadro_mol * 1e-27
au_rates = np.unique([*AU_COMPARISON_RATES_A_S, AU_BASELINE_RATE_A_S])
materials = [
    ("Ti", TI_THICKNESS_A, TI_RATE_A_S, ti_atoms_nm3),
    *[("Au", AU_THICKNESS_A, rate, au_atoms_nm3) for rate in au_rates],
]

print(f"\nN₂-equivalent gas impact flux: {gas_flux_nm2_s:.1f} molecules/(nm²·s)")
print("Metal  Rate (Å/s)   Time (s)   Atoms/(nm²·s)    Full-cap atoms     Dose (L)")
for metal, thickness_a, rate_a_s, density_atoms_nm3 in materials:
    duration_s = thickness_a / rate_a_s
    metal_flux_nm2_s = density_atoms_nm3 * rate_a_s / 10.0
    atom_count = metal_flux_nm2_s * duration_s * cap_area_nm2
    exposure_l = PRESSURE_TORR * duration_s * 1e6
    print(
        f"{metal:5s}  {rate_a_s:10.2f}  {duration_s:9.1f}  "
        f"{metal_flux_nm2_s:15.2f}  {atom_count:16.3g}  {exposure_l:11.1f}"
    )
ti_time_s = TI_THICKNESS_A / TI_RATE_A_S
au_baseline_time_s = AU_THICKNESS_A / AU_BASELINE_RATE_A_S
baseline_min = (ti_time_s + au_baseline_time_s) / 60.0
print(f"\nBaseline stack shutter-open time: {baseline_min:.2f} min.")

# --- Electrical contact budget ---
cap_area_um2 = cap_area_nm2 * 1e-6
parallel_resistance_ohm = RA_P_OHM_UM2 / cap_area_um2
series_budget_ohm = (
    TMR_LOSS_FRACTION / (1.0 - TMR_LOSS_FRACTION) * parallel_resistance_ohm
)
print(f"Illustrative intrinsic parallel resistance: {parallel_resistance_ohm / 1e3:.2f} kΩ")
print(
    f"Series resistance budget for {100 * TMR_LOSS_FRACTION:g}% TMR loss: "
    f"{series_budget_ohm:.1f} Ω"
)
print(
    "Full-area contact budget with zero lead resistance: "
    f"{series_budget_ohm * cap_area_um2:.3f} Ω·µm²"
)

# --- Visualization ---
fig, (geometry_ax, growth_ax) = plt.subplots(
    1, 2, figsize=(11, 4.3), layout="constrained"
)
coverage = geometry_ax.pcolormesh(
    widths_nm, angles_deg, 100.0 * exposed_fraction,
    shading="auto", cmap="cividis", vmin=0.0, vmax=100.0,
)
colorbar = fig.colorbar(coverage, ax=geometry_ax, pad=0.025)
colorbar.set_label("Cap area with direct line of sight (%)")
geometry_ax.plot(full_opening_nm, angles_deg, color="black", linewidth=1.5)
geometry_ax.scatter(
    OPENING_DIAMETER_NM, INCIDENCE_DEG, facecolor="white",
    edgecolor="black", linewidth=1.1, s=40, zorder=3,
)
geometry_ax.annotate(
    f"{example_fraction:.1%} exposed",
    (OPENING_DIAMETER_NM, INCIDENCE_DEG),
    xytext=(8, 9), textcoords="offset points",
    bbox={"facecolor": "white", "alpha": 0.9, "edgecolor": "none", "pad": 2},
)
geometry_ax.set(
    title="Initial cap exposure",
    xlabel="Effective opening diameter (nm)",
    ylabel="Vapor angle from wafer normal (°)",
    xlim=(widths_nm[0], widths_nm[-1]),
    ylim=(angles_deg[0], angles_deg[-1]),
)

au_durations_min = AU_THICKNESS_A / au_rates / 60.0
au_target_nm = AU_THICKNESS_A / 10.0
growth_ax.axhline(au_target_nm, color="0.78", linewidth=0.9, linestyle="--")
for rate_a_s, duration_min in zip(au_rates[::-1], au_durations_min[::-1]):
    is_baseline = np.isclose(rate_a_s, AU_BASELINE_RATE_A_S)
    growth_ax.plot(
        [0.0, duration_min], [0.0, au_target_nm],
        linewidth=2.5 if is_baseline else 1.65,
        label=f"{rate_a_s:g} Å/s  ·  {duration_min:.2f} min",
        markevery=[1], marker="o", markersize=4.5,
    )
growth_ax.set(
    title="Planar Au growth after Ti",
    xlabel="Time after Au shutter opens (min)",
    ylabel="Au thickness (nm)",
    xlim=(0.0, 1.07 * au_durations_min.max()),
    ylim=(0.0, 1.08 * au_target_nm),
)
growth_ax.grid(alpha=0.22)
growth_ax.legend(loc="lower right")
plt.show()
```
