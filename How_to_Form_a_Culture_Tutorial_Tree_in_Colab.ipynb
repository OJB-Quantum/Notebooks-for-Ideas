```python
!uv pip install graphviz ipython

"""This script constructs a directed culture-formation decision graph and renders it as an in-memory object.

The graph avoids orthogonal edges and maintains an absolute monochrome aesthetic. Control parameters
defined at the module level dictate the precise font, resolution, and vertical spacing characteristics.
The script applies a hard character limit to each graphical node, cascades parallel branches, and stretches
the diagram vertically. The output displays the rendered image directly in memory without HTML wrappers.
"""

import graphviz
from IPython.display import display, Image

# Control Knobs
GRAPH_DPI = "250"
NODE_FONT = "DejaVu Sans"
NODE_FONT_WEIGHT = "normal"
EDGE_ROUTING = "spline"
NODE_SHAPE = "box"
BACKGROUND_COLOR = "white"
FOREGROUND_COLOR = "black"
RANK_DIRECTION = "TB"
MAX_CHARACTERS_PER_LINE = 27
RANK_SEPARATION = "0.65"
VERTICAL_STAGGER_STEP = 1  # Multiplies the vertical drop distance across parallel branches

def wrap_text(text: str, max_width: int = MAX_CHARACTERS_PER_LINE) -> str:
    """Wraps text to a specified maximum width, hyphenating oversized strings.

    Args:
        text: The original string input.
        max_width: The maximum allowable characters per line.

    Returns:
        A formatted string featuring newline characters at proper intervals.
    """
    lines = []
    for paragraph in text.split('\n'):
        current_line = ""
        words = paragraph.split()
        for word in words:
            while len(word) > max_width:
                if current_line:
                    lines.append(current_line)
                    current_line = ""
                chunk = word[:max_width - 1] + "-"
                lines.append(chunk)
                word = word[max_width - 1:]

            if current_line:
                if len(current_line) + 1 + len(word) <= max_width:
                    current_line += " " + word
                else:
                    lines.append(current_line)
                    current_line = word
            else:
                current_line = word
        if current_line:
            lines.append(current_line)
    return '\n'.join(lines)

def build_culture_formation_tree() -> graphviz.Digraph:
    """Constructs the directed Graphviz object outlining the culture formation hierarchy."""
    tree = graphviz.Digraph(format='png')

    tree.attr(
        dpi=GRAPH_DPI,
        splines=EDGE_ROUTING,
        bgcolor=BACKGROUND_COLOR,
        rankdir=RANK_DIRECTION,
        ranksep=RANK_SEPARATION
    )
    tree.attr(
        'node',
        shape=NODE_SHAPE,
        style='solid',
        color=FOREGROUND_COLOR,
        fontcolor=FOREGROUND_COLOR,
        fontname=NODE_FONT
    )
    tree.attr(
        'edge',
        color=FOREGROUND_COLOR,
        arrowhead='normal',
        fontname=NODE_FONT,
        fontsize='10'
    )

    # Root and Primary Trunk
    tree.node('Root', wrap_text('HOW TO FORM A CULTURE'))
    tree.node('People', wrap_text('PEOPLE DOING THE SAME THING'))
    tree.edge('Root', 'People', minlen='1')

    # Branch 1: Repetition Question
    tree.node('Q_Rep', wrap_text('Does the activity happen repeatedly?'))
    tree.edge('People', 'Q_Rep', minlen=str(1 * VERTICAL_STAGGER_STEP))

    # Repetition -> No
    tree.node('Event', wrap_text('EVENT / GATHERING'))
    tree.edge('Q_Rep', 'Event', label='No', minlen='1')
    tree.node('Event_Desc', wrap_text('Shared participation exists, but continuity does not yet exist.'))
    tree.edge('Event', 'Event_Desc', minlen='1')

    # Repetition -> Yes -> Recurring Activity
    tree.node('Rec_Act', wrap_text('RECURRING ACTIVITY'))
    tree.edge('Q_Rep', 'Rec_Act', label='Yes', minlen='2')

    # Recurring Activity Sub-branch 1: Social Relationships
    tree.node('Q_Rel', wrap_text('Do stable social relationships develop?'))
    tree.edge('Rec_Act', 'Q_Rel', minlen=str(1 * VERTICAL_STAGGER_STEP))

    tree.node('Rec_Part', wrap_text('RECURRING PARTICIPATION'))
    tree.edge('Q_Rel', 'Rec_Part', label='No', minlen='1')
    tree.node(
        'Rec_Part_Desc',
        wrap_text('People repeatedly perform the same activity, but little enduring group structure develops.')
    )
    tree.edge('Rec_Part', 'Rec_Part_Desc', minlen='1')

    tree.node('Comm', wrap_text('COMMUNITY'))
    tree.edge('Q_Rel', 'Comm', label='Yes', minlen='2')

    # Recurring Activity Sub-branch 2: Shared Meanings or Practices
    tree.node('Q_Mean', wrap_text('Does the group develop shared meanings or practices?'))
    tree.edge('Rec_Act', 'Q_Mean', minlen=str(2 * VERTICAL_STAGGER_STEP))

    tree.node('Act_Comm', wrap_text('ACTIVITY-BASED COMMUNITY'))
    tree.edge('Q_Mean', 'Act_Comm', label='No', minlen='1')
    tree.node(
        'Act_Comm_Desc',
        wrap_text('Social relationships exist, but a distinctive group culture remains weak or undeveloped.')
    )
    tree.edge('Act_Comm', 'Act_Comm_Desc', minlen='1')

    tree.node('Q_Mean_Yes', wrap_text('Yes'))
    tree.edge('Q_Mean', 'Q_Mean_Yes', minlen='2')

    # Staggered Shared Meaning Indicators converging to Microculture
    shared_practices = [
        'Shared terminology?',
        'Shared norms?',
        'Shared rituals?',
        'Recognizable roles?',
        'Shared stories or memories?',
        'Shared symbols?',
        'Shared expectations?',
        'Shared humor or references?',
        'Characteristic ways of doing the activity?',
        'Informal rules about acceptable behavior?'
    ]

    tree.node('Micro', wrap_text('MICROCULTURE'))

    for idx, practice in enumerate(shared_practices):
        prac_id = f'Prac_{idx}'
        tree.node(prac_id, wrap_text(practice))
        tree.edge('Q_Mean_Yes', prac_id, minlen=str((idx + 1) * VERTICAL_STAGGER_STEP))
        tree.edge(prac_id, 'Micro', minlen='1')

    # Microculture Persistence Questions
    persistence_questions = [
        'Are these practices remembered between gatherings?',
        'Are newcomers taught or shown "how we do things here"?',
        'Do the practices persist when individual members leave?'
    ]

    tree.node('Persist_Gate', wrap_text('Persistence Evaluation'))
    for idx, question in enumerate(persistence_questions):
        pq_id = f'PQ_{idx}'
        tree.node(pq_id, wrap_text(question))
        tree.edge('Micro', pq_id, minlen=str((idx + 1) * VERTICAL_STAGGER_STEP))
        tree.edge(pq_id, 'Persist_Gate', minlen='1')

    tree.node('Temp_Micro', wrap_text('TEMPORARY MICROCULTURE'))
    tree.edge('Persist_Gate', 'Temp_Micro', label='No', minlen='1')

    tree.node('Dur_Cult', wrap_text('DURABLE CULTURE'))
    tree.edge('Persist_Gate', 'Dur_Cult', label='Yes', minlen='2')

    # Durable Culture Continuity Questions converging to Tradition
    continuity_questions = [
        'Does it acquire historical continuity?',
        'Are practices passed across generations of participants?',
        'Do recurring practices become expected?'
    ]

    tree.node('Trad', wrap_text('TRADITION'))
    for idx, question in enumerate(continuity_questions):
        cq_id = f'CQ_{idx}'
        tree.node(cq_id, wrap_text(question))
        tree.edge('Dur_Cult', cq_id, minlen=str((idx + 1) * VERTICAL_STAGGER_STEP))
        tree.edge(cq_id, 'Trad', minlen='1')

    # Tradition Identity Split
    tree.node('Q_Ident', wrap_text('Does the group develop a distinct identity relative to a broader society?'))
    tree.edge('Trad', 'Q_Ident', minlen='1')

    tree.node('Comm_Cult', wrap_text('COMMUNITY CULTURE'))
    tree.edge('Q_Ident', 'Comm_Cult', label='No', minlen='1')

    tree.node('Subcult', wrap_text('SUBCULTURE'))
    tree.edge('Q_Ident', 'Subcult', label='Yes', minlen='2')

    # Branch 2: Possible Further Expansion
    tree.node('Exp', wrap_text('Possible further expansion'))
    tree.edge('People', 'Exp', minlen=str(3 * VERTICAL_STAGGER_STEP))

    expansions = [
        ('Multiple independent groups adopt the practices', 'SCENE'),
        ('Practices spread geographically', 'DISTRIBUTED CULTURE'),
        ('Shared values motivate coordinated social change', 'MOVEMENT'),
        ('Practices become widely inherited or institutionalized', 'BROADER CULTURAL TRADITION')
    ]

    for idx, (condition, outcome) in enumerate(expansions):
        cond_id = f'Exp_Cond_{idx}'
        out_id = f'Exp_Out_{idx}'
        tree.node(cond_id, wrap_text(condition))
        tree.node(out_id, wrap_text(outcome))
        tree.edge('Exp', cond_id, minlen=str((idx + 1) * VERTICAL_STAGGER_STEP))
        tree.edge(cond_id, out_id, minlen='1')

    return tree

def display_graph(tree: graphviz.Digraph) -> None:
    """Renders the graph in memory and displays the image without HTML wrappers."""
    binary_png = tree.pipe(format='png')
    display(Image(data=binary_png))

if __name__ == '__main__':
    culture_tree = build_culture_formation_tree()
    display_graph(culture_tree)
```
