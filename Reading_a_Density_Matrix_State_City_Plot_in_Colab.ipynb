## Reading a Density-Matrix State City Plot

A **state city plot** displays a density matrix as two three-dimensional bar charts: one for its real components and one for its imaginary components. Each bar corresponds to a matrix entry $\rho_{ij}$, with the horizontal coordinates identifying its row and column and the signed bar height giving the plotted component.

The central reading rule is:

> **Diagonal entries give measurement probabilities in the plotted basis; off-diagonal entries describe quantum coherence between those basis states.**

### 1. Single-qubit probabilities occupy two opposite corners

In the computational basis $\{|0\rangle,|1\rangle\}$, a single-qubit density matrix has the form

$$
\rho =
\begin{pmatrix}
p_0 & c \\
c^* & p_1
\end{pmatrix},
\qquad
p_0+p_1=1.
$$

The diagonal entries directly give the computational-basis measurement probabilities:

$$
P(0)=\rho_{00}=p_0,
\qquad
P(1)=\rho_{11}=p_1.
$$

Because this matrix is $2\times2$, its two diagonal entries occupy opposite corners of the matrix footprint. Their heights in the **real-component plot** are probabilities between zero and one. Their heights in the **imaginary-component plot** are zero because the diagonal entries of a physical density matrix are real.

Use the row and column labels to identify these entries, since rotating the viewing angle changes their apparent positions on screen.

| Matrix entry | Physical meaning | Real-component bar height | Imaginary-component bar height |
|---|---|---|---|
| $\rho_{00}$ | Probability of measuring $|0\rangle$ | $p_0$ | $0$ |
| $\rho_{11}$ | Probability of measuring $|1\rangle$ | $p_1$ | $0$ |
| $\rho_{01}$ | Coherence between $|0\rangle$ and $|1\rangle$ | $\operatorname{Re}(c)$ | $\operatorname{Im}(c)$ |
| $\rho_{10}$ | Complex-conjugate coherence | $\operatorname{Re}(c)$ | $-\operatorname{Im}(c)$ |

### 2. The other two corners contain coherence information

The off-diagonal entries obey $\rho_{10}=\rho_{01}^*$ because a density matrix is Hermitian. Consequently, their real-component bars have equal heights, while their imaginary-component bars have opposite heights.

These signed heights are not probabilities. Together, they specify the complex coherence

$$
c=|c|e^{i\theta},
$$

where $|c|$ gives the coherence magnitude and, when $c\neq0$, $\theta$ gives its phase in the chosen basis. Physical positivity requires

$$
|c|^2\leq p_0p_1.
$$

For a pure state $|\psi\rangle=\alpha|0\rangle+\beta|1\rangle$,

$$
p_0=|\alpha|^2,
\qquad
p_1=|\beta|^2,
\qquad
c=\alpha\beta^*.
$$

Thus, the diagonal records outcome probabilities, while the off-diagonal entries retain relative-phase information.

### 3. Equal probabilities do not establish a coherent superposition

Compare the coherent state $|+\rangle=(|0\rangle+|1\rangle)/\sqrt{2}$ with an equal statistical mixture of $|0\rangle$ and $|1\rangle$:

$$
\rho_+=\frac12
\begin{pmatrix}
1&1\\
1&1
\end{pmatrix},
\qquad
\rho_{\mathrm{mixed}}=\frac12
\begin{pmatrix}
1&0\\
0&1
\end{pmatrix}.
$$

Both have diagonal heights of $1/2$, so both yield $P(0)=P(1)=1/2$. Their off-diagonal bars distinguish them: the coherent state has nonzero coherence, while the statistical mixture has none in this basis.

This difference affects measurement in another basis. An $X$-basis measurement gives $P(+)=1$ for $\rho_+$ but $P(+)=1/2$ for $\rho_{\mathrm{mixed}}$.

### 4. The probability interpretation follows the chosen measurement basis

A plotted diagonal gives probabilities for the orthonormal basis used to represent the matrix. For an arbitrary normalized reference state $|\phi\rangle$, the probability of its projective measurement outcome is

$$
P(\phi)=\langle\phi|\rho|\phi\rangle.
$$

This expression generally involves both diagonal and off-diagonal entries of the original matrix. To read another basis's probabilities directly from diagonal bars, first express the density matrix in that basis.

### 5. Larger systems require reading the entire diagonal

For a $d$-dimensional system, all $d$ diagonal entries give the corresponding basis-state probabilities:

$$
P(i)=\rho_{ii},
\qquad
\sum_{i=0}^{d-1}\rho_{ii}=1.
$$

The two diagonal corner endpoints represent only the first and last listed basis states. A qutrit also has a central diagonal entry for its third probability, while an $n$-qubit register has $2^n$ diagonal entries representing joint computational-basis outcomes.

For the qutrit example $|\psi\rangle=(|0\rangle+|2\rangle)/\sqrt{2}$,

$$
\rho=\frac12
\begin{pmatrix}
1&0&1\\
0&0&0\\
1&0&1
\end{pmatrix}.
$$

The diagonal gives $P(0)=1/2$, $P(1)=0$, and $P(2)=1/2$. The other two outer corners, $\rho_{02}$ and $\rho_{20}$, represent coherence between $|0\rangle$ and $|2\rangle$.

**Practical reading rule:** identify the plotted basis, read the entire diagonal for probabilities, and inspect the off-diagonal real and imaginary bars for coherence and phase information.

```python
"""Illustrate density-matrix probabilities, coherence, phase, and basis choice.

Run in Google Colab with %run density_matrix_city_plots.py or paste this entire
file into one Python cell. Missing dependencies install automatically.
An equivalent optional installation cell is:
    !uv pip install qutip==5.3.1 matplotlib numpy

Eighteen examples cover basis states, coherent versus mixed states, real and
imaginary coherence, unequal populations, partial dephasing, measurement-basis
changes, qutrits, two-qubit states, and seeded random states. Each example has
two city plots and numerical reading aids; a table compares all examples.

Figures render in memory at 250 DPI without tight_layout. Displayed figures
close unless KEEP_FIGURES_OPEN is enabled. Calling
generate_state_city_plots(show=False) returns open figures for further use.

References:
    https://learning.quantum.ibm.com/course/general-formulation-of-quantum-information/density-matrices
    https://learning.quantum.ibm.com/course/general-formulation-of-quantum-information/general-measurements
    https://qutip.readthedocs.io/en/qutip-5.3.x/apidoc/visualization.html
"""

import importlib.util
import math
import shutil
import subprocess
import sys

# - Example selection and physical controls -
RANDOM_SEED = 42
EXAMPLE_IDS = None  # None selects all; otherwise use IDs from the summary.
EXAMPLE_GROUPS = None  # E.g. ("coherence", "basis_change", "larger_systems").
POPULATION_ZERO = 0.80
QUBIT_RELATIVE_PHASE = math.pi / 3
DEPHASING_COHERENCE_FRACTION = 0.40  # 0: fully dephased; 1: fully coherent.
QUTRIT_POPULATIONS = (0.20, 0.30, 0.50)
QUTRIT_PHASES = (0.0, math.pi / 2, math.pi)
AUTO_INSTALL = True
SHOW_SUMMARY = True
KEEP_FIGURES_OPEN = False
VALIDATION_TOLERANCE = 1e-10

# - Figure controls -
FIGURE_DPI = 250
BASE_FONT_SIZE = 10
ELEVATION_ANGLE = 35
AZIMUTH_ANGLE = -45
FIGURE_WIDTH = 12
FIGURE_HEIGHT = 6.5
BOX_ASPECT = (1.0, 1.0, 0.85)
AXES_ZOOM = 0.90
COMPONENT_LIMITS = (-1.0, 1.0)
SHOW_COLORBARS = True
COLORBAR_PADDING = 0.16
COLORMAP = "RdBu_r"
LEFT_MARGIN = 0.02
RIGHT_MARGIN = 0.94
BOTTOM_MARGIN = 0.25
TOP_MARGIN = 0.81
SUBPLOT_SPACING = 0.28


def ensure_dependencies():
    """Install missing dependencies into the active Python interpreter."""
    requirements = {
        "numpy": "numpy",
        "matplotlib": "matplotlib>=3.6",
        "qutip": "qutip==5.3.1",
    }
    missing = [
        requirement
        for module, requirement in requirements.items()
        if importlib.util.find_spec(module) is None
    ]
    if not missing:
        return
    if not AUTO_INSTALL:
        raise ModuleNotFoundError(
            "Missing dependencies: " + ", ".join(missing)
        )
    uv_path = shutil.which("uv")
    if uv_path is None:
        subprocess.check_call(
            [sys.executable, "-m", "pip", "install", "--quiet", "uv"]
        )
        uv_command = [sys.executable, "-m", "uv"]
    else:
        uv_command = [uv_path]
    subprocess.check_call(
        uv_command
        + ["pip", "install", "--quiet", "--python", sys.executable]
        + missing
    )
    importlib.invalidate_caches()


def build_examples():
    """Construct physical states and their orthonormal plotting bases.

    Returns:
        A list of dictionaries containing physical density matrices,
        displayed representations, basis labels, and explanatory captions.
    """
    import numpy as np
    import qutip as qt

    if not 0 <= POPULATION_ZERO <= 1:
        raise ValueError("POPULATION_ZERO must be between zero and one.")
    if not 0 <= DEPHASING_COHERENCE_FRACTION <= 1:
        raise ValueError("DEPHASING_COHERENCE_FRACTION must lie in [0, 1].")
    populations = np.asarray(QUTRIT_POPULATIONS, dtype=float)
    phases = np.asarray(QUTRIT_PHASES, dtype=float)
    if populations.shape != (3,) or phases.shape != (3,):
        raise ValueError(
            "Qutrit populations and phases require three entries."
        )
    if (
        not np.all(np.isfinite(populations))
        or not np.all(np.isfinite(phases))
        or np.any(populations < 0)
        or not np.isclose(populations.sum(), 1, atol=1e-12, rtol=0)
    ):
        raise ValueError(
            "Qutrit populations must be finite, nonnegative, and sum to one."
        )

    zero, one = qt.basis(2, 0), qt.basis(2, 1)
    plus = (zero + one).unit()
    minus = (zero - one).unit()
    plus_i = (zero + 1j * one).unit()
    minus_i = (zero - 1j * one).unit()
    mixed = qt.qeye(2) / 2
    rho_plus = qt.ket2dm(plus)
    rho_plus_i = qt.ket2dm(plus_i)
    unequal = (
        np.sqrt(POPULATION_ZERO) * zero
        + np.exp(1j * QUBIT_RELATIVE_PHASE)
        * np.sqrt(1 - POPULATION_ZERO)
        * one
    )
    x_basis = np.column_stack((plus.full().ravel(), minus.full().ravel()))
    y_basis = np.column_stack((plus_i.full().ravel(), minus_i.full().ravel()))
    examples = []

    def add(identifier, group, title, rho, labels, note, basis=None):
        """Register a state using orthonormal basis vectors as columns."""
        physical = rho.full()
        unitary = np.eye(len(labels)) if basis is None else basis
        if not np.allclose(
            unitary.conj().T @ unitary,
            np.eye(len(labels)),
            atol=VALIDATION_TOLERANCE,
            rtol=0,
        ):
            raise ValueError("The plotting basis must be orthonormal.")
        displayed = unitary.conj().T @ physical @ unitary
        examples.append(
            {
                "id": identifier,
                "group": group,
                "title": title,
                "physical": physical,
                "displayed": displayed,
                "labels": labels,
                "note": note,
            }
        )

    add(
        "zero",
        "basis_states",
        r"Qubit $|0\rangle$",
        qt.ket2dm(zero),
        ("0", "1"),
        "The (0, 0) diagonal corner carries probability one.",
    )
    add(
        "one",
        "basis_states",
        r"Qubit $|1\rangle$",
        qt.ket2dm(one),
        ("0", "1"),
        "The opposite (1, 1) diagonal corner carries probability one.",
    )
    add(
        "plus",
        "coherence",
        r"Coherent qubit $|+\rangle$",
        rho_plus,
        ("0", "1"),
        "Equal diagonal populations accompany positive real coherence.",
    )
    add(
        "mixed",
        "coherence",
        "Equal statistical mixture",
        mixed,
        ("0", "1"),
        "The same populations as |+> accompany zero coherence in this basis.",
    )
    add(
        "minus",
        "coherence",
        r"Coherent qubit $|-\rangle$",
        qt.ket2dm(minus),
        ("0", "1"),
        "A relative phase of pi reverses the real off-diagonal signs.",
    )
    add(
        "plus_i",
        "coherence",
        r"Coherent qubit $|+i\rangle$",
        rho_plus_i,
        ("0", "1"),
        "Imaginary coherence: Im(rho_01) = -1/2 and Im(rho_10) = +1/2.",
    )
    add(
        "minus_i",
        "coherence",
        r"Coherent qubit $|-i\rangle$",
        qt.ket2dm(minus_i),
        ("0", "1"),
        "The opposite relative phase reverses the imaginary bar signs.",
    )
    add(
        "unequal_pure",
        "populations",
        "Unequal-population coherent qubit",
        qt.ket2dm(unequal),
        ("0", "1"),
        "Diagonal entries give populations; complex coherences retain phase.",
    )
    add(
        "unequal_mixed",
        "populations",
        "Unequal statistical mixture",
        qt.Qobj(np.diag([POPULATION_ZERO, 1 - POPULATION_ZERO])),
        ("0", "1"),
        "The same populations as the preceding pure state have zero coherence.",
    )
    add(
        "partial_dephasing",
        "dephasing",
        "Partially dephased qubit",
        DEPHASING_COHERENCE_FRACTION * rho_plus
        + (1 - DEPHASING_COHERENCE_FRACTION) * mixed,
        ("0", "1"),
        "Z-basis dephasing reduces coherence while preserving populations.",
    )
    add(
        "plus_in_x",
        "basis_change",
        r"The same $|+\rangle$ state in the X basis",
        rho_plus,
        ("+", "-"),
        "A basis change gives P(+) = 1; the physical state is unchanged.",
        basis=x_basis,
    )
    add(
        "plus_i_in_y",
        "basis_change",
        r"The same $|+i\rangle$ state in the Y basis",
        rho_plus_i,
        ("+i", "-i"),
        "A basis change gives P(+i) = 1; coherence depends on the plotted basis.",
        basis=y_basis,
    )
    qutrit_outer = (qt.basis(3, 0) + qt.basis(3, 2)).unit()
    qutrit_three = qt.Qobj(np.sqrt(populations) * np.exp(1j * phases))
    add(
        "qutrit_outer",
        "larger_systems",
        "Qutrit superposition of |0> and |2>",
        qt.ket2dm(qutrit_outer),
        ("0", "1", "2"),
        "Two corners give probabilities; the other two give 0-to-2 coherence.",
    )
    add(
        "qutrit_three",
        "larger_systems",
        "Three-level qutrit with relative phases",
        qt.ket2dm(qutrit_three),
        ("0", "1", "2"),
        "Read all three diagonal entries, including the interior |1> population.",
    )
    ket_00, ket_11 = qt.tensor(zero, zero), qt.tensor(one, one)
    bell = qt.ket2dm((ket_00 + ket_11).unit())
    correlated_mixture = (qt.ket2dm(ket_00) + qt.ket2dm(ket_11)) / 2
    add(
        "bell",
        "larger_systems",
        "Two-qubit Bell state",
        bell,
        ("00", "01", "10", "11"),
        "Four diagonal entries give joint probabilities; rho_00,11 is coherence.",
    )
    add(
        "bell_mixed",
        "larger_systems",
        "Two-qubit correlated statistical mixture",
        correlated_mixture,
        ("00", "01", "10", "11"),
        "The same joint populations as the Bell state have zero coherence.",
    )
    rng = np.random.default_rng(RANDOM_SEED)
    add(
        "random_qubit",
        "random",
        "Seeded random qubit",
        qt.rand_dm(2, density=1.0, seed=rng),
        ("0", "1"),
        "Check conjugate coherence pairs and normalized nonnegative populations.",
    )
    add(
        "random_qutrit",
        "random",
        "Seeded random qutrit",
        qt.rand_dm(3, density=1.0, seed=rng),
        ("0", "1", "2"),
        "A general mixed qutrit requires the full diagonal and coherence pairs.",
    )
    identifiers = {example["id"] for example in examples}
    groups = {example["group"] for example in examples}
    if EXAMPLE_IDS is not None and set(EXAMPLE_IDS) - identifiers:
        raise ValueError(
            "Unknown EXAMPLE_IDS: " + str(set(EXAMPLE_IDS) - identifiers)
        )
    if EXAMPLE_GROUPS is not None and set(EXAMPLE_GROUPS) - groups:
        raise ValueError(
            "Unknown EXAMPLE_GROUPS: " + str(set(EXAMPLE_GROUPS) - groups)
        )
    return [
        example
        for example in examples
        if (EXAMPLE_IDS is None or example["id"] in EXAMPLE_IDS)
        and (EXAMPLE_GROUPS is None or example["group"] in EXAMPLE_GROUPS)
    ]


def analyze_example(example):
    """Validate a state and calculate measurement and coherence statistics.

    Args:
        example: A dictionary returned by build_examples.

    Returns:
        A dictionary of populations, purity, l1 coherence, and single-qubit
        probabilities in the physical Z, X, and Y measurement bases.
    """
    import numpy as np

    matrix = example["displayed"]
    tolerance = VALIDATION_TOLERANCE
    if not np.all(np.isfinite(matrix)):
        raise ValueError("Density-matrix entries must be finite.")
    if not np.allclose(matrix, matrix.conj().T, atol=tolerance, rtol=0):
        raise ValueError("A density matrix must be Hermitian.")
    if not np.isclose(np.trace(matrix), 1, atol=tolerance, rtol=0):
        raise ValueError("A density matrix must have unit trace.")
    if np.linalg.eigvalsh(matrix).min() < -tolerance:
        raise ValueError("A density matrix must be positive semidefinite.")
    diagonal = np.diag(matrix).real
    purity = float(np.trace(matrix @ matrix).real)
    coherence = float(np.abs(matrix - np.diag(diagonal)).sum())
    measurements = None
    if matrix.shape == (2, 2):
        physical = example["physical"]
        c = physical[0, 1]
        measurements = {
            "Z": (float(physical[0, 0].real), float(physical[1, 1].real)),
            "X": (float(0.5 + c.real), float(0.5 - c.real)),
            "Y": (float(0.5 - c.imag), float(0.5 + c.imag)),
        }
    return {
        "populations": diagonal,
        "purity": purity,
        "coherence": coherence,
        "measurements": measurements,
    }


def display_summary(examples, statistics):
    """Display a Markdown comparison table in Colab or print it in a terminal."""
    lines = [
        "### Density-matrix reading guide",
        "",
        "Diagonal bars give probabilities in the **plotted basis**. "
        "Off-diagonal bars give signed coherence components, not probabilities. "
        "Imaginary diagonal entries vanish for a physical density matrix.",
        "",
        r"Purity is $\mathrm{Tr}(\rho^2)$; it equals one for a pure state. "
        r"The basis-dependent coherence is "
        r"$C_{l_1}=\sum_{i\ne j}|\rho_{ij}|$. "
        "A basis change preserves purity while changing populations and coherence.",
        "",
        "The last three columns give single-qubit probabilities in the original "
        "physical Z, X, and Y bases, even when the plot uses another basis. "
        "Two-qubit diagonal entries describe joint outcomes.",
        "",
        "| Example ID | Plotted basis order | Plotted populations | Purity | "
        "l1 coherence | P(0), Z | P(+), X | P(+i), Y |",
        "| --- | --- | --- | --- | --- | --- | --- | --- |",
    ]
    for example, stats in zip(examples, statistics):
        labels = ", ".join(example["labels"])
        probabilities = ", ".join(
            f"{value:.3f}" for value in stats["populations"]
        )
        measurements = stats["measurements"]
        measured = (
            [f"{measurements[basis][0]:.3f}" for basis in ("Z", "X", "Y")]
            if measurements is not None
            else ["N/A"] * 3
        )
        lines.append(
            f"| {example['id']} | {labels} | {probabilities} | "
            f"{stats['purity']:.3f} | {stats['coherence']:.3f} | "
            + " | ".join(measured)
            + " |"
        )
    text = "\n".join(lines)
    try:
        from IPython import get_ipython
        from IPython.display import Markdown, display
    except ImportError:
        print(text)
    else:
        if get_ipython() is None:
            print(text)
        else:
            display(Markdown(text))


def generate_state_city_plots(show=True):
    """Render selected examples and their numerical interpretation.

    Args:
        show: Whether to display figures and the optional comparison table.

    Returns:
        A list of Matplotlib figures. With show=False they remain open;
        with show=True they close unless KEEP_FIGURES_OPEN is enabled.
    """
    ensure_dependencies()
    import matplotlib.pyplot as plt
    import numpy as np
    import qutip as qt

    examples = build_examples()
    if not examples:
        raise ValueError("Example selection is empty.")
    statistics = [analyze_example(example) for example in examples]
    style = {
        "figure.dpi": FIGURE_DPI,
        "savefig.dpi": FIGURE_DPI,
        "figure.autolayout": False,
        "figure.constrained_layout.use": False,
        "font.family": "sans-serif",
        "font.sans-serif": ["DejaVu Sans"],
        "font.size": BASE_FONT_SIZE,
        "font.weight": "normal",
    }
    figures = []
    with plt.rc_context(style):
        for example, stats in zip(examples, statistics):
            matrix = example["displayed"]
            labels = example["labels"]
            basis_labels = [rf"$|{label}\rangle$" for label in labels]
            fig, axes = plt.subplots(
                1,
                2,
                figsize=(FIGURE_WIDTH, FIGURE_HEIGHT),
                subplot_kw={"projection": "3d"},
                layout="none",
            )
            fig.subplots_adjust(
                left=LEFT_MARGIN,
                right=RIGHT_MARGIN,
                bottom=BOTTOM_MARGIN,
                top=TOP_MARGIN,
                wspace=SUBPLOT_SPACING,
            )
            components = [
                (matrix.real, "Real component", r"$\mathrm{Re}(\rho_{ij})$"),
                (
                    matrix.imag,
                    "Imaginary component",
                    r"$\mathrm{Im}(\rho_{ij})$",
                ),
            ]
            for ax, (data, title, z_label) in zip(axes, components):
                qt.matrix_histogram(
                    data,
                    x_basis=basis_labels,
                    y_basis=basis_labels,
                    limits=list(COMPONENT_LIMITS),
                    color_limits=list(COMPONENT_LIMITS),
                    options={
                        "elev": ELEVATION_ANGLE,
                        "azim": AZIMUTH_ANGLE,
                        "cbar_pad": COLORBAR_PADDING,
                    },
                    cmap=plt.get_cmap(COLORMAP),
                    colorbar=SHOW_COLORBARS,
                    fig=fig,
                    ax=ax,
                )
                if SHOW_COLORBARS:
                    fig.axes[-1].set_ylabel("")
                ax.view_init(elev=ELEVATION_ANGLE, azim=AZIMUTH_ANGLE)
                ax.set_box_aspect(BOX_ASPECT, zoom=AXES_ZOOM)
                ax.set_title(title, pad=14)
                # QuTiP maps matrix rows to x and matrix columns to y.
                axis_label_pad = 12 if len(labels) > 2 else 6
                ax.set_xlabel("Row i", labelpad=axis_label_pad)
                ax.set_ylabel("Column j", labelpad=axis_label_pad)
                if len(labels) > 2:
                    for axis_name in ("x", "y"):
                        ax.tick_params(
                            axis=axis_name,
                            labelsize=BASE_FONT_SIZE - 2,
                            pad=0,
                        )
                ax.set_zlabel(z_label, labelpad=7)
                ax.set_zticks(np.linspace(*COMPONENT_LIMITS, 5))
            fig.suptitle(
                example["title"],
                y=0.965,
                fontsize=BASE_FONT_SIZE + 3,
                fontweight="bold",
            )
            fig.text(0.5, 0.905, example["note"], ha="center", va="center")
            probabilities = ";  ".join(
                f"P({label}) = {value:.3f}"
                for label, value in zip(labels, stats["populations"])
            )
            fig.text(
                0.5,
                0.16,
                "Plotted-basis probabilities: " + probabilities,
                ha="center",
                va="center",
            )
            fig.text(
                0.5,
                0.115,
                f"Purity = {stats['purity']:.3f}    |    "
                f"l1 coherence in plotted basis = {stats['coherence']:.3f}",
                ha="center",
                va="center",
            )
            measurements = stats["measurements"]
            if measurements is not None:
                outcomes = {"Z": "0,1", "X": "+,-", "Y": "+i,-i"}
                measurement_text = "    ".join(
                    f"{basis} ({outcomes[basis]}): "
                    f"({values[0]:.3f}, {values[1]:.3f})"
                    for basis, values in measurements.items()
                )
                footer = "Measurement probabilities: " + measurement_text
            else:
                footer = (
                    "All diagonal entries count; diagonal corners give only "
                    "the first and last outcomes."
                )
            fig.text(0.5, 0.07, footer, ha="center", va="center")
            figures.append(fig)
            if show:
                plt.show()
                if not KEEP_FIGURES_OPEN:
                    plt.close(fig)
    if show and SHOW_SUMMARY:
        display_summary(examples, statistics)
    return figures


if __name__ == "__main__":
    generate_state_city_plots()
```
